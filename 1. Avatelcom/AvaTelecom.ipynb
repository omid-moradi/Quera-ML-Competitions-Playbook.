```python
import pandas as pd

train = pd.read_csv(r'F:\Quera\olympic 2024\customer\Data\train.csv')
test = pd.read_csv(r'F:\Quera\olympic 2024\customer\Data\test.csv')
train
```

```python
import warnings
warnings.filterwarnings('ignore')

print("=" * 60)
print("TRAIN DATASET INFO")
print("=" * 60)
print(f"Shape: {train.shape}")
print(f"\nColumn Names:\n{train.columns.tolist()}")
print(f"\nData Types:\n{train.dtypes}")
print(f"\nFirst 5 rows:\n{train.head()}")
print(f"\nBasic Statistics:\n{train.describe()}")
print(f"\nMissing Values:\n{train.isnull().sum()}")
print(f"\nTarget Distribution:\n{train['custExit'].value_counts()}")

print("\n" + "=" * 60)
print("TEST DATASET INFO")
print("=" * 60)
print(f"Shape: {test.shape}")
print(f"\nMissing Values:\n{test.isnull().sum()}")
```

```python
print("=" * 80)
print("COMPLETE UNIQUE VALUES ANALYSIS FOR ALL OBJECT COLUMNS")
print("=" * 80)

# لیست ستون‌های object (به جز custId و customerFeedback که متنی هستند)
object_columns = train.select_dtypes(include='object').columns.tolist()
object_columns.remove('custId')
object_columns.remove('customerFeedback')

# بررسی کامل هر ستون
for col in object_columns:
    unique_vals = train[col].unique()
    print(f"\n{'='*60}")
    print(f"Column: {col}")
    print(f"{'='*60}")
    print(f"Total Unique Values: {len(unique_vals)}")
    print(f"All Unique Values: {sorted(unique_vals)}")
    print(f"\nValue Counts:")
    print(train[col].value_counts())
    
print("\n" + "=" * 80)
print("CHECKING FOR MIXED DATA TYPES IN NUMERIC-LIKE COLUMNS")
print("=" * 80)

# ستون‌هایی که باید عددی باشند ولی object هستند
numeric_like_cols = ['membershipDuration', 'monthlyDataUsageGb', 'cumulativeSpend', 
                     'billingIssues', 'supportTickets', 'lastContactRating', 
                     'mobileAppSatisfaction', 'networkStabilityScore', 'dataLimitWarnings']

for col in numeric_like_cols:
    print(f"\n{col}:")
    unique_vals = train[col].unique()
    
    # تفکیک مقادیر عددی و غیرعددی
    numeric_vals = []
    non_numeric_vals = []
    
    for val in unique_vals:
        try:
            float(val)
            numeric_vals.append(val)
        except (ValueError, TypeError):
            non_numeric_vals.append(val)
    
    print(f"  Total unique: {len(unique_vals)}")
    print(f"  Numeric-like: {len(numeric_vals)}")
    print(f"  Non-numeric: {len(non_numeric_vals)}")


    if non_numeric_vals:
        print(f"  Non-numeric values: {sorted(non_numeric_vals)}")
        print(f"  Non-numeric counts:")
        for val in non_numeric_vals:
            count = (train[col] == val).sum()
            print(f"    '{val}': {count}")

print("\n" + "=" * 80)
print("TARGET VARIABLE ANALYSIS")
print("=" * 80)
print(f"custExit unique values: {train['custExit'].unique()}")
print(f"\nValue counts:\n{train['custExit'].value_counts()}")
print(f"\nChurn Rate: {(train['custExit']=='Yes').sum() / len(train) * 100:.2f}%")
```

```python
import numpy as np

print("="*80)
print("DATA CLEANING: BINARY & NUMERIC COLUMNS")
print("="*80)

# گروه A: ستون‌های باینری معتبر - تبدیل Yes=1, No=0
binary_cols = {
    'billingIssues': {'Yes': 1, 'No': 0},
    'supportTickets': {'Yes': 1, 'No': 0},
    'dataLimitWarnings': {'Yes': 1, 'No': 0}
}

print("\nGroup A: Converting valid Yes/No to 1/0")
for col, mapping in binary_cols.items():
    # ذخیره مقادیر عددی
    train_numeric_mask = pd.to_numeric(train[col], errors='coerce').notna()
    test_numeric_mask = pd.to_numeric(test[col], errors='coerce').notna()
    
    train_numeric_vals = pd.to_numeric(train[col], errors='coerce')[train_numeric_mask]
    test_numeric_vals = pd.to_numeric(test[col], errors='coerce')[test_numeric_mask]
    
    # تبدیل Yes/No
    train[col] = train[col].replace(mapping)
    test[col] = test[col].replace(mapping)
    
    # برگرداندن مقادیر عددی
    train.loc[train_numeric_mask, col] = train_numeric_vals
    test.loc[test_numeric_mask, col] = test_numeric_vals
    
    # تبدیل به numeric
    train[col] = pd.to_numeric(train[col], errors='coerce')
    test[col] = pd.to_numeric(test[col], errors='coerce')
    
    print(f"  {col}: converted (missing={train[col].isnull().sum()})")

# گروه B: ستون‌های عددی - جایگزینی Yes با 1 و حذف سایر موارد
numeric_with_yes = ['lastContactRating', 'mobileAppSatisfaction', 'networkStabilityScore']

print("\nGroup B: Replacing Yes with 1 in rating columns")
for col in numeric_with_yes:
    train[col] = train[col].replace('Yes', 1)
    test[col] = test[col].replace('Yes', 1)
    
    train[col] = pd.to_numeric(train[col], errors='coerce')
    test[col] = pd.to_numeric(test[col], errors='coerce')
    
    print(f"  {col}: Yes→1 (missing={train[col].isnull().sum()})")

# گروه C: ستون‌های عددی خالص - حذف Yes/No
pure_numeric = ['membershipDuration', 'monthlyDataUsageGb', 'cumulativeSpend']

print("\nGroup C: Cleaning pure numeric columns")
for col in pure_numeric:
    train[col] = train[col].astype(str).str.strip()
    test[col] = test[col].astype(str).str.strip()
    
    train[col] = train[col].replace(['Yes', 'No', 'yes', 'no', '', ' '], np.nan)
    test[col] = test[col].replace(['Yes', 'No', 'yes', 'no', '', ' '], np.nan)
    
    train[col] = pd.to_numeric(train[col], errors='coerce')
    test[col] = pd.to_numeric(test[col], errors='coerce')
    
    print(f"  {col}: cleaned (missing={train[col].isnull().sum()})")

# پر کردن missing values با median
print("\n" + "="*80)
print("IMPUTING MISSING VALUES")
print("="*80)

all_cols = list(binary_cols.keys()) + numeric_with_yes + pure_numeric

for col in all_cols:
    median_val = train[col].median()
    train[col].fillna(median_val, inplace=True)
    test[col].fillna(median_val, inplace=True)
    print(f"{col}: filled with {median_val:.2f}")

print("\n✅ Cleaning completed!")
print(f"Train missing: {train.isnull().sum().sum()}")
print(f"Test missing: {test.isnull().sum().sum()}")
```

```python
# ============================================================================
# STEP 1: NLP - TEXT FEATURE EXTRACTION FROM customerFeedback
# ============================================================================
print("="*80)
print("STEP 1: TEXT FEATURE EXTRACTION WITH NLP")
print("="*80)

from textblob import TextBlob
import re

def extract_text_features(text):
    """استخراج ویژگی‌های متنی"""
    if pd.isna(text) or text == '':
        return {
            'text_length': 0,
            'word_count': 0,
            'sentiment_polarity': 0,
            'sentiment_subjectivity': 0,
            'negative_words': 0,
            'positive_words': 0
        }
    
    # تمیز کردن متن
    text_clean = re.sub(r'[^a-zA-Z\s]', '', str(text).lower())
    
    # طول متن
    text_length = len(str(text))
    
    # تعداد کلمات
    words = text_clean.split()
    word_count = len(words)
    
    # Sentiment Analysis با TextBlob
    blob = TextBlob(str(text))
    sentiment_polarity = blob.sentiment.polarity  # -1 to 1
    sentiment_subjectivity = blob.sentiment.subjectivity  # 0 to 1
    
    # کلمات منفی/مثبت
    negative_keywords = ['problem', 'issue', 'bad', 'poor', 'terrible', 'worst', 
                        'cancel', 'disappointed', 'frustrated', 'slow', 'expensive']
    positive_keywords = ['great', 'good', 'excellent', 'best', 'happy', 
                        'satisfied', 'love', 'amazing', 'perfect', 'fast']
    
    negative_words = sum(1 for word in words if word in negative_keywords)
    positive_words = sum(1 for word in words if word in positive_keywords)
    
    return {
        'text_length': text_length,
        'word_count': word_count,
        'sentiment_polarity': sentiment_polarity,
        'sentiment_subjectivity': sentiment_subjectivity,
        'negative_words': negative_words,
        'positive_words': positive_words
    }

# اعمال بر روی train و test
print("\nExtracting features from customerFeedback...")
train_text_features = train['customerFeedback'].apply(extract_text_features).apply(pd.Series)
test_text_features = test['customerFeedback'].apply(extract_text_features).apply(pd.Series)

# اضافه کردن به dataframe
for col in train_text_features.columns:
    train[col] = train_text_features[col]
    test[col] = test_text_features[col]

print(f"✅ Text features created: {list(train_text_features.columns)}")
print(f"Sample sentiment scores:\n{train[['sentiment_polarity', 'sentiment_subjectivity']].describe()}")

# ============================================================================
# STEP 2: ENCODING CATEGORICAL VARIABLES (FIXED VERSION)
# ============================================================================
print("\n" + "="*80)
print("STEP 2: ENCODING CATEGORICAL VARIABLES")
print("="*80)

from sklearn.preprocessing import LabelEncoder

# Binary categorical columns - Label Encoding
binary_cats = ['sex', 'isElderly', 'partner', 'dependents', 'phoneService', 
               'paperlessBilling', 'competitorOffers']

label_encoders = {}
for col in binary_cats:
    le = LabelEncoder()
    train[col] = le.fit_transform(train[col])
    test[col] = le.transform(test[col])
    label_encoders[col] = le
    print(f"  {col}: {le.classes_} → {le.transform(le.classes_)}")

# Multi-class categorical - One-Hot Encoding
multi_cats = ['agreementTerm', 'acquisitionChannel', 'internetService', 
              'transactionMethod', 'multipleLines', 'cyberProtectionService',
              'onlineBackup', 'deviceProtection', 'techSupport', 
              'streamingTv', 'streamingMovies']

print(f"\n✅ One-Hot Encoding for: {multi_cats}")

# اعمال get_dummies بر روی train
train = pd.get_dummies(train, columns=multi_cats, drop_first=True)

# اعمال get_dummies بر روی test
test = pd.get_dummies(test, columns=multi_cats, drop_first=True)

# ✅ FIX: هماهنگ کردن ستون‌های test با train
missing_cols = set(train.columns) - set(test.columns) - {'custExit'}
for col in missing_cols:
    test[col] = 0

# حذف ستون‌های اضافی در test که در train نیست
extra_cols = set(test.columns) - set(train.columns)
test = test.drop(columns=list(extra_cols))

# مرتب کردن ستون‌های test مطابق train
test = test[train.drop('custExit', axis=1).columns]

print(f"Train shape after encoding: {train.shape}")
print(f"Test shape after encoding: {test.shape}")
print(f"✅ Columns aligned: {train.shape[1] - 1 == test.shape[1]}")  # -1 برای custExit

# ============================================================================
# STEP 3: FEATURE ENGINEERING - CREATE NEW FEATURES (FIXED)
# ============================================================================
print("\n" + "="*80)
print("STEP 3: FEATURE ENGINEERING - NEW FEATURES")
print("="*80)

# Feature 1: tenure per spend ratio
train['tenure_per_spend'] = train['membershipDuration'] / (train['cumulativeSpend'] + 1)
test['tenure_per_spend'] = test['membershipDuration'] / (test['cumulativeSpend'] + 1)
print("✅ Created: tenure_per_spend")

# Feature 2: monthly average spend
train['monthly_avg_spend'] = train['cumulativeSpend'] / (train['membershipDuration'] + 1)
test['monthly_avg_spend'] = test['cumulativeSpend'] / (test['membershipDuration'] + 1)
print("✅ Created: monthly_avg_spend")

# Feature 3: service count (FIXED)
# استفاده از ستون‌های مشترک بین train و test
streaming_cols = [c for c in train.columns if 'streaming' in c.lower() and c in test.columns]
protection_cols = [c for c in train.columns if ('protection' in c.lower() or 'backup' in c.lower()) and c in test.columns]

train['services_count'] = train[streaming_cols + protection_cols].sum(axis=1) + train['phoneService']
test['services_count'] = test[streaming_cols + protection_cols].sum(axis=1) + test['phoneService']
print("✅ Created: services_count")

# Feature 4: customer engagement score
train['engagement_score'] = (
    train['serviceSatisfactionScore'] + 
    train['mobileAppSatisfaction'] + 
    train['networkStabilityScore']
) / 3
test['engagement_score'] = (
    test['serviceSatisfactionScore'] + 
    test['mobileAppSatisfaction'] + 
    test['networkStabilityScore']
) / 3
print("✅ Created: engagement_score")

# Feature 5: problem indicator
train['problem_indicator'] = (
    train['billingIssues'] + 
    train['supportTickets'] + 
    train['dataLimitWarnings']
)
test['problem_indicator'] = (
    test['billingIssues'] + 
    test['supportTickets'] + 
    test['dataLimitWarnings']
)
print("✅ Created: problem_indicator")

# Feature 6: negative sentiment indicator
train['is_negative_sentiment'] = (train['sentiment_polarity'] < -0.1).astype(int)
test['is_negative_sentiment'] = (test['sentiment_polarity'] < -0.1).astype(int)
print("✅ Created: is_negative_sentiment")

print(f"\n✅ Final train shape: {train.shape}")
print(f"✅ Final test shape: {test.shape}")

# حذف ستون‌های غیرضروری و آماده‌سازی نهایی
cols_to_drop = ['Unnamed: 0', 'custId', 'customerFeedback', 'custExit']
cols_to_drop = [c for c in cols_to_drop if c in train.columns]

X_train = train.drop(cols_to_drop, axis=1)
X_test = test.drop(['custId', 'customerFeedback'], axis=1)
y_train = (train['custExit'] == 'Yes').astype(int)

print(f"\n✅ Final X_train shape: {X_train.shape}")
print(f"\n✅ Final X_test shape: {X_test.shape}")
print(f"✅ Final y_train shape: {y_train.shape}")
print(f"✅ Columns match: {X_train.shape[1] == X_test.shape[1]}")

print("\n" + "="*80)
print("FEATURE ENGINEERING COMPLETED!")
print("="*80)

```

```python
# ============================================================================
# FIX: FINAL COLUMN ALIGNMENT
# ============================================================================
print("\n" + "="*80)
print("FIXING COLUMN MISMATCH")
print("="*80)

print(f"Before fix - X_train: {X_train.shape}, X_test: {X_test.shape}")

# یافتن ستون‌های گمشده در test
missing_in_test = set(X_train.columns) - set(X_test.columns)
if missing_in_test:
    print(f"\nColumns missing in test (adding with 0): {missing_in_test}")
    for col in missing_in_test:
        X_test[col] = 0

# یافتن ستون‌های اضافی در test
extra_in_test = set(X_test.columns) - set(X_train.columns)
if extra_in_test:
    print(f"\nColumns extra in test (removing): {extra_in_test}")
    X_test = X_test.drop(columns=list(extra_in_test))

# مرتب کردن ستون‌های test مطابق train
X_test = X_test[X_train.columns]

print(f"\nAfter fix - X_train: {X_train.shape}, X_test: {X_test.shape}")
print(f"✅ Columns match: {X_train.shape[1] == X_test.shape[1]}")
print(f"✅ Column order match: {list(X_train.columns) == list(X_test.columns)}")

# بررسی نهایی
print("\n" + "="*80)
print("FINAL DATA SUMMARY")
print("="*80)
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_train distribution:\n{y_train.value_counts()}")
print(f"Churn rate: {y_train.mean()*100:.2f}%")

print("\n✅ Data is ready for modeling!")
```

```python
# ============================================================================
# STEP 4: MODEL TRAINING WITH GPU OPTIMIZATION
# ============================================================================
print("\n" + "="*80)
print("STEP 4: GPU-OPTIMIZED MODEL TRAINING")
print("="*80)

from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score
import time

# محاسبه class weight برای imbalance
scale_pos_weight = (y_train==0).sum() / (y_train==1).sum()
print(f"Class imbalance ratio: {scale_pos_weight:.2f}")
print(f"Churn rate: {y_train.mean()*100:.2f}%")

# تنظیمات Cross-Validation
cv = StratifiedKFold(n_splits=6, shuffle=True, random_state=42)

# مدل‌های بهینه‌شده برای GPU
models = {
    'LightGBM_CPU': LGBMClassifier(
        device='cpu',  # استفاده از GPU
        gpu_use_dp=False,  # single precision برای GTX 1650
        n_estimators=300,
        max_depth=8,
        learning_rate=0.05,
        num_leaves=50,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        min_child_samples=20,
        reg_alpha=0.1,
        reg_lambda=0.1,
        random_state=42,
        verbose=-1
    ),
    'XGBoost_GPU': XGBClassifier(
        tree_method='hist',  # hist method برای GPU
        device='cuda',  # استفاده از GPU
        n_estimators=300,
        max_depth=7,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        min_child_weight=3,
        gamma=0.1,
        reg_alpha=0.1,
        reg_lambda=1.0,
        random_state=42,
        eval_metric='logloss'
    ),
    'RandomForest_CPU': RandomForestClassifier(
        n_estimators=200,
        max_depth=15,
        min_samples_split=10,
        min_samples_leaf=4,
        max_features='sqrt',
        class_weight='balanced',
        random_state=42,
        n_jobs=-1  # همه CPU cores
    )
}

print(f"\n📊 Training {len(models)} models with 5-fold CV...")

# آموزش و ارزیابی
results = {}

for name, model in models.items():
    print(f"\n{'='*60}")
    print(f"🚀 Training {name}...")
    print(f"{'='*60}")
    
    start_time = time.time()
    
    # Cross-validation
    print("  Running 5-fold cross-validation...")
    cv_start = time.time()
    cv_scores = cross_val_score(model, X_train, y_train, cv=cv, 
                                scoring='roc_auc', n_jobs=1 if 'GPU' in name else -1)
    cv_time = time.time() - cv_start
    
    # Train on full data
    print("  Training on full dataset...")
    fit_start = time.time()
    model.fit(X_train, y_train)
    fit_time = time.time() - fit_start
    
    # Predictions
    y_pred_proba = model.predict_proba(X_train)[:, 1]
    train_auc = roc_auc_score(y_train, y_pred_proba)
    
    total_time = time.time() - start_time
    
    # Store results
    results[name] = {
        'model': model,
        'cv_auc_mean': cv_scores.mean(),
        'cv_auc_std': cv_scores.std(),
        'train_auc': train_auc,
        'cv_time': cv_time,
        'fit_time': fit_time,
        'total_time': total_time
    }
    
    print(f"  ✅ CV ROC-AUC: {cv_scores.mean():.4f} (±{cv_scores.std():.4f})")
    print(f"  ✅ Train ROC-AUC: {train_auc:.4f}")
    print(f"  ⏱️ CV Time: {cv_time:.1f}s | Fit Time: {fit_time:.1f}s | Total: {total_time:.1f}s")

# مقایسه نتایج
print("\n" + "="*80)
print("📊 MODEL COMPARISON SUMMARY")
print("="*80)

comparison_df = pd.DataFrame({
    'Model': list(results.keys()),
    'CV_AUC': [f"{results[m]['cv_auc_mean']:.4f} ± {results[m]['cv_auc_std']:.4f}" for m in results.keys()],
    'Train_AUC': [f"{results[m]['train_auc']:.4f}" for m in results.keys()],
    'Total_Time': [f"{results[m]['total_time']:.1f}s" for m in results.keys()],
    'Fit_Time': [f"{results[m]['fit_time']:.1f}s" for m in results.keys()]
})

comparison_df_numeric = pd.DataFrame({
    'Model': list(results.keys()),
    'CV_AUC_Mean': [results[m]['cv_auc_mean'] for m in results.keys()],
    'Train_AUC': [results[m]['train_auc'] for m in results.keys()],
    'Total_Time': [results[m]['total_time'] for m in results.keys()]
})

comparison_df_numeric = comparison_df_numeric.sort_values('CV_AUC_Mean', ascending=False)
print("\n" + comparison_df.to_string(index=False))

# انتخاب بهترین مدل
best_model_name = comparison_df_numeric.iloc[0]['Model']
best_model = results[best_model_name]['model']
best_cv_auc = results[best_model_name]['cv_auc_mean']

print(f"\n{'='*80}")
print(f"🏆 BEST MODEL: {best_model_name}")
print(f"{'='*80}")
print(f"   CV ROC-AUC: {best_cv_auc:.4f}")
print(f"   Expected Score: {best_cv_auc * 100:.2f}/100")
print(f"   Training Time: {results[best_model_name]['total_time']:.1f}s")

print("\n✅ Model training completed successfully!")
```
