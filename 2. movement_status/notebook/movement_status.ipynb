```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

# بارگذاری داده‌ها
train = pd.read_csv('/content/drive/MyDrive/Oltmpic 2024/initial_project/data/train.csv')
test = pd.read_csv('/content/drive/MyDrive/Oltmpic 2024/initial_project/data/test.csv')

print("="*50)
print("TRAIN DATA INFO")
print("="*50)
print(f"Shape: {train.shape}")
print(f"\nColumns: {train.columns.tolist()}")
print(f"\nData types:\n{train.dtypes}")
print(f"\nFirst few rows:\n{train.head()}")
print(f"\nBasic statistics:\n{train.describe()}")
print(f"\nMissing values:\n{train.isnull().sum()}")
print(f"\nLabel distribution:\n{train['label'].value_counts().sort_index()}")
print(f"\nNumber of unique labels: {train['label'].nunique()}")

print("\n" + "="*50)
print("TEST DATA INFO")
print("="*50)
print(f"Shape: {test.shape}")
print(f"\nColumns: {test.columns.tolist()}")
print(f"\nFirst few rows:\n{test.head()}")
print(f"\nBasic statistics:\n{test.describe()}")
print(f"\nMissing values:\n{test.isnull().sum()}")

```

```python
# Time series analysis
print("="*50)
print("TIME SERIES ANALYSIS")
print("="*50)

# Check timestamp format and sampling rate
train['timestamp'] = pd.to_datetime(train['timestamp'])
test['timestamp'] = pd.to_datetime(test['timestamp'])

# Calculate time differences
train['time_diff'] = train['timestamp'].diff().dt.total_seconds()
test['time_diff'] = test['timestamp'].diff().dt.total_seconds()

print(f"\nTrain time range: {train['timestamp'].min()} to {train['timestamp'].max()}")
print(f"Test time range: {test['timestamp'].min()} to {test['timestamp'].max()}")
print(f"\nSampling rate (seconds):")
print(f"  Train mean: {train['time_diff'].mean():.4f}, std: {train['time_diff'].std():.6f}")
print(f"  Test mean: {test['time_diff'].mean():.4f}, std: {test['time_diff'].std():.6f}")

# Check for gaps in time series
train['time_gap'] = train['time_diff'] > 0.025  # More than 25ms = gap
test['time_gap'] = test['time_diff'] > 0.025

print(f"\nNumber of time gaps in train: {train['time_gap'].sum()}")
print(f"Number of time gaps in test: {test['time_gap'].sum()}")

# Label transition analysis
train['label_change'] = (train['label'] != train['label'].shift(1)).astype(int)
print(f"\nNumber of activity transitions: {train['label_change'].sum()}")
print(f"Average samples per activity segment: {len(train) / train['label_change'].sum():.2f}")

# Sample a few examples from each class
print("\n" + "="*50)
print("SAMPLES PER CLASS (first 3 rows of each)")
print("="*50)
for label in sorted(train['label'].unique()):
    print(f"\n--- Class {label} (count: {(train['label']==label).sum()}) ---")
    sample = train[train['label']==label].head(3)[['back_x', 'back_y', 'back_z', 'thigh_x', 'thigh_y', 'thigh_z']]
    print(sample.to_string(index=False))

# Clean temporary columns
train.drop(['time_diff', 'time_gap', 'label_change'], axis=1, inplace=True)
test.drop(['time_diff', 'time_gap'], axis=1, inplace=True)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt

# Feature Engineering - Magnitude and derived features
def create_basic_features(df):
    df_feat = df.copy()

    # Magnitude of acceleration (L2 norm)
    df_feat['back_magnitude'] = np.sqrt(df['back_x']**2 + df['back_y']**2 + df['back_z']**2)
    df_feat['thigh_magnitude'] = np.sqrt(df['thigh_x']**2 + df['thigh_y']**2 + df['thigh_z']**2)

    # Total magnitude
    df_feat['total_magnitude'] = np.sqrt(
        df['back_x']**2 + df['back_y']**2 + df['back_z']**2 +
        df['thigh_x']**2 + df['thigh_y']**2 + df['thigh_z']**2
    )

    # Differences between sensors
    df_feat['diff_x'] = df['back_x'] - df['thigh_x']
    df_feat['diff_y'] = df['back_y'] - df['thigh_y']
    df_feat['diff_z'] = df['back_z'] - df['thigh_z']
    df_feat['diff_magnitude'] = np.sqrt(df_feat['diff_x']**2 + df_feat['diff_y']**2 + df_feat['diff_z']**2)

    return df_feat

print("Creating basic features...")
train_feat = create_basic_features(train)
test_feat = create_basic_features(test)

# Check correlations with label
print("\n" + "="*50)
print("FEATURE CORRELATION WITH LABEL")
print("="*50)
feature_cols = [col for col in train_feat.columns if col not in ['timestamp', 'label']]
correlations = train_feat[feature_cols + ['label']].corr()['label'].drop('label').sort_values(ascending=False)
print(correlations)

print("\n" + "="*50)
print("BASELINE MODEL - Random Forest with basic features")
print("="*50)

# Prepare data
X_train = train_feat[feature_cols]
y_train = train_feat['label']
X_test = test_feat[feature_cols]

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train a baseline Random Forest
rf_baseline = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_split=10,
    random_state=42,
    n_jobs=-1,
    class_weight='balanced'  # Important for imbalanced data
)

rf_baseline.fit(X_train_scaled, y_train)

# Predict on train for evaluation
y_train_pred = rf_baseline.predict(X_train_scaled)
train_f1 = f1_score(y_train, y_train_pred, average='macro')

print(f"\nTrain F1 Score (macro): {train_f1:.4f}")
print(f"\nClassification Report on Train:\n")
print(classification_report(y_train, y_train_pred))

# Feature importance
feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': rf_baseline.feature_importances_
}).sort_values('importance', ascending=False)

print("\n" + "="*50)
print("FEATURE IMPORTANCE (Top 10)")
print("="*50)
print(feature_importance.head(10).to_string(index=False))
```

```python
from scipy import stats
from scipy.signal import find_peaks

def create_window_features(df, window_size=50, overlap=25):
    """
    Create statistical features from sliding windows
    window_size=50 means 1 second (50Hz sampling rate)
    overlap=25 means 50% overlap
    """

    features_list = []
    sensor_cols = ['back_x', 'back_y', 'back_z', 'thigh_x', 'thigh_y', 'thigh_z']

    # Add magnitude columns
    df['back_magnitude'] = np.sqrt(df['back_x']**2 + df['back_y']**2 + df['back_z']**2)
    df['thigh_magnitude'] = np.sqrt(df['thigh_x']**2 + df['thigh_y']**2 + df['thigh_z']**2)

    all_sensor_cols = sensor_cols + ['back_magnitude', 'thigh_magnitude']

    # Sliding window
    step = window_size - overlap

    for i in range(0, len(df) - window_size + 1, step):
        window = df.iloc[i:i+window_size]
        window_features = {}

        # For each sensor column
        for col in all_sensor_cols:
            values = window[col].values

            # Statistical features
            window_features[f'{col}_mean'] = np.mean(values)
            window_features[f'{col}_std'] = np.std(values)
            window_features[f'{col}_min'] = np.min(values)
            window_features[f'{col}_max'] = np.max(values)
            window_features[f'{col}_median'] = np.median(values)
            window_features[f'{col}_range'] = np.max(values) - np.min(values)
            window_features[f'{col}_q25'] = np.percentile(values, 25)
            window_features[f'{col}_q75'] = np.percentile(values, 75)
            window_features[f'{col}_iqr'] = np.percentile(values, 75) - np.percentile(values, 25)

            # Additional features
            window_features[f'{col}_rms'] = np.sqrt(np.mean(values**2))
            window_features[f'{col}_skew'] = stats.skew(values)
            window_features[f'{col}_kurtosis'] = stats.kurtosis(values)

            # Zero crossing rate
            window_features[f'{col}_zcr'] = np.sum(np.diff(np.sign(values)) != 0) / len(values)

            # Energy
            window_features[f'{col}_energy'] = np.sum(values**2)

        # Label (majority voting in window)
        if 'label' in df.columns:
            window_features['label'] = window['label'].mode()[0]

        # Keep middle timestamp
        window_features['timestamp'] = window['timestamp'].iloc[len(window)//2]

        features_list.append(window_features)

    return pd.DataFrame(features_list)

print("="*50)
print("CREATING WINDOW-BASED FEATURES")
print("="*50)
print("This may take a few minutes...")
print(f"Train samples: {len(train)} -> Windows: ~{len(train)//25}")
print(f"Test samples: {len(test)} -> Windows: ~{len(test)//25}")

# Create windowed features
train_windowed = create_window_features(train, window_size=50, overlap=25)
test_windowed = create_window_features(test, window_size=50, overlap=25)

print(f"\nDone!")
print(f"Train windowed shape: {train_windowed.shape}")
print(f"Test windowed shape: {test_windowed.shape}")
print(f"\nNumber of features created: {train_windowed.shape[1] - 2}")  # -2 for timestamp and label

print(f"\nSample of features:")
print(train_windowed.head(3))
```

```python
import time
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

# Check GPU availability
import torch
print("="*50)
print("GPU SETUP CHECK")
print("="*50)
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU Device: {torch.cuda.get_device_name(0)}")
    print(f"GPU Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")

print("\n" + "="*50)
print("TRAINING GPU-ACCELERATED MODELS")
print("="*50)

# Prepare windowed data
feature_cols_windowed = [col for col in train_windowed.columns if col not in ['timestamp', 'label']]
X_train_win = train_windowed[feature_cols_windowed]
y_train_win = train_windowed['label']
X_test_win = test_windowed[feature_cols_windowed]

# Scale features
scaler_win = StandardScaler()
X_train_win_scaled = scaler_win.fit_transform(X_train_win)
X_test_win_scaled = scaler_win.transform(X_test_win)

# Define GPU-accelerated models
models = {
    'XGBoost_GPU': XGBClassifier(
        n_estimators=500,
        max_depth=12,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        tree_method='gpu_hist',  # GPU acceleration
        gpu_id=0,
        random_state=42,
        eval_metric='mlogloss',
        verbosity=1
    ),
    'LightGBM_GPU': LGBMClassifier(
        n_estimators=500,
        max_depth=15,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        class_weight='balanced',
        device='gpu',  # GPU acceleration
        gpu_platform_id=0,
        gpu_device_id=0,
        random_state=42,
        verbose=-1
    ),
    'CatBoost_GPU': CatBoostClassifier(
        iterations=500,
        depth=10,
        learning_rate=0.1,
        task_type='GPU',  # GPU acceleration
        devices='0',
        random_state=42,
        verbose=100,
        auto_class_weights='Balanced'
    ),
    'RandomForest': RandomForestClassifier(
        n_estimators=200,  # کمتر برای سرعت
        max_depth=25,
        min_samples_split=5,
        min_samples_leaf=2,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1  # CPU parallel
    )
}

# Train and evaluate each model
results = {}

for name, model in models.items():
    print(f"\n{'='*50}")
    print(f"Training {name}...")
    print(f"{'='*50}")

    try:
        start_time = time.time()
        model.fit(X_train_win_scaled, y_train_win)
        train_time = time.time() - start_time

        # Predictions on train
        y_train_pred = model.predict(X_train_win_scaled)
        train_f1 = f1_score(y_train_win, y_train_pred, average='macro')

        results[name] = {
            'model': model,
            'train_f1': train_f1,
            'train_time': train_time
        }

        print(f"✓ Training time: {train_time:.2f} seconds")
        print(f"✓ Train F1 Score (macro): {train_f1:.4f}")
        print(f"\nClassification Report:")
        print(classification_report(y_train_win, y_train_pred))

    except Exception as e:
        print(f"✗ Error training {name}: {str(e)}")
        print(f"  Skipping this model...")

# Find best model
print("\n" + "="*50)
print("MODEL COMPARISON SUMMARY")
print("="*50)
comparison_df = pd.DataFrame({
    'Model': list(results.keys()),
    'Train F1': [results[m]['train_f1'] for m in results.keys()],
    'Training Time (s)': [results[m]['train_time'] for m in results.keys()]
}).sort_values('Train F1', ascending=False)

print(comparison_df.to_string(index=False))

best_model_name = comparison_df.iloc[0]['Model']
best_model = results[best_model_name]['model']

print(f"\n🏆 Best Model: {best_model_name} with F1 = {results[best_model_name]['train_f1']:.4f}")

# Feature importance از بهترین مدل
if hasattr(best_model, 'feature_importances_'):
    feature_importance = pd.DataFrame({
        'feature': feature_cols_windowed,
        'importance': best_model.feature_importances_
    }).sort_values('importance', ascending=False)

    print("\n" + "="*50)
    print("TOP 20 IMPORTANT FEATURES")
    print("="*50)
    print(feature_importance.head(20).to_string(index=False))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

print("="*50)
print("CROSS-VALIDATION FOR REALISTIC PERFORMANCE")
print("="*50)

# Test multiple models with CV
models_cv = {
    'LightGBM_Regularized': LGBMClassifier(
        n_estimators=300,
        max_depth=10,
        learning_rate=0.05,
        subsample=0.7,
        colsample_bytree=0.7,
        min_child_samples=20,
        reg_alpha=0.1,
        reg_lambda=0.1,
        class_weight='balanced',
        device='gpu',
        gpu_platform_id=0,
        gpu_device_id=0,
        random_state=42,
        verbose=-1
    ),
    'CatBoost_Regularized': CatBoostClassifier(
        iterations=300,
        depth=8,
        learning_rate=0.05,
        l2_leaf_reg=3,
        task_type='GPU',
        devices='0',
        random_state=42,
        verbose=0,
        auto_class_weights='Balanced'
    ),
    'RandomForest_Tuned': RandomForestClassifier(
        n_estimators=200,
        max_depth=20,
        min_samples_split=10,
        min_samples_leaf=4,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    )
}

# Stratified K-Fold (5-fold)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_results = {}

for name, model in models_cv.items():
    print(f"\n{'='*50}")
    print(f"Cross-validating {name}...")
    print(f"{'='*50}")

    start_time = time.time()

    # Cross-validation with F1 macro scoring
    cv_scores = cross_val_score(
        model,
        X_train_win_scaled,
        y_train_win,
        cv=skf,
        scoring='f1_macro',
        n_jobs=1  # برای GPU models باید 1 باشه
    )

    cv_time = time.time() - start_time

    cv_results[name] = {
        'cv_scores': cv_scores,
        'cv_mean': cv_scores.mean(),
        'cv_std': cv_scores.std(),
        'cv_time': cv_time
    }

    print(f"✓ CV Time: {cv_time:.2f} seconds")
    print(f"✓ CV F1 Scores: {cv_scores}")
    print(f"✓ Mean F1: {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")

# Compare models
print("\n" + "="*50)
print("CROSS-VALIDATION COMPARISON")
print("="*50)
cv_comparison = pd.DataFrame({
    'Model': list(cv_results.keys()),
    'Mean F1': [cv_results[m]['cv_mean'] for m in cv_results.keys()],
    'Std F1': [cv_results[m]['cv_std'] for m in cv_results.keys()],
    'CV Time (s)': [cv_results[m]['cv_time'] for m in cv_results.keys()]
}).sort_values('Mean F1', ascending=False)

print(cv_comparison.to_string(index=False))

best_cv_model_name = cv_comparison.iloc[0]['Model']
print(f"\n🏆 Best CV Model: {best_cv_model_name} with F1 = {cv_comparison.iloc[0]['Mean F1']:.4f} (+/- {cv_comparison.iloc[0]['Std F1']:.4f})")

print("\n" + "="*50)
print("TRAINING FINAL MODEL ON FULL TRAIN DATA")
print("="*50)

# حالا بهترین مدل رو انتخاب کن
final_model = models_cv[best_cv_model_name]

print(f"\nTraining {best_cv_model_name} on full training data...")
start_time = time.time()
final_model.fit(X_train_win_scaled, y_train_win)
train_time = time.time() - start_time

# Train evaluation
y_train_pred_final = final_model.predict(X_train_win_scaled)
train_f1_final = f1_score(y_train_win, y_train_pred_final, average='macro')

print(f"✓ Training time: {train_time:.2f} seconds")
print(f"✓ Train F1 Score (macro): {train_f1_final:.4f}")
print(f"\nClassification Report on Train:")
print(classification_report(y_train_win, y_train_pred_final))

# Confusion matrix
cm = confusion_matrix(y_train_win, y_train_pred_final)
print("\nConfusion Matrix:")
print(cm)

print("\n" + "="*50)
print("GENERATING TEST PREDICTIONS")
print("="*50)

# Predict on test
y_test_pred = final_model.predict(X_test_win_scaled)

print(f"Test predictions shape: {y_test_pred.shape}")
print(f"Test predictions distribution:")
print(pd.Series(y_test_pred).value_counts().sort_index())

# Create submission
test_windowed['predicted_label'] = y_test_pred

print("\nSample of predictions:")
print(test_windowed[['timestamp', 'predicted_label']].head(10))
```

```python
print("="*50)
print("MAPPING PREDICTIONS BACK TO ORIGINAL TEST DATA")
print("="*50)

# مشکل: ما predictions برای windows داریم، ولی باید برای هر row اصلی پیش‌بینی بدیم

# استراتژی: از timestamp استفاده می‌کنیم
# هر window یک timestamp میانی داره، باید اون رو به سطرهای اصلی map کنیم

# ساده‌ترین روش: Forward Fill
test_predictions = pd.DataFrame({
    'timestamp': test['timestamp'],
})

# Merge با windowed predictions
test_windowed_pred = test_windowed[['timestamp', 'predicted_label']].copy()

# Merge - از asof برای نزدیک‌ترین timestamp استفاده می‌کنیم
test['timestamp'] = pd.to_datetime(test['timestamp'])
test_windowed_pred['timestamp'] = pd.to_datetime(test_windowed_pred['timestamp'])

test_predictions = pd.merge_asof(
    test[['timestamp']].sort_values('timestamp'),
    test_windowed_pred.sort_values('timestamp'),
    on='timestamp',
    direction='nearest'
)

print(f"Original test shape: {test.shape}")
print(f"Predictions shape: {test_predictions.shape}")
print(f"\nPredictions distribution:")
print(test_predictions['predicted_label'].value_counts().sort_index())

print("\nSample of final predictions:")
print(test_predictions.head(20))

# Check for missing predictions
if test_predictions['predicted_label'].isnull().any():
    print(f"\n⚠️ Warning: {test_predictions['predicted_label'].isnull().sum()} missing predictions!")
    # Fill با forward fill
    test_predictions['predicted_label'].fillna(method='ffill', inplace=True)
    test_predictions['predicted_label'].fillna(method='bfill', inplace=True)

# تبدیل به int
test_predictions['predicted_label'] = test_predictions['predicted_label'].astype(int)

print("\n" + "="*50)
print("FINAL SUBMISSION PREPARATION")
print("="*50)
print(f"Total predictions: {len(test_predictions)}")
print(f"Expected: {len(test)}")
print(f"Match: {len(test_predictions) == len(test)}")

# Save predictions
submission = test_predictions[['predicted_label']].copy()
submission.columns = ['label']  # Rename column to 'label' if needed

print("\nSubmission preview:")
print(submission.head(10))
print(f"\nSubmission shape: {submission.shape}")
print(f"Label distribution in submission:")
print(submission['label'].value_counts().sort_index())

# Save to CSV
submission.to_csv('submission.csv', index=False)
print("\n✅ Submission file saved as 'submission.csv'")
```
