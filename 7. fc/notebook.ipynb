<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
برای این سوال باید هدف شما نوشتن یک تابع پیشنهاددهنده بازیکن مناسب، به نام <code>player_recommender</code> باشد.
    <br>
این تابع باید آیدی دو بازیکن فروخته شده به همراه دیتاست را دریافت کرده و به ازای هر بازیکن یک جانشین پیشنهاد کرده و آیدی آن را به عنوان خروجی دهد. یعنی در نهایت خروجی این تابع دو int است که آیدی بازیکنان پیشنهادی برای جایگزینی را مشخص میکند.
</font>
</p>

```python
from sklearn.metrics.pairwise import cosine_similarity

def player_recommender(sold_player1_id, sold_player2_id, dataset):
    """
    تابع پیشنهاد بازیکن - نسخه نهایی
    
    Args:
        sold_player1_id: ID بازیکن اول فروخته شده
        sold_player2_id: ID بازیکن دوم فروخته شده  
        dataset: DataFrame کامل بازیکنان
        
    Returns:
        tuple: (replacement1_id, replacement2_id)
    """
    
    df = dataset.copy()
    
    # استخراج اطلاعات بازیکنان فروخته شده
    sold1 = df[df['player_id'] == sold_player1_id].iloc[0]
    sold2 = df[df['player_id'] == sold_player2_id].iloc[0]
    
    total_budget = (sold1['current_mv'] + sold2['current_mv']) * 0.8
    
    # فیلتر کاندیداها
    def get_candidates(sold_player, df, exclude_ids):
        candidates = df[
            (~df['player_id'].isin(exclude_ids)) & 
            (df['current_mv'].notna()) & 
            (df['current_mv'] > 0) &
            (df['OVR'] >= max(75, sold_player['OVR'] - 7))
        ].copy()
        
        position = sold_player['Position']
        alt_positions = sold_player['Alternative positions']
        
        allowed_positions = {position}
        if pd.notna(alt_positions):
            allowed_positions.update([p.strip() for p in str(alt_positions).split(',')])
        
        def check_position(row):
            if row['Position'] in allowed_positions:
                return True
            if pd.notna(row['Alternative positions']):
                row_alts = {p.strip() for p in str(row['Alternative positions']).split(',')}
                if allowed_positions & row_alts:
                    return True
            return False
        
        candidates = candidates[candidates.apply(check_position, axis=1)]
        
        def position_priority(row):
            if row['Position'] == position:
                return 1.0
            elif position in str(row['Alternative positions']):
                return 0.8
            else:
                return 0.6
        
        candidates['position_priority'] = candidates.apply(position_priority, axis=1)
        return candidates
    
    exclude_ids = [sold_player1_id, sold_player2_id]
    candidates1 = get_candidates(sold1, df, exclude_ids)
    candidates2 = get_candidates(sold2, df, exclude_ids)
    
    if len(candidates1) == 0 or len(candidates2) == 0:
        return None, None
    
    # امتیازدهی
    tech_features = ['PAC', 'SHO', 'PAS', 'DRI', 'DEF', 'PHY']
    
    def compute_similarity(sold, cand):
        sold_vec = sold[tech_features].values.astype(float)
        cand_vec = cand[tech_features].values.astype(float)
        
        position = sold['Position']
        weights = np.ones(6)
        
        if position in ['ST', 'CF', 'LW', 'RW']:
            weights = np.array([1.2, 1.5, 1.0, 1.3, 0.5, 1.0])
        elif position in ['CAM', 'CM', 'LM', 'RM']:
            weights = np.array([1.0, 1.0, 1.5, 1.3, 0.8, 0.8])
        elif position in ['CDM']:
            weights = np.array([0.8, 0.7, 1.3, 0.9, 1.5, 1.2])
        elif position in ['CB', 'LB', 'RB']:
            weights = np.array([1.0, 0.5, 0.9, 0.7, 1.5, 1.3])
        
        sold_vec = sold_vec * weights
        cand_vec = cand_vec * weights
        
        similarity = np.dot(sold_vec, cand_vec) / (np.linalg.norm(sold_vec) * np.linalg.norm(cand_vec))
        return similarity
    
    def score_pair(sold1, sold2, cand1, cand2, budget):
        total_price = cand1['current_mv'] + cand2['current_mv']
        if total_price > budget:
            return -1
        
        # شباهت پروفایل
        sim1 = compute_similarity(sold1, cand1)
        sim2 = compute_similarity(sold2, cand2)
        similarity_score = (sim1 + sim2) / 2
        position_score = (cand1['position_priority'] + cand2['position_priority']) / 2
        
        # حفظ کیفیت
        def ovr_score(diff):
            if diff >= 0:
                return 1.0 + diff * 0.04
            elif diff >= -2:
                return 1.0 + diff * 0.03
            elif diff >= -4:
                return 0.85 + diff * 0.03
            else:
                return 0.65 + diff * 0.02
        
        quality1 = ovr_score(cand1['OVR'] - sold1['OVR'])
        quality2 = ovr_score(cand2['OVR'] - sold2['OVR'])
        quality_score = (quality1 + quality2) / 2
        
        # جوانگرایی
        def youth_score(diff):
            if diff >= 3:
                return 1.25
            elif diff >= 1:
                return 1.15
            elif diff >= -1:
                return 1.05
            elif diff >= -3:
                return 0.95
            else:
                return 0.80
        
        youth1 = youth_score(sold1['Age'] - cand1['Age'])
        youth2 = youth_score(sold2['Age'] - cand2['Age'])
        youth_score_val = (youth1 + youth2) / 2
        
        # ارزش مالی
        price_ratio = total_price / budget
        if price_ratio <= 0.7:
            financial_score = 1.15
        elif price_ratio <= 0.85:
            financial_score = 1.25
        elif price_ratio <= 0.95:
            financial_score = 1.15
        else:
            financial_score = 1.0
        
        pair_score = (
            similarity_score * 0.35 * position_score +
            quality_score * 0.35 +
            youth_score_val * 0.20 +
            financial_score * 0.10
        )
        
        return pair_score
    
    # جستجو
    candidates1 = candidates1.nlargest(400, 'OVR')
    candidates2 = candidates2.nlargest(400, 'OVR')
    
    best_score = -1
    best_pair = None
    
    for _, cand1 in candidates1.iterrows():
        for _, cand2 in candidates2.iterrows():
            if cand1['player_id'] == cand2['player_id']:
                continue
            
            score = score_pair(sold1, sold2, cand1, cand2, total_budget)
            
            if score > best_score:
                best_score = score
                best_pair = (cand1, cand2)
    
    if best_pair is None:
        return None, None
    
    best_cand1, best_cand2 = best_pair
    
    return int(best_cand1['player_id']), int(best_cand2['player_id'])



    
```
