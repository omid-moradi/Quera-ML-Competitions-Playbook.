```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
import warnings
warnings.filterwarnings('ignore')

# Load the dataset
path = r"F:\Quera\olympic 2024\fc\Data\players.csv"
df = pd.read_csv(path)

# Initial exploration
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nFirst few rows:")
print(df.head())
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
print("\nBasic statistics:")
print(df.describe())
```

```python
print("Original shape:", df.shape)

# 1. Drop unnecessary columns
columns_to_drop = ['url', 'Unnamed: 57', 'Rank', 'Name']
df = df.drop(columns=columns_to_drop, errors='ignore')

print(f"\nAfter dropping {columns_to_drop}:")
print("New shape:", df.shape)

# 2. Fill GK-specific columns with 0 for non-GK players
gk_columns = ['GK Diving', 'GK Handling', 'GK Kicking', 'GK Positioning', 'GK Reflexes']
for col in gk_columns:
    df[col] = df[col].fillna(0)
    
print("\nGK columns filled with 0 for non-GK players")
print("Missing values in GK columns after filling:")
print(df[gk_columns].isnull().sum())

# 3. Check current_mv column (has 0 values and missing values)
print("\n--- Analysis of current_mv column ---")
print(f"Missing values: {df['current_mv'].isnull().sum()}")
print(f"Zero values: {(df['current_mv'] == 0).sum()}")
print(f"Min value (excluding zeros): {df[df['current_mv'] > 0]['current_mv'].min()}")
print(f"Median value: {df['current_mv'].median()}")
print("\nSample of rows with missing current_mv:")
print(df[df['current_mv'].isnull()][['player_id', 'Position', 'OVR', 'Age', 'Team', 'current_mv']].head(10))

# 4. Check other missing values
print("\n--- Remaining missing values ---")
print(df.isnull().sum()[df.isnull().sum() > 0])

# 5. Display final cleaned dataset info
print("\n--- Final Dataset Info ---")
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData types:")
print(df.dtypes.value_counts())

# Save a preview
print("\n--- First few rows after cleaning ---")
print(df.head())
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

# Assuming df is already loaded and cleaned from previous step
# Let's handle missing values strategically

print("=== Handling Missing Values ===\n")

# 1. Handle Age (only 1 missing)
age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)
print(f"✓ Age: Filled 1 missing value with median ({age_median})")

# 2. Handle Alternative positions
# Fill with empty string for players with only one position
df['Alternative positions'] = df['Alternative positions'].fillna('')
print(f"✓ Alternative positions: Filled {7247} missing values with empty string")

# 3. Handle play style
# Fill with 'Unknown' for now
df['play style'] = df['play style'].fillna('Unknown')
print(f"✓ Play style: Filled {8682} missing values with 'Unknown'")

# 4. Handle current_mv (most critical)
print("\n--- Handling current_mv ---")

# Identify rows with valid current_mv (> 0 and not null)
valid_mv_mask = (df['current_mv'] > 0) & (df['current_mv'].notna())
missing_mv_mask = (df['current_mv'] == 0) | (df['current_mv'].isna())

print(f"Valid current_mv: {valid_mv_mask.sum()}")
print(f"Missing/Zero current_mv: {missing_mv_mask.sum()}")

# Create features for predicting current_mv
# We'll use: OVR, Age, PAC, SHO, PAS, DRI, DEF, PHY, Position, League
feature_cols = ['OVR', 'Age', 'PAC', 'SHO', 'PAS', 'DRI', 'DEF', 'PHY', 
                'Weak foot', 'Skill moves', 'height', 'weight']

# Prepare training data (only rows with valid current_mv)
X_train = df[valid_mv_mask][feature_cols].copy()
y_train = df[valid_mv_mask]['current_mv'].copy()

# Prepare prediction data (rows with missing current_mv)
X_predict = df[missing_mv_mask][feature_cols].copy()

print(f"\nTraining samples: {len(X_train)}")
print(f"Prediction samples: {len(X_predict)}")

# Train a Random Forest model
print("\nTraining Random Forest model to predict missing current_mv values...")
rf_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1, max_depth=15)
rf_model.fit(X_train, y_train)

# Predict missing values
predicted_mv = rf_model.predict(X_predict)

# Ensure no negative predictions
predicted_mv = np.maximum(predicted_mv, 5000)  # Minimum value observed is 5000

print(f"✓ Predicted current_mv for {len(predicted_mv)} players")
print(f"  - Min predicted: {predicted_mv.min():,.0f}")
print(f"  - Max predicted: {predicted_mv.max():,.0f}")
print(f"  - Mean predicted: {predicted_mv.mean():,.0f}")
print(f"  - Median predicted: {np.median(predicted_mv):,.0f}")

# Fill the missing values
df.loc[missing_mv_mask, 'current_mv'] = predicted_mv

# Verify no missing values remain
print("\n=== Final Missing Values Check ===")
missing_summary = df.isnull().sum()
print(missing_summary[missing_summary > 0])
if missing_summary.sum() == 0:
    print("✓ No missing values remaining!")

# Display statistics
print("\n=== Final Dataset Statistics ===")
print(f"Total players: {len(df)}")
print(f"Total features: {len(df.columns)}")
print(f"\ncurrent_mv statistics:")
print(df['current_mv'].describe())

# Show some examples of predicted values
print("\n=== Sample of Predicted current_mv ===")
sample_predicted = df[missing_mv_mask][['player_id', 'Position', 'OVR', 'Age', 'Team', 'current_mv']].head(10)
print(sample_predicted)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for better visualizations
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# Assuming df is already cleaned from previous steps

print("=== Correlation Analysis with current_mv (Target) ===\n")

# Select only numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

# Calculate correlation with current_mv
correlations = df[numeric_cols].corr()['current_mv'].sort_values(ascending=False)

print("Top 20 Features Most Correlated with current_mv:")
print("=" * 60)
for i, (feature, corr) in enumerate(correlations.head(20).items(), 1):
    print(f"{i:2d}. {feature:25s}: {corr:7.4f}")

print("\n" + "=" * 60)
print("\nBottom 10 Features (Least/Negative Correlation):")
print("=" * 60)
for i, (feature, corr) in enumerate(correlations.tail(10).items(), 1):
    print(f"{i:2d}. {feature:25s}: {corr:7.4f}")

# === VISUALIZATION 1: Top 20 Correlations Bar Plot ===
plt.figure(figsize=(12, 8))
top_20 = correlations.head(20)
colors = ['green' if x > 0 else 'red' for x in top_20.values]
plt.barh(range(len(top_20)), top_20.values, color=colors, alpha=0.7)
plt.yticks(range(len(top_20)), top_20.index, fontsize=10)
plt.xlabel('Correlation with current_mv', fontsize=12, fontweight='bold')
plt.title('Top 20 Features Correlated with Market Value', fontsize=14, fontweight='bold')
plt.axvline(x=0, color='black', linestyle='--', linewidth=0.8)
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('correlation_top20.png', dpi=300, bbox_inches='tight')
print("\n✓ Saved: correlation_top20.png")
plt.show()

# === VISUALIZATION 2: All Features Correlation (sorted) ===
plt.figure(figsize=(10, 14))
all_corr = correlations.drop('current_mv')  # Remove self-correlation
colors = ['green' if x > 0 else 'red' for x in all_corr.values]
plt.barh(range(len(all_corr)), all_corr.values, color=colors, alpha=0.6)
plt.yticks(range(len(all_corr)), all_corr.index, fontsize=8)
plt.xlabel('Correlation Coefficient', fontsize=12, fontweight='bold')
plt.title('All Features Correlation with Market Value (Sorted)', fontsize=14, fontweight='bold')
plt.axvline(x=0, color='black', linestyle='--', linewidth=1)
plt.axvline(x=0.5, color='blue', linestyle=':', linewidth=0.8, label='Strong Positive (>0.5)')
plt.axvline(x=-0.5, color='orange', linestyle=':', linewidth=0.8, label='Strong Negative (<-0.5)')
plt.legend(fontsize=10)
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('correlation_all_features.png', dpi=300, bbox_inches='tight')
print("✓ Saved: correlation_all_features.png")
plt.show()

# Group correlations by category
print("\n" + "=" * 60)
print("Correlation by Feature Category:")
print("=" * 60)

main_attributes = ['OVR', 'PAC', 'SHO', 'PAS', 'DRI', 'DEF', 'PHY']
technical_skills = ['Acceleration', 'Sprint Speed', 'Positioning', 'Finishing', 
                    'Shot Power', 'Long Shots', 'Volleys', 'Penalties']
passing_skills = ['Vision', 'Crossing', 'Free Kick Accuracy', 'Short Passing', 
                  'Long Passing', 'Curve']
dribbling_skills = ['Dribbling', 'Agility', 'Balance', 'Reactions', 'Ball Control', 'Composure']
defensive_skills = ['Interceptions', 'Heading Accuracy', 'Def Awareness', 
                    'Standing Tackle', 'Sliding Tackle']
physical_skills = ['Jumping', 'Stamina', 'Strength', 'Aggression']
gk_skills = ['GK Diving', 'GK Handling', 'GK Kicking', 'GK Positioning', 'GK Reflexes']
other_attrs = ['Weak foot', 'Skill moves', 'height', 'weight', 'Age']

categories = {
    'Main Attributes': main_attributes,
    'Technical Skills': technical_skills,
    'Passing Skills': passing_skills,
    'Dribbling Skills': dribbling_skills,
    'Defensive Skills': defensive_skills,
    'Physical Skills': physical_skills,
    'GK Skills': gk_skills,
    'Other Attributes': other_attrs
}

category_stats = []
for category, features in categories.items():
    available_features = [f for f in features if f in correlations.index]
    if available_features:
        avg_corr = correlations[available_features].mean()
        max_corr = correlations[available_features].max()
        max_feature = correlations[available_features].idxmax()
        category_stats.append({
            'Category': category,
            'Avg Correlation': avg_corr,
            'Max Correlation': max_corr,
            'Top Feature': max_feature
        })
        print(f"\n{category}:")
        print(f"  Average Correlation: {avg_corr:7.4f}")
        print(f"  Max Correlation: {max_corr:7.4f} ({max_feature})")

# === VISUALIZATION 3: Category-wise Average Correlation ===
plt.figure(figsize=(10, 6))
cat_df = pd.DataFrame(category_stats)
cat_df_sorted = cat_df.sort_values('Avg Correlation', ascending=True)
colors_cat = ['green' if x > 0 else 'red' for x in cat_df_sorted['Avg Correlation'].values]
plt.barh(cat_df_sorted['Category'], cat_df_sorted['Avg Correlation'], color=colors_cat, alpha=0.7)
plt.xlabel('Average Correlation with current_mv', fontsize=12, fontweight='bold')
plt.title('Average Correlation by Feature Category', fontsize=14, fontweight='bold')
plt.axvline(x=0, color='black', linestyle='--', linewidth=1)
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('correlation_by_category.png', dpi=300, bbox_inches='tight')
print("\n✓ Saved: correlation_by_category.png")
plt.show()

# === VISUALIZATION 4: Correlation Matrix - Main Attributes ===
print("\n" + "=" * 60)
print("Correlation Matrix - Main Attributes (Multicollinearity Check):")
print("=" * 60)
main_attr_corr = df[main_attributes].corr()
print(main_attr_corr.round(3))

plt.figure(figsize=(10, 8))
mask = np.triu(np.ones_like(main_attr_corr, dtype=bool), k=1)
sns.heatmap(main_attr_corr, annot=True, fmt='.3f', cmap='coolwarm', center=0,
            square=True, linewidths=1, cbar_kws={"shrink": 0.8},
            vmin=-1, vmax=1, mask=mask)
plt.title('Correlation Matrix - Main Attributes', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('correlation_matrix_main_attributes.png', dpi=300, bbox_inches='tight')
print("\n✓ Saved: correlation_matrix_main_attributes.png")
plt.show()

# === VISUALIZATION 5: Heatmap of Top Correlated Features ===
top_features = correlations.head(15).index.tolist()
top_features_corr = df[top_features].corr()

plt.figure(figsize=(12, 10))
mask = np.triu(np.ones_like(top_features_corr, dtype=bool), k=1)
sns.heatmap(top_features_corr, annot=True, fmt='.2f', cmap='RdYlGn', center=0,
            square=True, linewidths=0.5, cbar_kws={"shrink": 0.8},
            vmin=-1, vmax=1, mask=mask)
plt.title('Correlation Matrix - Top 15 Features (by current_mv correlation)', 
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('correlation_matrix_top15.png', dpi=300, bbox_inches='tight')
print("✓ Saved: correlation_matrix_top15.png")
plt.show()

# Identify highly correlated pairs (potential multicollinearity)
print("\n" + "=" * 60)
print("Highly Correlated Feature Pairs (|r| > 0.8):")
print("=" * 60)

corr_matrix = df[numeric_cols].corr()
upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

high_corr_pairs = []
for column in upper_triangle.columns:
    for index in upper_triangle.index:
        corr_value = upper_triangle.loc[index, column]
        if pd.notna(corr_value) and abs(corr_value) > 0.8:
            high_corr_pairs.append((index, column, corr_value))

high_corr_pairs.sort(key=lambda x: abs(x[2]), reverse=True)

for i, (feat1, feat2, corr) in enumerate(high_corr_pairs[:20], 1):
    print(f"{i:2d}. {feat1:25s} <-> {feat2:25s}: {corr:7.4f}")

# Summary insights
print("\n" + "=" * 60)
print("KEY INSIGHTS:")
print("=" * 60)
print(f"1. Most important feature: {correlations.index[0]} (r={correlations.iloc[0]:.4f})")
print(f"2. Average correlation (all features): {correlations.mean():.4f}")
print(f"3. Features with r > 0.5: {(correlations > 0.5).sum()}")
print(f"4. Features with r > 0.7: {(correlations > 0.7).sum()}")
print(f"5. Highly correlated pairs found: {len(high_corr_pairs)}")
print(f"6. Strong positive correlations: {(correlations > 0.5).sum()}")
print(f"7. Strong negative correlations: {(correlations < -0.5).sum()}")

print("\n✓ All visualizations saved successfully!")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
import warnings
warnings.filterwarnings('ignore')

# ========== OBJECTIVE FUNCTION DESIGN ==========
"""
چون ground truth نداریم، objective function رو بر اساس این اصول طراحی می‌کنیم:
1. تعادل بین همه 4 فاکتور (نه over-emphasis روی یکی)
2. نتایج منطقی (OVR نباید خیلی افت کنه، Age نباید خیلی زیاد شه)
3. استفاده خوب از بودجه (نه خیلی کم، نه خیلی زیاد)
4. تنوع در solutions (نه همیشه یک pattern)
"""

def create_test_scenarios(df):
    """
    Create diverse test scenarios for optimization.
    """
    scenarios = []
    
    # Scenario 1: High-value players (expensive)
    high_value_players = df.nlargest(100, 'current_mv')
    for i in range(0, min(20, len(high_value_players)), 2):
        scenarios.append((
            int(high_value_players.iloc[i]['player_id']),
            int(high_value_players.iloc[i+1]['player_id'])
        ))
    
    # Scenario 2: Mid-tier players (realistic)
    mid_ovr_players = df[(df['OVR'] >= 75) & (df['OVR'] <= 82)].sample(min(40, len(df)))
    for i in range(0, min(20, len(mid_ovr_players)), 2):
        scenarios.append((
            int(mid_ovr_players.iloc[i]['player_id']),
            int(mid_ovr_players.iloc[i+1]['player_id'])
        ))
    
    # Scenario 3: Different positions
    for pos_cat in ['Attacker', 'Midfielder', 'Defender', 'Goalkeeper']:
        pos_players = df[df['Position_Category'] == pos_cat].sample(min(10, len(df)))
        if len(pos_players) >= 2:
            scenarios.append((
                int(pos_players.iloc[0]['player_id']),
                int(pos_players.iloc[1]['player_id'])
            ))
    
    return scenarios[:30]  # Use 30 test scenarios


def evaluate_weights(weights, df, scenarios, verbose=False):
    """
    Evaluate a set of weights across multiple scenarios.
    
    weights: [sim_weight, quality_weight, youth_weight, financial_weight,
              quality_scale, youth_scale, financial_vfm_weight]
    """
    
    # Unpack weights
    sim_w = weights[0]
    quality_w = weights[1]
    youth_w = weights[2]
    financial_w = weights[3]
    quality_scale = weights[4]
    youth_scale = weights[5]
    financial_vfm_w = weights[6]
    
    # Normalize component weights to sum to 1
    total_w = sim_w + quality_w + youth_w + financial_w
    sim_w /= total_w
    quality_w /= total_w
    youth_w /= total_w
    financial_w /= total_w
    
    scores = []
    
    for sold_id_1, sold_id_2 in scenarios:
        try:
            # Run recommender with these weights
            result = player_recommender_optimizable(
                sold_id_1, sold_id_2, df,
                sim_w, quality_w, youth_w, financial_w,
                quality_scale, youth_scale, financial_vfm_w,
                verbose=False
            )
            
            if result is None:
                scores.append(-1000)  # Penalty for failure
                continue
            
            # Evaluate quality of recommendation
            score = evaluate_recommendation_quality(result, df, sold_id_1, sold_id_2)
            scores.append(score)
            
        except Exception as e:
            scores.append(-1000)
    
    # Aggregate score
    avg_score = np.mean(scores)
    min_score = np.min(scores)
    
    # Penalty for very bad worst-case
    final_score = avg_score + min_score * 0.1
    
    if verbose:
        print(f"Weights: sim={sim_w:.3f} qual={quality_w:.3f} youth={youth_w:.3f} fin={financial_w:.3f}")
        print(f"Scales: q_scale={quality_scale:.2f} y_scale={youth_scale:.2f} f_vfm={financial_vfm_w:.2f}")
        print(f"Score: avg={avg_score:.2f} min={min_score:.2f} final={final_score:.2f}")
    
    return final_score


def evaluate_recommendation_quality(result, df, sold_id_1, sold_id_2):
    """
    Evaluate how good a recommendation is based on principles.
    """
    if result is None or 'player1_id' not in result:
        return -1000
    
    sold_1 = df[df['player_id'] == sold_id_1].iloc[0]
    sold_2 = df[df['player_id'] == sold_id_2].iloc[0]
    
    rec_1 = df[df['player_id'] == result['player1_id']].iloc[0]
    rec_2 = df[df['player_id'] == result['player2_id']].iloc[0]
    
    score = 0
    
    # 1. OVR maintenance (very important)
    ovr_diff = ((rec_1['OVR'] - sold_1['OVR']) + (rec_2['OVR'] - sold_2['OVR'])) / 2
    if ovr_diff >= 0:
        score += ovr_diff * 20  # Reward improvement
    elif ovr_diff >= -2:
        score += ovr_diff * 5   # Small penalty for small drop
    else:
        score += (ovr_diff + 2) * 15 - 10  # Big penalty for big drop
    
    # 2. Age factor (important)
    age_diff = ((sold_1['Age'] - rec_1['Age']) + (sold_2['Age'] - rec_2['Age'])) / 2
    if age_diff >= 0:
        score += age_diff * 5   # Reward younger
    elif age_diff >= -3:
        score += age_diff * 2   # Small penalty
    else:
        score += (age_diff + 3) * 5 - 6  # Big penalty
    
    # 3. Budget usage (should use 40-85%)
    budget = (sold_1['current_mv'] + sold_2['current_mv']) * 0.8
    cost = rec_1['current_mv'] + rec_2['current_mv']
    budget_pct = (cost / budget) * 100
    
    if cost > budget:
        score -= 100  # Invalid
    elif 40 <= budget_pct <= 85:
        score += 20
    elif budget_pct < 40:
        score += budget_pct / 2  # Too cheap is suspicious
    else:
        score += max(20 - (budget_pct - 85), 0)
    
    # 4. Value for money
    sold_vfm = (sold_1['OVR'] + sold_2['OVR']) / ((sold_1['current_mv'] + sold_2['current_mv']) / 1e6)
    rec_vfm = (rec_1['OVR'] + rec_2['OVR']) / ((rec_1['current_mv'] + rec_2['current_mv']) / 1e6)
    
    if rec_vfm > sold_vfm * 1.2:
        score += 15  # Great value
    elif rec_vfm > sold_vfm:
        score += 10
    
    # 5. Diversity penalty (don't always pick same pattern)
    # This is implicit in having diverse scenarios
    
    return score


# ========== SIMPLIFIED RECOMMENDER FOR OPTIMIZATION ==========
def player_recommender_optimizable(sold_id_1, sold_id_2, df, 
                                   sim_w, qual_w, youth_w, fin_w,
                                   qual_scale, youth_scale, fin_vfm_w,
                                   verbose=False):
    """
    Streamlined recommender that accepts weight parameters.
    """
    
    df = df.copy()
    
    # Data prep (same as before)
    if 'Age' in df.columns:
        df['Age'] = df['Age'].fillna(df['Age'].median())
    if 'Alternative positions' in df.columns:
        df['Alternative positions'] = df['Alternative positions'].fillna('')
    
    gk_cols = ['GK Diving', 'GK Handling', 'GK Kicking', 'GK Positioning', 'GK Reflexes']
    for col in gk_cols:
        if col in df.columns:
            df[col] = df[col].fillna(0)
    
    df['GK_Overall'] = df[gk_cols].mean(axis=1)
    df['DEF_Composite'] = (df['Interceptions'] + df['Def Awareness'] + 
                           df['Standing Tackle'] + df['Sliding Tackle']) / 4
    
    def get_position_category(position):
        if pd.isna(position) or position == '':
            return 'Midfielder'
        position = str(position).upper()
        if 'GK' in position:
            return 'Goalkeeper'
        elif any(x in position for x in ['CB', 'LB', 'RB', 'RWB', 'LWB']):
            return 'Defender'
        elif any(x in position for x in ['CDM', 'CM', 'CAM', 'LM', 'RM']):
            return 'Midfielder'
        elif any(x in position for x in ['ST', 'CF', 'LW', 'RW']):
            return 'Attacker'
        else:
            return 'Midfielder'
    
    df['Position_Category'] = df['Position'].apply(get_position_category)
    
    sold_1 = df[df['player_id'] == sold_id_1].iloc[0]
    sold_2 = df[df['player_id'] == sold_id_2].iloc[0]
    
    budget = (sold_1['current_mv'] + sold_2['current_mv']) * 0.8
    
    team_remaining = df[(df['player_id'] != sold_id_1) & (df['player_id'] != sold_id_2)]
    current_team_ovr = team_remaining['OVR'].mean()
    current_team_age = team_remaining['Age'].mean()
    
    # Position matching
    def can_play_position(player_row, target_position):
        main_pos = str(player_row['Position']).upper()
        alt_pos = str(player_row['Alternative positions']).upper()
        target = str(target_position).upper()
        return target in main_pos or target in alt_pos
    
    def get_eligible_candidates(sold_player):
        candidates = df[
            (df['player_id'] != sold_id_1) &
            (df['player_id'] != sold_id_2) &
            (df['Position_Category'] == sold_player['Position_Category'])
        ].copy()
        
        pos_match = candidates.apply(
            lambda row: can_play_position(row, sold_player['Position']), 
            axis=1
        )
        return candidates[pos_match]
    
    candidates_1 = get_eligible_candidates(sold_1)
    candidates_2 = get_eligible_candidates(sold_2)
    
    # Feature weights
    FEATURE_WEIGHTS = {
        'Goalkeeper': {'GK_Overall': 0.45, 'Reactions': 0.20, 'Positioning': 0.15,
                      'height': 0.10, 'Composure': 0.05, 'OVR': 0.05},
        'Defender': {'DEF_Composite': 0.25, 'DEF': 0.20, 'PHY': 0.18,
                    'Reactions': 0.12, 'Heading Accuracy': 0.10, 'Strength': 0.08, 'OVR': 0.07},
        'Midfielder': {'PAS': 0.22, 'Vision': 0.15, 'DRI': 0.15, 'Short Passing': 0.12,
                      'Ball Control': 0.10, 'Reactions': 0.10, 'Stamina': 0.08, 'OVR': 0.08},
        'Attacker': {'SHO': 0.22, 'DRI': 0.18, 'PAC': 0.15, 'Finishing': 0.13,
                    'Positioning': 0.12, 'Reactions': 0.10, 'Acceleration': 0.05, 'OVR': 0.05}
    }
    
    # Similarity calculation
    def calculate_similarity(sold_player, candidates):
        pos_cat = sold_player['Position_Category']
        features = list(FEATURE_WEIGHTS[pos_cat].keys())
        weights = np.array([FEATURE_WEIGHTS[pos_cat][f] for f in features])
        
        sold_features = sold_player[features].values.reshape(1, -1)
        candidate_features = candidates[features].values
        
        scaler = StandardScaler()
        all_features = np.vstack([sold_features, candidate_features])
        all_features_scaled = scaler.fit_transform(all_features)
        
        sold_scaled = all_features_scaled[0:1]
        candidates_scaled = all_features_scaled[1:]
        
        sold_weighted = sold_scaled * weights
        candidates_weighted = candidates_scaled * weights
        
        similarities = cosine_similarity(sold_weighted, candidates_weighted)[0]
        return similarities
    
    candidates_1['similarity_score'] = calculate_similarity(sold_1, candidates_1)
    candidates_2['similarity_score'] = calculate_similarity(sold_2, candidates_2)
    
    # Filtering
    candidates_1_filtered = candidates_1[
        (candidates_1['OVR'] >= sold_1['OVR'] - 3) &
        (candidates_1['OVR'] <= sold_1['OVR'] + 10) &
        (candidates_1['Age'] <= sold_1['Age'] + 8) &
        (candidates_1['similarity_score'] > -0.3)
    ].nlargest(min(100, len(candidates_1)), 'similarity_score')
    
    candidates_2_filtered = candidates_2[
        (candidates_2['OVR'] >= sold_2['OVR'] - 3) &
        (candidates_2['OVR'] <= sold_2['OVR'] + 10) &
        (candidates_2['Age'] <= sold_2['Age'] + 8) &
        (candidates_2['similarity_score'] > -0.3)
    ].nlargest(min(100, len(candidates_2)), 'similarity_score')
    
    if len(candidates_1_filtered) < 5:
        candidates_1_filtered = candidates_1.nlargest(50, 'similarity_score')
    if len(candidates_2_filtered) < 5:
        candidates_2_filtered = candidates_2.nlargest(50, 'similarity_score')
    
    # Scoring function with parameterized weights
    def calculate_score(p1, p2):
        total_cost = p1['current_mv'] + p2['current_mv']
        if total_cost > budget:
            return None
        
        # Similarity
        sim = ((p1['similarity_score'] + p2['similarity_score']) / 2 + 1) / 2 * 100
        
        # Quality
        ovr_diff = ((p1['OVR'] - sold_1['OVR']) + (p2['OVR'] - sold_2['OVR'])) / 2
        qual = np.clip(ovr_diff * qual_scale, -100, 100)
        
        # Youth
        age_diff = ((sold_1['Age'] - p1['Age']) + (sold_2['Age'] - p2['Age'])) / 2
        youth = np.clip(age_diff * youth_scale, -100, 100)
        
        # Financial
        vfm = ((p1['OVR'] + p2['OVR']) / 2) / max((total_cost / 1e6) / 2, 0.1)
        fin = min(vfm * fin_vfm_w, 100)
        
        # Combined
        score = sim * sim_w + qual * qual_w + youth * youth_w + fin * fin_w
        
        return {
            'score': score,
            'player1_id': int(p1['player_id']),
            'player2_id': int(p2['player_id'])
        }
    
    # Find best pair
    best = None
    best_score = -999999
    
    for _, c1 in candidates_1_filtered.iterrows():
        for _, c2 in candidates_2_filtered.iterrows():
            result = calculate_score(c1, c2)
            if result and result['score'] > best_score:
                best_score = result['score']
                best = result
    
    return best


# ========== GENETIC ALGORITHM ==========
def genetic_algorithm_optimize(df, population_size=50, generations=20, scenarios=None):
    """
    Use Genetic Algorithm to find optimal weights.
    """
    
    if scenarios is None:
        print("Creating test scenarios...")
        scenarios = create_test_scenarios(df)
        print(f"Created {len(scenarios)} test scenarios")
    
    # Initialize population
    # Genes: [sim_w, qual_w, youth_w, fin_w, qual_scale, youth_scale, fin_vfm_w]
    population = []
    for _ in range(population_size):
        genes = [
            np.random.uniform(0.1, 0.4),   # sim_w
            np.random.uniform(0.2, 0.5),   # qual_w
            np.random.uniform(0.1, 0.3),   # youth_w
            np.random.uniform(0.1, 0.3),   # fin_w
            np.random.uniform(10, 30),     # qual_scale
            np.random.uniform(8, 20),      # youth_scale
            np.random.uniform(1, 5)        # fin_vfm_w
        ]
        population.append(genes)
    
    print(f"\n{'='*80}")
    print(f"GENETIC ALGORITHM OPTIMIZATION")
    print(f"{'='*80}")
    print(f"Population size: {population_size}")
    print(f"Generations: {generations}")
    print(f"Test scenarios: {len(scenarios)}")
    print(f"{'='*80}\n")
    
    best_overall = None
    best_overall_score = -999999
    
    for gen in range(generations):
        # Evaluate fitness
        fitness_scores = []
        for individual in population:
            score = evaluate_weights(individual, df, scenarios)
            fitness_scores.append(score)
        
        # Track best
        best_idx = np.argmax(fitness_scores)
        if fitness_scores[best_idx] > best_overall_score:
            best_overall_score = fitness_scores[best_idx]
            best_overall = population[best_idx].copy()
        
        print(f"Gen {gen+1}/{generations}: Best={fitness_scores[best_idx]:.2f}, "
              f"Avg={np.mean(fitness_scores):.2f}, Overall Best={best_overall_score:.2f}")
        
        # Selection (tournament)
        selected = []
        for _ in range(population_size // 2):
            tournament = np.random.choice(len(population), 3, replace=False)
            winner = tournament[np.argmax([fitness_scores[i] for i in tournament])]
            selected.append(population[winner])
        
        # Crossover
        offspring = []
        for i in range(0, len(selected), 2):
            if i + 1 < len(selected):
                parent1 = selected[i]
                parent2 = selected[i+1]
                
                # Single-point crossover
                point = np.random.randint(1, len(parent1))
                child1 = parent1[:point] + parent2[point:]
                child2 = parent2[:point] + parent1[point:]
                
                offspring.extend([child1, child2])
        
        # Mutation
        for individual in offspring:
            if np.random.random() < 0.2:  # 20% mutation rate
                gene_idx = np.random.randint(len(individual))
                if gene_idx < 4:  # Weights
                    individual[gene_idx] *= np.random.uniform(0.8, 1.2)
                    individual[gene_idx] = np.clip(individual[gene_idx], 0.05, 0.6)
                elif gene_idx == 4:  # qual_scale
                    individual[gene_idx] *= np.random.uniform(0.8, 1.2)
                    individual[gene_idx] = np.clip(individual[gene_idx], 5, 40)
                elif gene_idx == 5:  # youth_scale
                    individual[gene_idx] *= np.random.uniform(0.8, 1.2)
                    individual[gene_idx] = np.clip(individual[gene_idx], 5, 25)
                else:  # fin_vfm_w
                    individual[gene_idx] *= np.random.uniform(0.8, 1.2)
                    individual[gene_idx] = np.clip(individual[gene_idx], 0.5, 8)
        
        # Elitism (keep best 20%)
        elite_count = population_size // 5
        elite_indices = np.argsort(fitness_scores)[-elite_count:]
        elite = [population[i] for i in elite_indices]
        
        # New population
        population = elite + offspring[:population_size - elite_count]
    
    print(f"\n{'='*80}")
    print("OPTIMIZATION COMPLETE!")
    print(f"{'='*80}")
    print(f"Best score: {best_overall_score:.2f}")
    print(f"\nOptimal weights:")
    
    # Normalize
    total = sum(best_overall[:4])
    norm_weights = [w/total for w in best_overall[:4]]
    
    print(f"  Similarity:  {norm_weights[0]:.3f}")
    print(f"  Quality:     {norm_weights[1]:.3f}")
    print(f"  Youth:       {norm_weights[2]:.3f}")
    print(f"  Financial:   {norm_weights[3]:.3f}")
    print(f"\nScaling parameters:")
    print(f"  Quality scale:   {best_overall[4]:.2f}")
    print(f"  Youth scale:     {best_overall[5]:.2f}")
    print(f"  Financial VFM:   {best_overall[6]:.2f}")
    print(f"{'='*80}\n")
    
    return best_overall, norm_weights


# ========== RUN OPTIMIZATION ==========
print("🧬 Starting Genetic Algorithm Optimization...")
print("This may take 5-10 minutes...\n")

optimal_params, optimal_weights = genetic_algorithm_optimize(
    df, 
    population_size=30,  # Smaller for speed
    generations=15
)

print("\n✅ Optimization complete!")
print("Use these weights in your final recommender function.")
```
