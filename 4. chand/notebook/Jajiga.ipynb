```python
import pandas as pd
import numpy as np

# بارگذاری دیتا
train = pd.read_csv(r"/content/drive/MyDrive/Oltmpic 2024/chand/data/train.csv")
test = pd.read_csv(r"/content/drive/MyDrive/Oltmpic 2024/chand/data/test.csv")

# نمایش اطلاعات کلی
print("=" * 50)
print("TRAIN DATASET INFO")
print("=" * 50)
print(f"Shape: {train.shape}")
print(f"\nColumns: {train.columns.tolist()}")
print(f"\nFirst 3 rows:")
print(train.head(3))

print("\n" + "=" * 50)
print("TARGET VARIABLE (min_price) STATISTICS")
print("=" * 50)
print(train['min_price'].describe())

print("\n" + "=" * 50)
print("MISSING VALUES IN TRAIN")
print("=" * 50)
missing_train = train.isnull().sum()
missing_train = missing_train[missing_train > 0].sort_values(ascending=False)
print(missing_train)

print("\n" + "=" * 50)
print("MISSING VALUES IN TEST")
print("=" * 50)
missing_test = test.isnull().sum()
missing_test = missing_test[missing_test > 0].sort_values(ascending=False)
print(missing_test)

print("\n" + "=" * 50)
print("DATA TYPES")
print("=" * 50)
print(train.dtypes.value_counts())

# بررسی دقیق‌تر برای یافتن data leakage
print("\n" + "=" * 50)
print("CHECKING FOR POTENTIAL LEAKAGE")
print("=" * 50)
print(f"Unique cities in train: {train['city.name'].nunique()}")
print(f"Unique cities in test: {test['city.name'].nunique()}")
print(f"Cities in test not in train: {set(test['city.name'].unique()) - set(train['city.name'].unique())}")
```

```python
print("="*50)
print("STEP 1: INITIAL CLEANING")
print("="*50)

# کپی از دیتاست‌ها
train_clean = train.copy()
test_clean = test.copy()

# 1. حذف ردیف با قیمت صفر
print(f"Rows before removing zero prices: {len(train_clean)}")
train_clean = train_clean[train_clean['min_price'] > 0]
print(f"Rows after removing zero prices: {len(train_clean)}")

# 2. حذف ستون‌های غیرضروری
cols_to_drop = [
    'Unnamed: 0', 'id', 'status',  # شناسایی و وضعیت
    'province.name', 'province.url',  # تکراری با province.id
    'city.name', 'city.url',  # تکراری با city.id
    'title', 'description',  # متن‌های طولانی (می‌تونیم بعدا فیچر ازشون بگیریم)
    'additional_safety',  # Missing value خیلی زیاد (99.5%)
    'stays_max',  # Missing value خیلی زیاد (97.8%)
    'sleep_description'  # Missing value زیاد (64.7%)
]

print(f"\nColumns before drop: {train_clean.shape[1]}")
train_clean = train_clean.drop(columns=cols_to_drop)
test_clean = test_clean.drop(columns=[col for col in cols_to_drop if col in test_clean.columns])
print(f"Columns after drop: {train_clean.shape[1]}")

print(f"\nRemaining columns: {train_clean.columns.tolist()}")
print(f"\nTrain shape: {train_clean.shape}")
print(f"Test shape: {test_clean.shape}")
```

```python
import ast

print("="*50)
print("STEP 2: FEATURE ENGINEERING - LIST COLUMNS")
print("="*50)

def extract_list_features(df, col_name):
    """استخراج فیچر از ستون‌های لیستی"""

    # تبدیل string به list
    df[col_name] = df[col_name].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else [])

    # تعداد آیتم‌ها
    df[f'{col_name}_count'] = df[col_name].apply(len)

    # گرفتن تمام آیتم‌های یونیک
    all_items = set()
    for items in df[col_name]:
        all_items.update(items)

    print(f"\n{col_name}:")
    print(f"  Unique items: {sorted(all_items)}")
    print(f"  Count range: {df[f'{col_name}_count'].min()} to {df[f'{col_name}_count'].max()}")

    # ساخت binary features برای هر آیتم
    for item in all_items:
        df[f'{col_name}_{item}'] = df[col_name].apply(lambda x: 1 if item in x else 0)

    return df, all_items

# پردازش برای train
train_clean, types_items = extract_list_features(train_clean, 'types')
train_clean, regions_items = extract_list_features(train_clean, 'regions')
train_clean, rules_items = extract_list_features(train_clean, 'rules')

# پردازش برای test
test_clean, _ = extract_list_features(test_clean, 'types')
test_clean, _ = extract_list_features(test_clean, 'regions')
test_clean, _ = extract_list_features(test_clean, 'rules')

print(f"\n✅ Features added!")
print(f"Train shape: {train_clean.shape}")
print(f"Test shape: {test_clean.shape}")
```

```python
import ast

print("="*50)
print("STEP 3: FEATURE ENGINEERING - SLEEP_ARRANGE")
print("="*50)

def extract_sleep_features(df):
    singles = []
    doubles = []
    mattresses = []

    for row in df['sleep_arrange']:
        if pd.isnull(row):  # اگر مقدار NaN بود
            singles.append(0)
            doubles.append(0)
            mattresses.append(0)
        else:
            try:
                arr = ast.literal_eval(row)
            except Exception:
                arr = []
            singles.append(sum(item.get('single',0) for item in arr))
            doubles.append(sum(item.get('double',0) for item in arr))
            mattresses.append(sum(item.get('mattress',0) for item in arr))

    df['sleep_single_count'] = singles
    df['sleep_double_count'] = doubles
    df['sleep_mattress_count'] = mattresses

    return df

train_clean = extract_sleep_features(train_clean)
test_clean = extract_sleep_features(test_clean)

print("Sleep features added! Example:")
print(train_clean[['sleep_single_count', 'sleep_double_count', 'sleep_mattress_count']].head(3))
```

```python
print("="*50)
print("STEP 4: FEATURE ENGINEERING - FEATURES")
print("="*50)

def extract_features_names(df):
    all_features = set()
    for row in df['features'].dropna():
        try:
            feature_list = ast.literal_eval(row)
            names = [item.get('name') for item in feature_list if 'name' in item]
            all_features.update(names)
        except Exception:
            continue

    print(f"Unique feature names: {sorted(all_features)}")

    # ساخت فیچر باینری برای هر name
    for name in all_features:
        df[f'feature_{name}'] = df['features'].apply(lambda x: int(name in [item.get('name') for item in ast.literal_eval(x)] if pd.notnull(x) else []) )

    return df, all_features

train_clean, feature_names = extract_features_names(train_clean)
test_clean, _ = extract_features_names(test_clean)

print(f"Features (binary columns) added! Example columns:")
feat_cols = [f'feature_{name}' for name in list(feature_names)[:5]]
print(train_clean[feat_cols].head())
```

```python
print("="*50)
print("STEP 5: BOOLEAN COLUMNS ANALYSIS")
print("="*50)

bool_cols = train_clean.select_dtypes(include='bool').columns.tolist()
print(f"Boolean columns: {bool_cols}\n")

# بررسی توزیع و تاثیر روی قیمت
for col in bool_cols:
    print(f"\n--- {col} ---")
    print(f"Distribution:")
    print(f"  True:  {train_clean[col].sum():4d} ({train_clean[col].sum()/len(train_clean)*100:5.1f}%)")
    print(f"  False: {(~train_clean[col]).sum():4d} ({(~train_clean[col]).sum()/len(train_clean)*100:5.1f}%)")

    # میانگین قیمت
    if train_clean[col].sum() > 0:
        mean_true = train_clean[train_clean[col]]['min_price'].mean()
        mean_false = train_clean[~train_clean[col]]['min_price'].mean()
        diff = ((mean_true - mean_false) / mean_false * 100)
        print(f"  Mean price (True):  {mean_true:,.0f}")
        print(f"  Mean price (False): {mean_false:,.0f}")
        print(f"  Difference: {diff:+.1f}%")

print("\n✅ Boolean columns are ready (no preprocessing needed)")
```

```python
print("="*50)
print("STEP 6: DROP is_clean & ANALYZE FLOAT COLUMNS")
print("="*50)

# حذف is_clean
train_clean = train_clean.drop(columns=['is_clean'])
test_clean = test_clean.drop(columns=['is_clean'])
print("✅ is_clean dropped\n")

# بررسی ستون‌های float
float_cols = train_clean.select_dtypes(include='float64').columns.tolist()
print(f"Float columns ({len(float_cols)}): {float_cols}\n")

for col in float_cols:
    print(f"\n--- {col} ---")
    print(f"Missing values: {train_clean[col].isnull().sum()} ({train_clean[col].isnull().sum()/len(train_clean)*100:.1f}%)")
    print(f"Stats:")
    print(f"  Min:    {train_clean[col].min()}")
    print(f"  Mean:   {train_clean[col].mean():.2f}")
    print(f"  Median: {train_clean[col].median():.2f}")
    print(f"  Max:    {train_clean[col].max()}")
    print(f"  Std:    {train_clean[col].std():.2f}")

    # Correlation با target
    if col != 'min_price':
        corr = train_clean[[col, 'min_price']].corr().iloc[0,1]
        print(f"  Correlation with target: {corr:.3f}")

print(f"\n✅ Float columns analyzed")
```

```python
print("="*50)
print("STEP 7: RATINGS ANALYSIS & FEATURE ENGINEERING")
print("="*50)

# بررسی اقامتگاه‌های بدون rating
has_rating = (train_clean['ratings.count'] > 0).astype(int)
no_rating = (train_clean['ratings.count'] == 0).astype(int)

print(f"اقامتگاه‌های با rating: {has_rating.sum()} ({has_rating.sum()/len(train_clean)*100:.1f}%)")
print(f"اقامتگاه‌های بدون rating: {no_rating.sum()} ({no_rating.sum()/len(train_clean)*100:.1f}%)")

# مقایسه قیمت
price_with_rating = train_clean[train_clean['ratings.count'] > 0]['min_price'].mean()
price_without_rating = train_clean[train_clean['ratings.count'] == 0]['min_price'].mean()
diff = ((price_with_rating - price_without_rating) / price_without_rating * 100)

print(f"\nمیانگین قیمت با rating: {price_with_rating:,.0f}")
print(f"میانگین قیمت بدون rating: {price_without_rating:,.0f}")
print(f"تفاوت: {diff:+.1f}%")

# بررسی تعداد صفرها در هر rating
rating_cols = ['ratings.total', 'ratings.cleanliness', 'ratings.location',
               'ratings.checkin', 'ratings.value']

print("\n--- تعداد صفرها در هر rating ---")
for col in rating_cols:
    zero_count = (train_clean[col] == 0).sum()
    print(f"{col:25s}: {zero_count:4d} ({zero_count/len(train_clean)*100:5.1f}%)")

# ساخت فیچر جدید
print("\n--- ساخت فیچرهای جدید ---")

# 1. آیا rating دارد؟
train_clean['has_rating'] = (train_clean['ratings.count'] > 0).astype(int)
test_clean['has_rating'] = (test_clean['ratings.count'] > 0).astype(int)
print("✅ has_rating created")

# 2. میانگین کل ratingها (فقط برای اقامتگاه‌های با rating)
train_clean['ratings_mean'] = train_clean[rating_cols].replace(0, np.nan).mean(axis=1).fillna(0)
test_clean['ratings_mean'] = test_clean[rating_cols].replace(0, np.nan).mean(axis=1).fillna(0)
print("✅ ratings_mean created")

# 3. تعداد رتینگ‌های داده شده
train_clean['ratings_given_count'] = (train_clean[rating_cols] > 0).sum(axis=1)
test_clean['ratings_given_count'] = (test_clean[rating_cols] > 0).sum(axis=1)
print("✅ ratings_given_count created")

print(f"\n🎯 New features summary:")
print(train_clean[['has_rating', 'ratings_mean', 'ratings_given_count', 'ratings.count']].describe())
```

```python
print("="*50)
print("STEP 8: INTEGER COLUMNS ANALYSIS")
print("="*50)

# گرفتن ستون‌های int (بدون target)
int_cols = train_clean.select_dtypes(include=['int64', 'int32']).columns.tolist()
int_cols = [col for col in int_cols if col != 'min_price' and not col.startswith(('types_', 'regions_', 'rules_', 'feature_', 'sleep_', 'has_', 'ratings_'))]

print(f"Integer columns to analyze ({len(int_cols)}):")
print(int_cols)
print()

# بررسی هر ستون
for col in int_cols:
    print(f"\n--- {col} ---")
    print(f"Missing: {train_clean[col].isnull().sum()} ({train_clean[col].isnull().sum()/len(train_clean)*100:.1f}%)")
    print(f"Unique values: {train_clean[col].nunique()}")
    print(f"Range: [{train_clean[col].min()}, {train_clean[col].max()}]")
    print(f"Mean: {train_clean[col].mean():.2f}, Median: {train_clean[col].median():.0f}")

    # Correlation با target
    corr = train_clean[[col, 'min_price']].corr().iloc[0,1]
    print(f"Correlation with target: {corr:.3f}")

    # نمونه مقادیر
    if train_clean[col].nunique() < 20:
        print(f"Value counts:\n{train_clean[col].value_counts().head(10)}")

print("\n✅ Integer columns analyzed")
```

```python
print("="*50)
print("STEP 9: INTEGER COLUMNS PREPROCESSING")
print("="*50)

# 1. حذف floors_count (همه صفر هستند)
print("\n1. حذف floors_count...")
train_clean = train_clean.drop(columns=['floors_count'])
test_clean = test_clean.drop(columns=['floors_count'])
print("✅ floors_count dropped")

# 2. Handle کردن outliers در land_area
print("\n2. پردازش outliers در land_area...")
print(f"   Before: max={train_clean['land_area'].max()}")
# مقادیر بیش از 10000 احتمالاً خطای ثبت است - کلیپ کنیم
train_clean['land_area'] = train_clean['land_area'].clip(upper=10000)
test_clean['land_area'] = test_clean['land_area'].clip(upper=10000)
print(f"   After: max={train_clean['land_area'].max()}")
print("✅ land_area outliers handled")

# 3. ساخت فیچر ترکیبی: نسبت مساحت به تعداد نفر
print("\n3. ساخت فیچرهای ترکیبی...")
train_clean['area_per_guest'] = train_clean['floor_area'] / (train_clean['guest_number'] + 1)
test_clean['area_per_guest'] = test_clean['floor_area'] / (test_clean['guest_number'] + 1)
print("✅ area_per_guest created")

# 4. ساخت فیچر: نسبت تخت به مهمان
train_clean['bedroom_per_guest'] = train_clean['bedrooms'] / (train_clean['guest_number'] + 1)
test_clean['bedroom_per_guest'] = test_clean['bedrooms'] / (test_clean['guest_number'] + 1)
print("✅ bedroom_per_guest created")

# 5. ساخت فیچر: آیا اقامتگاه چند واحده است؟
train_clean['is_multi_unit'] = (train_clean['units_count'] > 1).astype(int)
test_clean['is_multi_unit'] = (test_clean['units_count'] > 1).astype(int)
print("✅ is_multi_unit created")

# 6. تبدیل success_books به دسته‌های معنی‌دار
def categorize_success(x):
    if x == 0:
        return 0  # بدون رزرو موفق
    elif x <= 10:
        return 1  # کم
    elif x <= 50:
        return 2  # متوسط
    else:
        return 3  # زیاد

train_clean['success_category'] = train_clean['success_books'].apply(categorize_success)
test_clean['success_category'] = test_clean['success_books'].apply(categorize_success)
print("✅ success_category created")

print(f"\n✅ Preprocessing completed!")
print(f"Train shape: {train_clean.shape}")
print(f"Test shape: {test_clean.shape}")
```

```python
print("="*50)
print("STEP 10: CATEGORICAL (OBJECT) COLUMNS ANALYSIS")
print("="*50)

# گرفتن ستون‌های object که هنوز پردازش نشده‌اند
object_cols = train_clean.select_dtypes(include='object').columns.tolist()

# فیلتر کردن ستون‌هایی که قبلاً پردازش شدند
processed_cols = ['types', 'regions', 'rules', 'features', 'sleep_arrange']
object_cols = [col for col in object_cols if col not in processed_cols]

print(f"Remaining object columns ({len(object_cols)}):")
print(object_cols)
print()

# بررسی هر ستون
for col in object_cols:
    print(f"\n{'='*40}")
    print(f"--- {col} ---")
    print(f"{'='*40}")

    # اطلاعات کلی
    print(f"Unique values: {train_clean[col].nunique()}")
    print(f"Missing values: {train_clean[col].isnull().sum()} ({train_clean[col].isnull().sum()/len(train_clean)*100:.1f}%)")

    # نمونه مقادیر
    print(f"\nTop 10 most common values:")
    print(train_clean[col].value_counts().head(10))

    # بررسی correlation با قیمت (اگر تعداد unique کم باشد)
    if train_clean[col].nunique() < 100:
        print(f"\nMean price by category:")
        price_by_cat = train_clean.groupby(col)['min_price'].agg(['mean', 'count']).sort_values('mean', ascending=False)
        print(price_by_cat.head(10))

print("\n" + "="*50)
print("✅ Categorical columns analyzed")
print("="*50)
```

```python
print("="*50)
print("STEP 11: CATEGORICAL PREPROCESSING")
print("="*50)

# 1. حذف allocation_encoded و province_price_level قبلی
if 'allocation_encoded' in train_clean.columns:
    train_clean = train_clean.drop(columns=['allocation_encoded'])
    test_clean = test_clean.drop(columns=['allocation_encoded'])
    print("⚠️ allocation_encoded removed (had data leakage)")

if 'province_price_level' in train_clean.columns:
    train_clean = train_clean.drop(columns=['province_price_level'])
    test_clean = test_clean.drop(columns=['province_price_level'])
    print("⚠️ province_price_level removed (had data leakage)")

# 2. One-Hot Encoding برای allocation (بدون استفاده از target)
allocation_dummies = pd.get_dummies(train_clean['allocation'], prefix='allocation', drop_first=True)
train_clean = pd.concat([train_clean, allocation_dummies], axis=1)

allocation_dummies_test = pd.get_dummies(test_clean['allocation'], prefix='allocation', drop_first=True)
test_clean = pd.concat([test_clean, allocation_dummies_test], axis=1)
print("✅ allocation One-Hot encoded (no leakage)")

# 3. One-Hot Encoding برای province.id (بدون استفاده از target)
province_dummies = pd.get_dummies(train_clean['province.id'], prefix='province', drop_first=True)
train_clean = pd.concat([train_clean, province_dummies], axis=1)

province_dummies_test = pd.get_dummies(test_clean['province.id'], prefix='province', drop_first=True)
test_clean = pd.concat([test_clean, province_dummies_test], axis=1)
print("✅ province.id One-Hot encoded (no leakage)")

# 4. حذف ستون‌های اصلی
train_clean = train_clean.drop(columns=['allocation', 'province.id'])
test_clean = test_clean.drop(columns=['allocation', 'province.id'])
print("✅ Original categorical columns dropped")

print(f"\n✅ Fixed! No data leakage")
print(f"Train shape: {train_clean.shape}")
print(f"Test shape: {test_clean.shape}")
```

```python
print("="*50)
print("FINDING EXTRA COLUMN")
print("="*50)

train_cols = set(train_clean.columns)
test_cols = set(test_clean.columns)

extra_in_train = train_cols - test_cols
extra_in_test = test_cols - train_cols

print(f"\nColumns in TRAIN but not in TEST:")
print(extra_in_train)

print(f"\nColumns in TEST but not in TRAIN:")
print(extra_in_test)
```

```python
print("="*50)
print("STEP 12: VERIFICATION & CLEANUP")
print("="*50)

# بررسی ستون‌های باقیمانده
print("\nستون‌های Object باقیمانده:")
object_cols = train_clean.select_dtypes(include='object').columns.tolist()
print(object_cols)

# ستون‌هایی که باید حذف شده باشند
should_be_removed = ['types', 'regions', 'rules', 'features', 'sleep_arrange', 'additional_feature', 'additional_rule']
remaining = [col for col in should_be_removed if col in train_clean.columns]

if remaining:
    print(f"\n⚠️ این ستون‌ها هنوز باقی‌مانده‌اند: {remaining}")
    print("حذف می‌کنیم...")
    train_clean = train_clean.drop(columns=remaining)
    test_clean = test_clean.drop(columns=[col for col in remaining if col in test_clean.columns])
    print("✅ حذف شدند")
else:
    print("\n✅ همه ستون‌های اولیه حذف شده‌اند")

# رفع مشکل feature_gym
print("\n3. رفع مشکل feature_gym...")
if 'feature_gym' in train_clean.columns and 'feature_gym' not in test_clean.columns:
    print("⚠️ feature_gym در test وجود ندارد")
    print("راه حل: اضافه کردن به test با مقدار 0")
    test_clean['feature_gym'] = 0
    print("✅ feature_gym به test اضافه شد")

# همگام‌سازی ستون‌ها
print("\n4. همگام‌سازی ستون‌های train و test...")
train_cols = set(train_clean.columns) - {'min_price'}
test_cols = set(test_clean.columns)

missing_in_test = train_cols - test_cols
missing_in_train = test_cols - train_cols

print(f"ستون‌های موجود در train ولی نه در test: {missing_in_test}")
print(f"ستون‌های موجود در test ولی نه در train: {missing_in_train}")

# اضافه کردن ستون‌های گمشده
for col in missing_in_test:
    test_clean[col] = 0
    print(f"  ✅ {col} به test اضافه شد")

for col in missing_in_train:
    train_clean[col] = 0
    print(f"  ✅ {col} به train اضافه شد")

print(f"\n✅ همگام‌سازی کامل شد!")
print(f"Train shape: {train_clean.shape}")
print(f"Test shape: {test_clean.shape}")
print(f"تفاوت ستون‌ها: {train_clean.shape[1] - test_clean.shape[1]} (باید 1 باشد برای min_price)")
```

```python
print("="*50)
print("STEP 13: FINAL DATA TYPE CHECK & NORMALIZATION")
print("="*50)

# 1. بررسی انواع داده
print("\n1. بررسی انواع داده در train...")
print(train_clean.dtypes.value_counts())

# بررسی ستون‌های object باقیمانده
object_cols = train_clean.select_dtypes(include='object').columns.tolist()
if object_cols:
    print(f"\n⚠️ ستون‌های object باقیمانده: {object_cols}")
    print("این ستون‌ها باید پردازش شوند!")
else:
    print("\n✅ هیچ ستون object باقی نمانده")

# 2. جدا کردن X و y
print("\n2. جداسازی Features و Target...")
X_train = train_clean.drop(columns=['min_price'])
y_train = train_clean['min_price']
X_test = test_clean.copy()

print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")

# 3. بررسی Missing Values
print("\n3. بررسی Missing Values...")
missing_train = X_train.isnull().sum().sum()
missing_test = X_test.isnull().sum().sum()
print(f"Total missing in X_train: {missing_train}")
print(f"Total missing in X_test: {missing_test}")

if missing_train > 0 or missing_test > 0:
    print("\n⚠️ Missing values detected! Filling with 0...")
    X_train = X_train.fillna(0)
    X_test = X_test.fillna(0)
    print("✅ Missing values filled")

# 4. تبدیل همه به numeric
print("\n4. تبدیل تمام ستون‌ها به numeric...")
print(f"Boolean columns: {X_train.select_dtypes(include='bool').columns.tolist()}")

# تبدیل bool به int
bool_cols = X_train.select_dtypes(include='bool').columns
if len(bool_cols) > 0:
    X_train[bool_cols] = X_train[bool_cols].astype(int)
    X_test[bool_cols] = X_test[bool_cols].astype(int)
    print(f"✅ {len(bool_cols)} boolean columns converted to int")

print("\nData types after conversion:")
print(X_train.dtypes.value_counts())

# 5. لیست فیچرهای نهایی
print(f"\n5. تعداد فیچرهای نهایی: {X_train.shape[1]}")
print(f"نمونه از فیچرها:")
print(X_train.columns.tolist()[:20])

print("\n✅ داده آماده مدل‌سازی است!")
print(f"Final shapes:")
print(f"  X_train: {X_train.shape}")
print(f"  y_train: {y_train.shape}")
print(f"  X_test: {X_test.shape}")
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder

print("="*50)
print("STEP 15: SCALING & FINAL ENCODING")
print("="*50)

# 1. Label Encoding برای city.id
print("\n1. Label Encoding city.id...")
if 'city.id' in X_train.columns:
    le = LabelEncoder()
    # Fit روی train و transform روی هر دو
    X_train['city.id_encoded'] = le.fit_transform(X_train['city.id'])
    # برای test: اگر شهر جدیدی بود، -1 بده
    X_test['city.id_encoded'] = X_test['city.id'].apply(
        lambda x: le.transform([x])[0] if x in le.classes_ else -1
    )
    # حذف ستون اصلی
    X_train = X_train.drop(columns=['city.id'])
    X_test = X_test.drop(columns=['city.id'])
    print(f"✅ city.id encoded: {len(le.classes_)} شهر")

# 2. شناسایی ستون‌های عددی برای scaling
print("\n2. شناسایی ستون‌های عددی برای scaling...")
numeric_cols = ['floor_area', 'land_area', 'bedrooms', 'guest_number',
                'max_guest_number', 'stays_min', 'entrance_time_min',
                'entrance_time_max', 'leaving_time', 'geo.lat', 'geo.lng',
                'ratings.count', 'ratings.total', 'ratings.cleanliness',
                'ratings.location', 'ratings.checkin', 'ratings.value',
                'units_count', 'success_books', 'area_per_guest',
                'bedroom_per_guest', 'ratings_mean', 'ratings_given_count',
                'sleep_single_count', 'sleep_double_count', 'sleep_mattress_count']

# فقط ستون‌هایی که واقعاً وجود دارند
numeric_cols = [col for col in numeric_cols if col in X_train.columns]
print(f"ستون‌های عددی: {len(numeric_cols)} عدد")

# 3. Scaling با StandardScaler
print("\n3. Scaling ستون‌های عددی...")
scaler = StandardScaler()
X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])
print("✅ Scaling completed (mean=0, std=1)")

# 4. بررسی نهایی
print("\n4. بررسی نهایی...")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"Data types:\n{X_train.dtypes.value_counts()}")
print(f"Missing values: {X_train.isnull().sum().sum()}")

# نمونه از داده scaled
print("\nنمونه از داده scaled:")
print(X_train[numeric_cols[:5]].head(3))

print("\n🎯 داده کاملاً آماده مدل‌سازی است!")
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score, KFold
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

print("="*50)
print("STEP 16: BASELINE MODEL - RANDOM FOREST")
print("="*50)

# 1. تنظیم Cross-Validation
print("\n1. تنظیم Cross-Validation...")
kfold = KFold(n_splits=5, shuffle=True, random_state=42)
print("✅ 5-Fold CV آماده شد")

# 2. مدل اولیه با hyperparameters پیش‌فرض
print("\n2. آموزش مدل Random Forest (baseline)...")
rf_baseline = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,  # استفاده از تمام CPU
    verbose=0
)

# 3. ارزیابی با Cross-Validation
print("\n3. ارزیابی با 5-Fold CV...")
cv_scores = cross_val_score(
    rf_baseline,
    X_train,
    y_train,
    cv=kfold,
    scoring='r2',
    n_jobs=-1
)

print(f"\nR² Scores برای هر fold:")
for i, score in enumerate(cv_scores, 1):
    print(f"  Fold {i}: {score:.4f}")

print(f"\n📊 نتایج Cross-Validation:")
print(f"  میانگین R²: {cv_scores.mean():.4f}")
print(f"  انحراف معیار: {cv_scores.std():.4f}")
print(f"  بازه R²: [{cv_scores.min():.4f}, {cv_scores.max():.4f}]")

# 4. آموزش روی کل train
print("\n4. آموزش نهایی روی کل train...")
rf_baseline.fit(X_train, y_train)
print("✅ مدل آموزش داده شد")

# 5. پیش‌بینی و ارزیابی روی train
print("\n5. ارزیابی روی train set...")
y_train_pred = rf_baseline.predict(X_train)
train_r2 = r2_score(y_train, y_train_pred)
train_mae = mean_absolute_error(y_train, y_train_pred)
train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))

print(f"  R² Score: {train_r2:.4f}")
print(f"  MAE: {train_mae:,.0f}")
print(f"  RMSE: {train_rmse:,.0f}")

# 6. بررسی feature importance
print("\n6. Top 10 مهم‌ترین فیچرها:")
feature_importance = pd.DataFrame({
    'feature': X_train.columns,
    'importance': rf_baseline.feature_importances_
}).sort_values('importance', ascending=False)

print(feature_importance.head(10).to_string(index=False))

# 7. وضعیت مدل
print("\n" + "="*50)
print("📌 خلاصه:")
print("="*50)
print(f"✅ CV R² Score: {cv_scores.mean():.4f} ± {cv_scores.std():.4f}")
print(f"✅ Train R² Score: {train_r2:.4f}")

if cv_scores.mean() >= 0.4:
    print(f"🎉 به حدنصاب رسیدیم! (حداقل 0.4 بود)")
    estimated_score = round(cv_scores.mean(), 3) * 250
    print(f"🏆 امتیاز تخمینی: {estimated_score:.1f} از 250")
else:
    print(f"⚠️ هنوز به حدنصاب نرسیدیم (نیاز: 0.4)")

print("\n✅ مدل baseline آماده است!")
```

```python
print("="*50)
print("STEP 17: OPTIMIZED RANDOM FOREST")
print("="*50)

# مدل بهینه‌شده با تنظیمات anti-overfitting
print("\n1. آموزش مدل با hyperparameters بهینه...")
rf_optimized = RandomForestRegressor(
    n_estimators=200,        # افزایش تعداد درخت
    max_depth=15,            # محدود کردن عمق (جلوگیری از overfitting)
    min_samples_split=10,    # حداقل نمونه برای split
    min_samples_leaf=5,      # حداقل نمونه در برگ
    max_features='sqrt',     # استفاده از sqrt فیچرها
    random_state=42,
    n_jobs=-1,
    verbose=0
)

# ارزیابی با CV
print("\n2. ارزیابی با Cross-Validation...")
cv_scores_opt = cross_val_score(
    rf_optimized,
    X_train,
    y_train,
    cv=kfold,
    scoring='r2',
    n_jobs=-1
)

print(f"\nR² Scores برای هر fold:")
for i, score in enumerate(cv_scores_opt, 1):
    print(f"  Fold {i}: {score:.4f}")

print(f"\n📊 نتایج Optimized Model:")
print(f"  میانگین R²: {cv_scores_opt.mean():.4f}")
print(f"  انحراف معیار: {cv_scores_opt.std():.4f}")

# آموزش نهایی
rf_optimized.fit(X_train, y_train)
y_train_pred_opt = rf_optimized.predict(X_train)
train_r2_opt = r2_score(y_train, y_train_pred_opt)

print(f"\n  Train R²: {train_r2_opt:.4f}")
print(f"  Gap (Train - CV): {train_r2_opt - cv_scores_opt.mean():.4f}")

# مقایسه
print("\n" + "="*50)
print("📊 مقایسه Baseline vs Optimized:")
print("="*50)
print(f"Baseline:  CV={cv_scores.mean():.4f}, Train={train_r2:.4f}, Gap={train_r2 - cv_scores.mean():.4f}")
print(f"Optimized: CV={cv_scores_opt.mean():.4f}, Train={train_r2_opt:.4f}, Gap={train_r2_opt - cv_scores_opt.mean():.4f}")

if cv_scores_opt.mean() > cv_scores.mean():
    print("\n🎉 Optimized model بهتر است!")
    estimated_score = round(cv_scores_opt.mean(), 3) * 250
    print(f"🏆 امتیاز تخمینی جدید: {estimated_score:.1f} از 250")
```

```python
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import cross_val_score
from sklearn.metrics import r2_score
import numpy as np
import pandas as pd

print("="*80)
print("COMPREHENSIVE MODEL COMPARISON WITH GPU ACCELERATION")
print("="*80)

# بررسی GPU
try:
    import torch
    gpu_available = torch.cuda.is_available()
    print(f"\n🔍 GPU Status: {'✅ Available' if gpu_available else '❌ Not Available'}")
    if gpu_available:
        print(f"   Device: {torch.cuda.get_device_name(0)}")
except:
    gpu_available = False
    print("\n🔍 GPU Status: ❌ Not Available")

print("\n" + "="*80)

# Dictionary برای ذخیره مدل‌ها
models = {}
cv_results = {}

# ─────────────────────────────────────────────────────────────
# 1. XGBoost (GPU support - XGBoost 3.1+)
# ─────────────────────────────────────────────────────────────
print("\n1️⃣ Training XGBoost...")
try:
    xgb_params = {
        'n_estimators': 200,
        'max_depth': 6,
        'learning_rate': 0.05,
        'subsample': 0.8,
        'colsample_bytree': 0.8,
        'random_state': 42,
        'verbosity': 0
    }
    if gpu_available:
        xgb_params['device'] = 'cuda'
        xgb_params['tree_method'] = 'hist'
        print("   🚀 Using GPU acceleration")
    else:
        xgb_params['n_jobs'] = -1

    models['XGBoost'] = XGBRegressor(**xgb_params)
    cv_results['XGBoost'] = cross_val_score(models['XGBoost'], X_train, y_train, cv=kfold, scoring='r2', n_jobs=-1)
    print(f"   ✅ CV R²: {cv_results['XGBoost'].mean():.4f} ± {cv_results['XGBoost'].std():.4f}")
except Exception as e:
    print(f"   ❌ XGBoost failed: {str(e)[:100]}")

# ─────────────────────────────────────────────────────────────
# 2. LightGBM (GPU support - safer approach)
# ─────────────────────────────────────────────────────────────
print("\n2️⃣ Training LightGBM...")
try:
    lgbm_params = {
        'n_estimators': 200,
        'max_depth': 6,
        'learning_rate': 0.05,
        'subsample': 0.8,
        'colsample_bytree': 0.8,
        'random_state': 42,
        'verbose': -1
    }
    if gpu_available:
        try:
            # محتاط‌تر: اول بدون GPU تست می‌کنیم
            lgbm_params['device'] = 'gpu'
            lgbm_params['gpu_platform_id'] = 0
            lgbm_params['gpu_device_id'] = 0
            print("   🚀 Attempting GPU acceleration")
        except:
            lgbm_params['n_jobs'] = -1
            print("   ⚠️ GPU failed, using CPU")
    else:
        lgbm_params['n_jobs'] = -1

    models['LightGBM'] = LGBMRegressor(**lgbm_params)
    cv_results['LightGBM'] = cross_val_score(models['LightGBM'], X_train, y_train, cv=kfold, scoring='r2', n_jobs=-1)
    print(f"   ✅ CV R²: {cv_results['LightGBM'].mean():.4f} ± {cv_results['LightGBM'].std():.4f}")
except Exception as e:
    print(f"   ❌ LightGBM failed: {str(e)[:100]}")
    # Fallback: CPU version
    print("   🔄 Retrying with CPU...")
    models['LightGBM'] = LGBMRegressor(
        n_estimators=200, max_depth=6, learning_rate=0.05,
        subsample=0.8, colsample_bytree=0.8, random_state=42,
        n_jobs=-1, verbose=-1
    )
    cv_results['LightGBM'] = cross_val_score(models['LightGBM'], X_train, y_train, cv=kfold, scoring='r2', n_jobs=-1)
    print(f"   ✅ CV R² (CPU): {cv_results['LightGBM'].mean():.4f} ± {cv_results['LightGBM'].std():.4f}")

# ─────────────────────────────────────────────────────────────
# 3. CatBoost (GPU support)
# ─────────────────────────────────────────────────────────────
print("\n3️⃣ Training CatBoost...")
try:
    catboost_params = {
        'iterations': 200,
        'depth': 6,
        'learning_rate': 0.05,
        'subsample': 0.8,
        'random_state': 42,
        'verbose': 0
    }
    if gpu_available:
        catboost_params['task_type'] = 'GPU'
        print("   🚀 Using GPU acceleration")
    else:
        catboost_params['thread_count'] = -1

    models['CatBoost'] = CatBoostRegressor(**catboost_params)
    cv_results['CatBoost'] = cross_val_score(models['CatBoost'], X_train, y_train, cv=kfold, scoring='r2', n_jobs=-1)
    print(f"   ✅ CV R²: {cv_results['CatBoost'].mean():.4f} ± {cv_results['CatBoost'].std():.4f}")
except Exception as e:
    print(f"   ❌ CatBoost failed: {str(e)[:100]}")

# ─────────────────────────────────────────────────────────────
# 4. Sklearn GradientBoosting (CPU only - always works)
# ─────────────────────────────────────────────────────────────
print("\n4️⃣ Training Sklearn GradientBoosting...")
try:
    models['Sklearn_GB'] = GradientBoostingRegressor(
        n_estimators=200,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        random_state=42,
        verbose=0
    )
    cv_results['Sklearn_GB'] = cross_val_score(models['Sklearn_GB'], X_train, y_train, cv=kfold, scoring='r2', n_jobs=-1)
    print(f"   ✅ CV R²: {cv_results['Sklearn_GB'].mean():.4f} ± {cv_results['Sklearn_GB'].std():.4f}")
except Exception as e:
    print(f"   ❌ Sklearn_GB failed: {str(e)[:100]}")

# ─────────────────────────────────────────────────────────────
# 5. جدول مقایسه (فقط مدل‌های موفق)
# ─────────────────────────────────────────────────────────────
if cv_results:
    print("\n" + "="*80)
    print("📊 FINAL MODEL COMPARISON")
    print("="*80)

    comparison_df = pd.DataFrame({
        'Model': list(cv_results.keys()),
        'CV_R2': [scores.mean() for scores in cv_results.values()],
        'CV_Std': [scores.std() for scores in cv_results.values()],
        'Score_Estimate': [(scores.mean() * 1000).round() / 1000 * 250 for scores in cv_results.values()]
    }).sort_values('CV_R2', ascending=False)

    print(comparison_df.to_string(index=False))

    # بهترین مدل
    best_model_name = comparison_df.iloc[0]['Model']
    best_model = models[best_model_name]
    best_cv_score = comparison_df.iloc[0]['CV_R2']

    print("\n" + "="*80)
    print(f"🏆 BEST MODEL: {best_model_name}")
    print("="*80)
    print(f"   CV R² Score: {best_cv_score:.4f} ± {comparison_df.iloc[0]['CV_Std']:.4f}")
    print(f"   Estimated Score: {comparison_df.iloc[0]['Score_Estimate']:.1f} / 250")

    # آموزش نهایی
    print(f"\n⚙️ Training {best_model_name} on full dataset...")
    best_model.fit(X_train, y_train)

    y_train_pred = best_model.predict(X_train)
    train_r2 = r2_score(y_train, y_train_pred)
    gap = train_r2 - best_cv_score

    print(f"   Train R²: {train_r2:.4f}")
    print(f"   Gap: {gap:.4f} {'✅ Good' if gap < 0.15 else '⚠️ Overfitting'}")

    print("\n✅ Best model ready for prediction!")
else:
    print("\n❌ همه مدل‌ها با خطا مواجه شدند!")
```

```python
# نصب (اگر نصب نیست)
!pip install optuna -q

import optuna
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import cross_val_score
import time

print("="*80)
print("STEP 19: BAYESIAN OPTIMIZATION FOR GRADIENT BOOSTING")
print("="*80)

# ─────────────────────────────────────────────────────────────
# 1. تعریف Objective Function
# ─────────────────────────────────────────────────────────────
def objective(trial):
    """
    Optuna به صورت هوشمند hyperparameters رو پیشنهاد میده
    """
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 500, step=50),
        'max_depth': trial.suggest_int('max_depth', 3, 10),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.1, log=True),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'max_features': trial.suggest_categorical('max_features', ['sqrt', 0.7, 0.8, 0.9]),
        'min_samples_split': trial.suggest_int('min_samples_split', 2, 20),
        'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 10),
        'random_state': 42,
        'verbose': 0
    }

    # ساخت مدل با params پیشنهادی
    model = GradientBoostingRegressor(**params)

    # ارزیابی با Cross-Validation
    scores = cross_val_score(model, X_train, y_train, cv=kfold, scoring='r2', n_jobs=-1)

    return scores.mean()

# ─────────────────────────────────────────────────────────────
# 2. ساخت Study و شروع Optimization
# ─────────────────────────────────────────────────────────────
print("\n🔍 شروع Bayesian Optimization...")

# ایجاد study
study = optuna.create_study(
    direction='maximize',  # چون می‌خواهیم R² را بیشینه کنیم
    sampler=optuna.samplers.TPESampler(seed=42),  # بهترین sampler
    pruner=optuna.pruners.MedianPruner()  # اگر trial ضعیف بود، زود متوقف می‌شود
)

# شروع timing
start_time = time.time()
print(f"⏱️ شروع: {time.strftime('%H:%M:%S')}")

# اجرای optimization
study.optimize(
    objective,
    n_trials=50,  # تعداد آزمایش (50 کافیه، GridSearch 72+ داشت!)
    timeout=600,  # حداکثر 10 دقیقه
    show_progress_bar=True,
    n_jobs=1  # برای stability بهتره 1 باشه
)

elapsed_time = time.time() - start_time
print(f"\n⏱️ پایان: {time.strftime('%H:%M:%S')}")
print(f"⏱️ زمان سپری شده: {elapsed_time/60:.1f} دقیقه")

# ─────────────────────────────────────────────────────────────
# 3. نمایش نتایج
# ─────────────────────────────────────────────────────────────
print("\n" + "="*80)
print("🏆 بهترین نتایج:")
print("="*80)

best_params = study.best_params
best_score = study.best_value

print(f"📊 بهترین CV R² Score: {best_score:.4f}")
print(f"\n🎯 بهترین hyperparameters:")
for param, value in sorted(best_params.items()):
    print(f"  {param:20s}: {value}")

# امتیاز تخمینی
estimated_score = round(best_score, 3) * 250
print(f"\n🏆 امتیاز تخمینی نهایی: {estimated_score:.1f} / 250")

# ─────────────────────────────────────────────────────────────
# 4. آموزش مدل نهایی با بهترین params
# ─────────────────────────────────────────────────────────────
print("\n⚙️ آموزش مدل نهایی با بهترین hyperparameters...")
best_gb_bayesian = GradientBoostingRegressor(**best_params, random_state=42, verbose=0)
best_gb_bayesian.fit(X_train, y_train)

# ارزیابی
y_train_pred = best_gb_bayesian.predict(X_train)
train_r2 = r2_score(y_train, y_train_pred)
gap = train_r2 - best_score

print(f"   Train R²: {train_r2:.4f}")
print(f"   CV R²: {best_score:.4f}")
print(f"   Gap: {gap:.4f} {'✅ Good' if gap < 0.15 else '⚠️ Overfitting'}")

# ─────────────────────────────────────────────────────────────
# 5. مقایسه با روش‌های قبلی
# ─────────────────────────────────────────────────────────────
print("\n" + "="*80)
print("📊 مقایسه Bayesian vs Default:")
print("="*80)

if 'cv_results' in locals() and 'Sklearn_GB' in cv_results:
    default_score = cv_results['Sklearn_GB'].mean()
    print(f"Default model CV R²:    {default_score:.4f}")
    print(f"Bayesian model CV R²:   {best_score:.4f}")
    improvement = (best_score - default_score) * 100
    print(f"بهبود:                  {improvement:+.2f}%")

    if improvement > 0:
        print("\n🎉 Bayesian Optimization موفق بود!")
    else:
        print("\n⚠️ مدل default بهتر بود (overfitting احتمالی)")

# ─────────────────────────────────────────────────────────────
# 6. Visualization (اختیاری)
# ─────────────────────────────────────────────────────────────
print("\n📈 رسم نمودار Optimization History...")
try:
    import matplotlib.pyplot as plt

    fig = optuna.visualization.matplotlib.plot_optimization_history(study)
    plt.title('Bayesian Optimization History')
    plt.tight_layout()
    plt.show()

    # اهمیت پارامترها
    fig = optuna.visualization.matplotlib.plot_param_importances(study)
    plt.title('Hyperparameter Importances')
    plt.tight_layout()
    plt.show()
except:
    print("⚠️ رسم نمودار ممکن نیست")

# ─────────────────────────────────────────────────────────────
# 7. نمایش Top 5 trials
# ─────────────────────────────────────────────────────────────
print("\n📋 Top 5 بهترین trials:")
trials_df = study.trials_dataframe().nlargest(5, 'value')
for idx, row in enumerate(trials_df.itertuples(), 1):
    print(f"{idx}. Score: {row.value:.4f} (Trial #{row.number})")

print("\n✅ Bayesian Optimization completed!")
print(f"✅ بهترین مدل ذخیره شد در: best_gb_bayesian")
print("="*80)
```
