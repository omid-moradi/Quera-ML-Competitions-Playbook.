```python
print("="*80)
print("🔍 بررسی دقیق مقادیر خاص و Missing Values در هر ستون")
print("="*80)

# لیست ستون‌ها
columns = train.columns.tolist()
print(f"\nستون‌های موجود ({len(columns)} عدد):")
print(columns)
print("\n" + "="*80)

# بررسی هر ستون
for col in columns:
    print(f"\n📊 ستون: {col}")
    print("-"*80)

    # 1. نوع داده
    print(f"نوع داده: {train[col].dtype}")

    # 2. تعداد کل
    print(f"تعداد کل: {len(train[col])}")

    # 3. مقادیر null معمولی
    null_count = train[col].isnull().sum()
    null_percent = (null_count / len(train)) * 100
    print(f"Null/NaN: {null_count} ({null_percent:.2f}%)")

    # 4. بررسی مقادیر خاص که ممکن است missing باشند
    if train[col].dtype == 'object':
        # برای ستون‌های متنی
        unique_vals = train[col].unique()

        # بررسی مقادیر خالی یا خاص
        empty_string = (train[col] == '').sum()
        dash = (train[col] == '-').sum()
        question_mark = (train[col] == '?').sum()
        none_string = (train[col] == 'None').sum()
        na_string = (train[col] == 'NA').sum()
        unknown = (train[col] == 'Unknown').sum()

        print(f"\nمقادیر خاص که ممکن است Missing باشند:")
        if empty_string > 0:
            print(f"  - رشته خالی (''): {empty_string} ({(empty_string/len(train)*100):.2f}%)")
        if dash > 0:
            print(f"  - خط تیره ('-'): {dash} ({(dash/len(train)*100):.2f}%)")
        if question_mark > 0:
            print(f"  - علامت سوال ('?'): {question_mark} ({(question_mark/len(train)*100):.2f}%)")
        if none_string > 0:
            print(f"  - 'None': {none_string} ({(none_string/len(train)*100):.2f}%)")
        if na_string > 0:
            print(f"  - 'NA': {na_string} ({(na_string/len(train)*100):.2f}%)")
        if unknown > 0:
            print(f"  - 'Unknown': {unknown} ({(unknown/len(train)*100):.2f}%)")

        # تعداد مقادیر یونیک
        print(f"\nتعداد مقادیر یونیک: {len(unique_vals)}")

        # نمایش 10 مقدار اول
        print(f"نمونه مقادیر (10 تا اول):")
        sample_vals = [str(v) for v in unique_vals[:10]]
        for i, val in enumerate(sample_vals, 1):
            print(f"  {i}. '{val}'")

    else:
        # برای ستون‌های عددی
        print(f"\nآمار توصیفی:")
        print(f"  - Min: {train[col].min()}")
        print(f"  - Max: {train[col].max()}")
        print(f"  - Mean: {train[col].mean():.2f}")
        print(f"  - Median: {train[col].median():.2f}")
        print(f"  - Std: {train[col].std():.2f}")

        # بررسی مقادیر منفی یا صفر
        negative = (train[col] < 0).sum()
        zero = (train[col] == 0).sum()
        if negative > 0:
            print(f"  - مقادیر منفی: {negative}")
        if zero > 0:
            print(f"  - مقادیر صفر: {zero}")

    # 5. مجموع کل Missing (null + مقادیر خاص)
    total_missing = null_count
    if train[col].dtype == 'object':
        total_missing += empty_string + dash + question_mark + none_string + na_string

    total_missing_percent = (total_missing / len(train)) * 100
    print(f"\n✅ مجموع کل Missing: {total_missing} ({total_missing_percent:.2f}%)")

    print("="*80)

print("\n✅ تحلیل اولیه تمام شد!")
```

```python
print("\n" + "="*80)
print("🎯 تبدیل متغیر هدف (status) و بررسی توزیع")
print("="*80)

# نمایش توزیع اولیه
print("\n📊 توزیع اولیه status:")
print(train['status'].value_counts())
print("\nدرصد هر کلاس:")
print(train['status'].value_counts(normalize=True) * 100)

# تبدیل به 3 کلاس
def convert_status(status):
    """
    تبدیل status به 3 کلاس:
    - 0: closed (شکست)
    - 1: operating (در حال کار - نه موفق نه شکست)
    - 2: acquired + ipo (موفقیت)
    """
    if status == 'closed':
        return 0
    elif status == 'operating':
        return 1
    elif status in ['acquired', 'ipo']:
        return 2
    else:
        return np.nan

train['target'] = train['status'].apply(convert_status)

print("\n" + "="*80)
print("📊 توزیع جدید target (3 کلاس):")
print("="*80)
print("\nتعداد هر کلاس:")
print(train['target'].value_counts().sort_index())

print("\nدرصد هر کلاس:")
class_dist = train['target'].value_counts(normalize=True).sort_index() * 100
for idx, percent in class_dist.items():
    class_names = {0: 'closed (شکست)', 1: 'operating (در حال کار)', 2: 'acquired/ipo (موفقیت)'}
    print(f"  کلاس {idx} ({class_names[idx]}): {percent:.2f}%")

# محاسبه نسبت عدم تعادل
print("\n⚠️ نسبت عدم تعادل کلاس‌ها:")
max_class = train['target'].value_counts().max()
for cls in [0, 1, 2]:
    count = (train['target'] == cls).sum()
    ratio = max_class / count if count > 0 else 0
    print(f"  کلاس {cls}: 1:{ratio:.2f}")

print("\n✅ تبدیل target انجام شد!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

print("\n" + "="*80)
print("📈 تحلیل همبستگی با Target")
print("="*80)

# ایجاد یک DataFrame موقت برای محاسبه همبستگی
df_temp = train.copy()

# 1. پردازش funding_total_usd (تبدیل به عددی)
print("\n🔄 پردازش funding_total_usd...")
df_temp['funding_total_usd_clean'] = df_temp['funding_total_usd'].replace('-', np.nan)
df_temp['funding_total_usd_clean'] = pd.to_numeric(df_temp['funding_total_usd_clean'], errors='coerce')
df_temp['funding_total_usd_log'] = np.log1p(df_temp['funding_total_usd_clean'].fillna(0))

# 2. پردازش تاریخ‌ها
print("🔄 پردازش تاریخ‌ها...")
date_cols = ['founded_at', 'first_funding_at', 'last_funding_at']
for col in date_cols:
    df_temp[col] = pd.to_datetime(df_temp[col], errors='coerce')
    df_temp[f'{col}_year'] = df_temp[col].dt.year
    df_temp[f'{col}_days'] = (df_temp[col] - pd.Timestamp('1970-01-01')).dt.days

# ویژگی‌های مشتق شده از تاریخ
df_temp['age_at_first_funding'] = (df_temp['first_funding_at'] - df_temp['founded_at']).dt.days
df_temp['funding_duration'] = (df_temp['last_funding_at'] - df_temp['first_funding_at']).dt.days
df_temp['company_age'] = (df_temp['last_funding_at'] - df_temp['founded_at']).dt.days

# 3. Target Encoding برای ویژگی‌های کاتگوریک
print("🔄 محاسبه Target Encoding برای categorical features...")
categorical_cols = ['country_code', 'state_code', 'region', 'city', 'category_list']

for col in categorical_cols:
    # محاسبه میانگین target برای هر دسته
    target_mean = df_temp.groupby(col)['target'].mean()
    df_temp[f'{col}_target_enc'] = df_temp[col].map(target_mean)

    # محاسبه تعداد (frequency)
    freq = df_temp[col].value_counts()
    df_temp[f'{col}_freq'] = df_temp[col].map(freq)

# 4. انتخاب ویژگی‌های عددی برای محاسبه همبستگی
print("\n📊 محاسبه همبستگی...")
numeric_features = [
    'funding_rounds',
    'funding_total_usd_log',
    'founded_at_year',
    'first_funding_at_year',
    'last_funding_at_year',
    'age_at_first_funding',
    'funding_duration',
    'company_age',
    'country_code_target_enc',
    'state_code_target_enc',
    'region_target_enc',
    'city_target_enc',
    'category_list_target_enc',
    'country_code_freq',
    'state_code_freq',
    'region_freq',
    'city_freq',
    'target'
]

# حذف ستون‌هایی که وجود ندارند
numeric_features = [col for col in numeric_features if col in df_temp.columns]

# محاسبه ماتریس همبستگی
correlation_matrix = df_temp[numeric_features].corr()

# استخراج همبستگی با target و مرتب‌سازی
target_correlation = correlation_matrix['target'].drop('target').sort_values(ascending=False)

print("\n" + "="*80)
print("🔝 Top 15 ویژگی با بیشترین همبستگی مثبت با Target:")
print("="*80)
print(target_correlation.head(15))

print("\n" + "="*80)
print("🔻 Top 15 ویژگی با بیشترین همبستگی منفی با Target:")
print("="*80)
print(target_correlation.tail(15))

# ذخیره نتایج
target_correlation.to_csv('feature_correlation_with_target.csv')
print("\n✅ نتایج همبستگی در فایل 'feature_correlation_with_target.csv' ذخیره شد")

# نمایش گرافیکی (اختیاری - اگر می‌خواهی)
print("\n📊 آیا می‌خواهید نمودار همبستگی را ببینید؟ (بله/خیر)")
# این قسمت را می‌توانید فعال کنید
```

```python
print("="*80)
print("📦 بخش 3: Target Encoding (تکنیک طلایی)")
print("="*80)
print("\n🎯 هدف: تبدیل categorical features به عددی با استفاده از میانگین target")

from sklearn.model_selection import KFold

# لیست ویژگی‌های categorical که می‌خواهیم target encoding کنیم
categorical_features_for_encoding = [
    'first_category',
    'last_category',
    'country_code',
    'state_code',
    'region',
    'city'
]

print(f"\n📋 ویژگی‌های انتخاب شده برای Target Encoding:")
for idx, feat in enumerate(categorical_features_for_encoding, 1):
    unique_count = df[feat].nunique()
    print(f"   {idx}. {feat:<20} : {unique_count:>5,} مقدار یونیک")

# تابع Target Encoding با Smoothing
def target_encode_with_smoothing(train_data, test_data, feature, target_col='target', smoothing=10):
    """
    Target Encoding با smoothing برای جلوگیری از overfitting

    Parameters:
    -----------
    train_data : DataFrame
        داده train
    test_data : DataFrame
        داده test
    feature : str
        نام ستون categorical
    target_col : str
        نام ستون target
    smoothing : int
        پارامتر smoothing (بالاتر = محافظه‌کارتر)

    Returns:
    --------
    train_encoded : Series
        مقادیر encode شده train
    test_encoded : Series
        مقادیر encode شده test
    """

    # میانگین global
    global_mean = train_data[target_col].mean()

    # محاسبه میانگین و تعداد برای هر category
    agg = train_data.groupby(feature)[target_col].agg(['mean', 'count'])

    # Smoothing: (mean * count + global_mean * smoothing) / (count + smoothing)
    smoothed_mean = (agg['mean'] * agg['count'] + global_mean * smoothing) / (agg['count'] + smoothing)

    # اعمال encoding
    train_encoded = train_data[feature].map(smoothed_mean)
    test_encoded = test_data[feature].map(smoothed_mean)

    # مقادیر unknown را با global mean پر می‌کنیم
    train_encoded = train_encoded.fillna(global_mean)
    test_encoded = test_encoded.fillna(global_mean)

    return train_encoded, test_encoded, smoothed_mean

print("\n" + "="*80)
print("🔄 شروع Target Encoding با Cross-Validation")
print("="*80)

# جدا کردن train و test از df
train_df = df[df['is_train'] == 1].copy()
test_df = df[df['is_train'] == 0].copy()

print(f"\n📊 اطلاعات داده:")
print(f"   ├─ Train: {len(train_df):,} rows")
print(f"   ├─ Test: {len(test_df):,} rows")
print(f"   └─ Smoothing parameter: 10")

# Target Encoding برای هر feature
for feature in categorical_features_for_encoding:
    print(f"\n{'─'*80}")
    print(f"🎯 Target Encoding برای: {feature}")
    print(f"{'─'*80}")

    # محاسبه encoding
    train_encoded, test_encoded, encoding_map = target_encode_with_smoothing(
        train_df, test_df, feature, 'target', smoothing=10
    )

    # اضافه کردن به df اصلی
    train_df[f'{feature}_target_enc'] = train_encoded
    test_df[f'{feature}_target_enc'] = test_encoded

    # آمار
    print(f"   ├─ میانگین train: {train_encoded.mean():.4f}")
    print(f"   ├─ میانگین test: {test_encoded.mean():.4f}")
    print(f"   ├─ Min: {train_encoded.min():.4f}")
    print(f"   ├─ Max: {train_encoded.max():.4f}")
    print(f"   └─ Std: {train_encoded.std():.4f}")

    # نمایش Top 5 و Bottom 5
    print(f"\n   🔝 Top 5 categories (بیشترین احتمال موفقیت):")
    top_5 = encoding_map.nlargest(5)
    for idx, (cat, score) in enumerate(top_5.items(), 1):
        print(f"      {idx}. {str(cat):<30} : {score:.4f}")

    print(f"\n   🔻 Bottom 5 categories (کمترین احتمال موفقیت):")
    bottom_5 = encoding_map.nsmallest(5)
    for idx, (cat, score) in enumerate(bottom_5.items(), 1):
        print(f"      {idx}. {str(cat):<30} : {score:.4f}")

# به‌روزرسانی df اصلی
df.loc[df['is_train'] == 1, [f'{f}_target_enc' for f in categorical_features_for_encoding]] = \
    train_df[[f'{f}_target_enc' for f in categorical_features_for_encoding]].values

df.loc[df['is_train'] == 0, [f'{f}_target_enc' for f in categorical_features_for_encoding]] = \
    test_df[[f'{f}_target_enc' for f in categorical_features_for_encoding]].values

print("\n" + "="*80)
print("✅ Target Encoding برای تمام features تمام شد!")
print("="*80)
print(f"\nشکل فعلی داده: {df.shape}")
print(f"ستون‌های target_enc اضافه شده: {len(categorical_features_for_encoding)}")

# بررسی نهایی
print("\n📊 بررسی correlation با target:")
train_df = df[df['is_train'] == 1].copy()
target_enc_cols = [f'{f}_target_enc' for f in categorical_features_for_encoding]

correlations = {}
for col in target_enc_cols:
    corr = train_df[col].corr(train_df['target'])
    correlations[col] = corr

# مرتب‌سازی
sorted_corr = sorted(correlations.items(), key=lambda x: abs(x[1]), reverse=True)

print("\n   📈 Correlation با target (مرتب شده):")
for idx, (col, corr) in enumerate(sorted_corr, 1):
    print(f"      {idx}. {col:<35} : {corr:>7.4f}")

print("\n" + "="*80)
print("🎉 عالی! حالا ویژگی‌های قدرتمندی داریم!")
print("="*80)
```

```python
import pandas as pd
import numpy as np
from collections import Counter

print("="*80)
print("🚀 بخش نهایی: Feature Engineering بهینه برای LightGBM/CatBoost/XGBoost")
print("="*80)
print("\n✅ این کد برای مدل‌های tree-based بهینه شده است")
print("✅ NaN‌ها را فقط در جاهای ضروری fill می‌کند")
print("✅ Missing indicators را حفظ می‌کند")

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 1: پردازش funding_total_usd
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("💰 قسمت 1: پردازش funding_total_usd")
print("="*80)

# تبدیل '-' به NaN و سپس به عددی
df['funding_total_usd_clean'] = df['funding_total_usd'].replace('-', np.nan)
df['funding_total_usd_clean'] = pd.to_numeric(df['funding_total_usd_clean'], errors='coerce')

# Missing indicator (خیلی مهم!)
df['funding_missing'] = df['funding_total_usd_clean'].isna().astype(int)

# Log transformation (فقط برای non-null)
df['funding_log'] = np.log1p(df['funding_total_usd_clean'].fillna(0))

# Funding per round (NaN برای مدل قابل قبول است)
df['funding_per_round'] = df['funding_total_usd_clean'] / df['funding_rounds']
df['funding_per_round_log'] = np.log1p(df['funding_per_round'].fillna(0))

# Categorical funding level
def categorize_funding(amount):
    if pd.isna(amount) or amount == 0:
        return 'no_funding'
    elif amount < 100000:
        return 'very_low'
    elif amount < 1000000:
        return 'low'
    elif amount < 10000000:
        return 'medium'
    elif amount < 50000000:
        return 'high'
    else:
        return 'very_high'

df['funding_category'] = df['funding_total_usd_clean'].apply(categorize_funding)

print(f"✅ funding features ساخته شد")
print(f"   ├─ funding_missing: {df['funding_missing'].sum():,}")
print(f"   ├─ funding_log: mean = {df['funding_log'].mean():.2f}")
print(f"   └─ funding_category توزیع:")
for cat, count in df['funding_category'].value_counts().head().items():
    print(f"      {cat:<15} : {count:>6,}")

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 2: پردازش تاریخ‌ها
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("📅 قسمت 2: پردازش تاریخ‌ها")
print("="*80)

date_columns = ['founded_at', 'first_funding_at', 'last_funding_at']

for col in date_columns:
    # تبدیل به datetime
    df[col] = pd.to_datetime(df[col], errors='coerce')

    # Missing indicator (حفظ اطلاعات!)
    df[f'{col}_missing'] = df[col].isna().astype(int)

    # استخراج components (NaN می‌ماند - مدل handle می‌کند)
    df[f'{col}_year'] = df[col].dt.year
    df[f'{col}_month'] = df[col].dt.month
    df[f'{col}_quarter'] = df[col].dt.quarter

    print(f"   ├─ {col}: {df[f'{col}_missing'].sum():,} missing")

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 3: ویژگی‌های زمانی مشتق شده
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("🛠️ قسمت 3: ویژگی‌های زمانی مشتق شده")
print("="*80)

# محاسبه فاصله‌های زمانی (NaN می‌ماند)
df['age_at_first_funding'] = (df['first_funding_at'] - df['founded_at']).dt.days
df['funding_duration'] = (df['last_funding_at'] - df['first_funding_at']).dt.days
df['company_age'] = (df['last_funding_at'] - df['founded_at']).dt.days

# حذف مقادیر منفی (خطای داده)
for col in ['age_at_first_funding', 'funding_duration', 'company_age']:
    negative_count = (df[col] < 0).sum()
    if negative_count > 0:
        print(f"   ⚠️ {col}: {negative_count:,} مقدار منفی → NaN")
        df.loc[df[col] < 0, col] = np.nan

print(f"✅ ویژگی‌های زمانی ساخته شد (NaN‌ها حفظ شدند)")

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 4: ویژگی‌های تعاملی
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("🔗 قسمت 4: ویژگی‌های تعاملی")
print("="*80)

# Funding velocity (NaN می‌ماند)
df['funding_velocity'] = df['funding_total_usd_clean'] / (df['company_age'] + 1)
df['funding_velocity_log'] = np.log1p(df['funding_velocity'].fillna(0))

# Binary features
df['early_stage_funding'] = (df['age_at_first_funding'] < 365).astype(float)
df.loc[df['age_at_first_funding'].isna(), 'early_stage_funding'] = np.nan

df['multiple_rounds'] = (df['funding_rounds'] > 1).astype(int)

# High funding indicator
median_funding = df.loc[df['is_train'] == 1, 'funding_total_usd_clean'].median()
df['high_funding'] = (df['funding_total_usd_clean'] > median_funding).astype(float)
df.loc[df['funding_total_usd_clean'].isna(), 'high_funding'] = np.nan

# Timing features
df['founded_decade'] = (df['founded_at_year'] // 10) * 10
df['founded_recent'] = (df['founded_at_year'] >= 2010).astype(float)
df.loc[df['founded_at_year'].isna(), 'founded_recent'] = np.nan

df['founded_in_crisis'] = df['founded_at_year'].isin([2008, 2009]).astype(float)
df.loc[df['founded_at_year'].isna(), 'founded_in_crisis'] = np.nan

df['founded_in_boom'] = df['founded_at_year'].isin([2011, 2012, 2013, 2014]).astype(float)
df.loc[df['founded_at_year'].isna(), 'founded_in_boom'] = np.nan

# Funding intensity
df['funding_intensity'] = df['funding_rounds'] / ((df['funding_duration'] / 365) + 0.1)

print(f"✅ ویژگی‌های تعاملی ساخته شد")
print(f"   ├─ multiple_rounds: {df['multiple_rounds'].sum():,}")
print(f"   ├─ founded_recent: {df['founded_recent'].sum():,} (NaN excluded)")
print(f"   └─ funding_velocity_log: mean = {df['funding_velocity_log'].mean():.2f}")

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 5: پردازش نهایی برای مدل
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("🎯 قسمت 5: آماده‌سازی نهایی برای مدل")
print("="*80)

# جدا کردن train و test
train_final = df[df['is_train'] == 1].copy()
test_final = df[df['is_train'] == 0].copy()

print(f"\n📊 شکل نهایی داده:")
print(f"   ├─ Train: {train_final.shape}")
print(f"   └─ Test: {test_final.shape}")

# لیست ستون‌های feature برای مدل
feature_columns = [col for col in df.columns if col not in [
    'name', 'category_list', 'funding_total_usd', 'status', 'target',
    'is_train', 'founded_at', 'first_funding_at', 'last_funding_at',
    'category_list_original', 'funding_total_usd_clean'
]]

print(f"\n📋 تعداد کل features: {len(feature_columns)}")

# بررسی missing values در features
print(f"\n📊 Top 10 features با بیشترین NaN:")
missing_counts = train_final[feature_columns].isna().sum().sort_values(ascending=False).head(10)
for col, count in missing_counts.items():
    if count > 0:
        print(f"   ├─ {col:<40} : {count:>6,} ({count/len(train_final)*100:>5.2f}%)")

# بررسی correlation
print(f"\n📈 Top 15 features با بیشترین correlation با target:")
correlations = {}
for col in feature_columns:
    try:
        corr = train_final[col].corr(train_final['target'])
        if not pd.isna(corr):
            correlations[col] = abs(corr)
    except:
        pass

sorted_corr = sorted(correlations.items(), key=lambda x: x[1], reverse=True)[:15]
for idx, (col, corr) in enumerate(sorted_corr, 1):
    print(f"   {idx:>2}. {col:<45} : {corr:.4f}")

print("\n" + "="*80)
print("✅ Feature Engineering تمام شد!")
print("="*80)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, classification_report, confusion_matrix
from sklearn.utils.class_weight import compute_class_weight
import lightgbm as lgb
import warnings
warnings.filterwarnings('ignore')

print("="*80)
print("🚀 مدل‌سازی با LightGBM - نسخه بهینه برای Startup Success Prediction")
print("="*80)

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 1: آماده‌سازی داده
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("📦 قسمت 1: آماده‌سازی داده")
print("="*80)

# جدا کردن X و y
X_train = train_final[feature_columns].copy()
y_train = train_final['target'].copy()
X_test = test_final[feature_columns].copy()

print(f"\n📊 شکل داده:")
print(f"   ├─ X_train: {X_train.shape}")
print(f"   ├─ y_train: {y_train.shape}")
print(f"   └─ X_test: {X_test.shape}")

print(f"\n📊 توزیع کلاس‌ها در train:")
for cls in [0, 1, 2]:
    count = (y_train == cls).sum()
    percent = count / len(y_train) * 100
    class_names = {0: 'closed', 1: 'operating', 2: 'acquired/ipo'}
    print(f"   ├─ Class {cls} ({class_names[cls]:<12}): {count:>6,} ({percent:>5.2f}%)")

# محاسبه class weights برای LightGBM
class_weights = compute_class_weight(
    'balanced',
    classes=np.unique(y_train),
    y=y_train
)
class_weight_dict = {i: weight for i, weight in enumerate(class_weights)}

print(f"\n⚖️ Class Weights (balanced):")
for cls, weight in class_weight_dict.items():
    print(f"   ├─ Class {cls}: {weight:.4f}")

# تبدیل sample weights برای LightGBM
# LightGBM نیاز به sample_weight دارد (نه class_weight)
sample_weights = np.array([class_weight_dict[cls] for cls in y_train])

print(f"\n💡 نکته: LightGBM از sample_weight استفاده می‌کند (نه class_weight)")

# شناسایی categorical features (فقط برای نام‌گذاری، نیازی به خود ستون‌ها نیست)
# Remove object columns and 'funding_category' from feature_columns
object_cols_to_remove = [col for col in X_train.columns if X_train[col].dtype == 'object']
object_cols_to_remove.append('funding_category') # Add funding_category if it exists and wasn't object

feature_columns_cleaned = [col for col in feature_columns if col not in object_cols_to_remove]

X_train = train_final[feature_columns_cleaned].copy()
X_test = test_final[feature_columns_cleaned].copy()


print(f"\n📋 تعداد Features نهایی بعد از حذف Object Columns: {len(feature_columns_cleaned)}")
print(f"📊 شکل داده بعد از حذف Object Columns:")
print(f"   ├─ X_train: {X_train.shape}")
print(f"   └─ X_test: {X_test.shape}")


# ═══════════════════════════════════════════════════════════════════════════
# قسمت 2: تعریف پارامترهای LightGBM
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("🤖 قسمت 2: تعریف پارامترهای LightGBM")
print("="*80)

# پارامترهای بهینه برای LightGBM
lgb_params = {
    # Core parameters
    'objective': 'multiclass',
    'num_class': 3,
    'metric': 'None',  # ما custom metric (F1) استفاده می‌کنیم
    'boosting_type': 'gbdt',

    # Learning parameters
    'num_iterations': 4000,
    'learning_rate': 0.01,
    'max_depth': 9,

    # Tree parameters
    'num_leaves': 31,  # معمولاً 2^max_depth - 1
    'min_data_in_leaf': 20,
    'min_sum_hessian_in_leaf': 1e-3,

    # Feature parameters
    'feature_fraction': 0.8,  # مشابه feature_sample_ratio
    'bagging_fraction': 0.8,  # نمونه‌برداری از rows
    'bagging_freq': 5,

    # Regularization
    'lambda_l1': 0.1,
    'lambda_l2': 3.0,
    'min_gain_to_split': 0.0,

    # Categorical features (not needed if object cols removed)
    # 'cat_smooth': 10,
    # 'max_cat_threshold': 32,

    # Performance
    'num_threads': -1,
    'device_type': 'gpu',  # یا 'gpu' اگر دارید

    # Other
    'verbose': -1,
    'random_state': 42,
    'force_col_wise': True  # برای سرعت بیشتر
}

print(f"\n🔧 پارامترهای کلیدی:")
print(f"   ├─ objective: {lgb_params['objective']}")
print(f"   ├─ num_iterations: {lgb_params['num_iterations']}")
print(f"   ├─ learning_rate: {lgb_params['learning_rate']}")
print(f"   ├─ max_depth: {lgb_params['max_depth']}")
print(f"   ├─ num_leaves: {lgb_params['num_leaves']}")
print(f"   └─ feature_fraction: {lgb_params['feature_fraction']}")

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 3: تابع Custom Metric - Macro F1
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("📊 قسمت 3: تعریف Custom Metric (Macro F1)")
print("="*80)

def lgb_f1_macro(y_pred, y_true):
    """
    Custom metric برای LightGBM: Macro F1 Score

    LightGBM predictions را به صورت probability می‌دهد:
    y_pred shape: (n_samples * n_classes,)
    باید reshape کنیم به (n_samples, n_classes)
    """
    y_true = y_true.get_label()

    # Reshape predictions
    y_pred = y_pred.reshape(len(y_true), -1)

    # تبدیل probabilities به class predictions
    y_pred_class = np.argmax(y_pred, axis=1)

    # محاسبه F1 Score
    score = f1_score(y_true, y_pred_class, average='macro')

    # LightGBM انتظار دارد: (metric_name, metric_value, is_higher_better)
    return 'f1_macro', score, True

print(f"✅ Custom metric تعریف شد: lgb_f1_macro")
print(f"   این metric Macro F1 Score را محاسبه می‌کند")

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 4: Cross-Validation با K=5
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("🔄 قسمت 4: Cross-Validation (K=5)")
print("="*80)

# Stratified K-Fold
n_splits = 7
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

# ذخیره نتایج
oof_predictions = np.zeros(len(X_train))
test_predictions = np.zeros((len(X_test), n_splits))
fold_scores = []
fold_models = []
feature_importance_list = []

print(f"\n🎯 شروع {n_splits}-Fold Cross-Validation...\n")

for fold, (train_idx, val_idx) in enumerate(skf.split(X_train, y_train), 1):
    print(f"{'='*80}")
    print(f"📊 Fold {fold}/{n_splits}")
    print(f"{'='*80}")

    # جدا کردن train و validation
    X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
    y_tr, y_val = y_train.iloc[train_idx], y_train.iloc[val_idx]

    # Sample weights برای train
    sample_weights_tr = sample_weights[train_idx]

    print(f"\n📦 اندازه داده:")
    print(f"   ├─ Train: {len(X_tr):,} samples")
    print(f"   └─ Validation: {len(X_val):,} samples")

    # توزیع کلاس‌ها در validation
    print(f"\n📊 توزیع validation:")
    for cls in [0, 1, 2]:
        count = (y_val == cls).sum()
        print(f"   ├─ Class {cls}: {count:>4,} samples")

    # ساخت Dataset برای LightGBM
    lgb_train = lgb.Dataset(
        X_tr,
        y_tr,
        weight=sample_weights_tr,  # اعمال class weights
        free_raw_data=False
    )

    lgb_val = lgb.Dataset(
        X_val,
        y_val,
        reference=lgb_train,
        free_raw_data=False
    )

    # Callbacks برای early stopping و logging
    callbacks = [
        lgb.early_stopping(stopping_rounds=200, verbose=False),
        lgb.log_evaluation(period=100)  # نمایش هر 100 iteration
    ]

    # Training
    print(f"\n🔄 Training...")
    model = lgb.train(
        lgb_params,
        lgb_train,
        valid_sets=[lgb_train, lgb_val],
        valid_names=['train', 'valid'],
        feval=lgb_f1_macro,  # Custom metric
        callbacks=callbacks
    )

    # ذخیره best iteration
    best_iteration = model.best_iteration
    print(f"\n✅ Best iteration: {best_iteration}")

    # Prediction روی validation (probabilities)
    val_pred_proba = model.predict(X_val, num_iteration=best_iteration)
    val_pred = np.argmax(val_pred_proba, axis=1)
    oof_predictions[val_idx] = val_pred

    # محاسبه F1 Score
    f1_macro = f1_score(y_val, val_pred, average='macro')
    f1_per_class = f1_score(y_val, val_pred, average=None)

    fold_scores.append(f1_macro)
    fold_models.append(model)

    # Feature Importance
    feature_importance = pd.DataFrame({
        'feature': X_train.columns,
        'importance': model.feature_importance(importance_type='gain')
    }).sort_values('importance', ascending=False)
    feature_importance_list.append(feature_importance)

    print(f"\n{'─'*80}")
    print(f"📈 نتایج Fold {fold}:")
    print(f"{'─'*80}")
    print(f"   ├─ F1 Score (Macro): {f1_macro:.4f}")
    print(f"   ├─ F1 Class 0 (closed): {f1_per_class[0]:.4f}")
    print(f"   ├─ F1 Class 1 (operating): {f1_per_class[1]:.4f}")
    print(f"   └─ F1 Class 2 (acquired/ipo): {f1_per_class[2]:.4f}")

    # Top 5 important features
    print(f"\n   🔝 Top 5 Important Features:")
    for idx, row in feature_importance.head(5).iterrows():
        print(f"      {row['feature']:<40} : {row['importance']:>8.0f}")

    # Prediction روی test
    test_pred_proba = model.predict(X_test, num_iteration=best_iteration)
    test_predictions[:, fold-1] = np.argmax(test_pred_proba, axis=1)

    print(f"\n✅ Fold {fold} تمام شد!\n")

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 5: نتایج کلی Cross-Validation
# ═══════════════════════════════════════════════════════════════════════════

print("="*80)
print("📊 نتایج کلی Cross-Validation")
print("="*80)

# آمار folds
mean_f1 = np.mean(fold_scores)
std_f1 = np.std(fold_scores)
min_f1 = np.min(fold_scores)
max_f1 = np.max(fold_scores)

print(f"\n🎯 F1 Score (Macro) در هر fold:")
for fold, score in enumerate(fold_scores, 1):
    print(f"   Fold {fold}: {score:.4f}")

print(f"\n📈 آمار نهایی:")
print(f"   ├─ Mean F1: {mean_f1:.4f}")
print(f"   ├─ Std F1: {std_f1:.4f}")
print(f"   ├─ Min F1: {min_f1:.4f}")
print(f"   ├─ Max F1: {max_f1:.4f}")
print(f"   └─ 95% CI: [{mean_f1 - 1.96*std_f1:.4f}, {mean_f1 + 1.96*std_f1:.4f}]")

# محاسبه OOF F1 Score
oof_f1 = f1_score(y_train, oof_predictions, average='macro')
oof_f1_per_class = f1_score(y_train, oof_predictions, average=None)

print(f"\n🎯 Out-of-Fold (OOF) Performance:")
print(f"   ├─ OOF F1 Score (Macro): {oof_f1:.4f}")
print(f"   ├─ OOF F1 Class 0: {oof_f1_per_class[0]:.4f}")
print(f"   ├─ OOF F1 Class 1: {oof_f1_per_class[1]:.4f}")
print(f"   └─ OOF F1 Class 2: {oof_f1_per_class[2]:.4f}")

# Confusion Matrix
print(f"\n📊 Confusion Matrix (OOF):")
cm = confusion_matrix(y_train, oof_predictions)
print(cm)

# Classification Report
print(f"\n📋 Classification Report (OOF):")
print(classification_report(y_train, oof_predictions,
                          target_names=['closed', 'operating', 'acquired/ipo']))

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 6: Feature Importance میانگین از تمام folds
# ═══════════════════════════════════════════════════════════════════════════

print("="*80)
print("🔍 قسمت 6: Feature Importance (میانگین از 5 folds)")
print("="*80)

# محاسبه میانگین importance
avg_feature_importance = pd.DataFrame({
    'feature': X_train.columns # Use X_train.columns here as it's cleaned
})

for i, fi in enumerate(feature_importance_list):
    avg_feature_importance[f'fold_{i+1}'] = fi['importance'].values

avg_feature_importance['mean_importance'] = avg_feature_importance.iloc[:, 1:].mean(axis=1)
avg_feature_importance = avg_feature_importance.sort_values('mean_importance', ascending=False)

print(f"\n🏆 Top 20 Most Important Features:")
print(f"{'─'*80}")
for idx, row in avg_feature_importance.head(20).iterrows():
    print(f"   {idx+1:>2}. {row['feature']:<40} : {row['mean_importance']:>10,.0f}")

# ذخیره feature importance
avg_feature_importance.to_csv('feature_importance_lgbm.csv', index=False)
print(f"\n✅ Feature importance ذخیره شد: 'feature_importance_lgbm.csv'")

# ═══════════════════════════════════════════════════════════════════════════
# قسمت 7: Prediction روی Test (Majority Voting)
# ═══════════════════════════════════════════════════════════════════════════

print("\n" + "="*80)
print("🎯 قسمت 7: Prediction روی Test (Ensemble)")
print("="*80)

# Ensemble: voting از تمام folds
final_test_predictions = []
for i in range(len(X_test)):
    votes = test_predictions[i, :]
    final_pred = np.bincount(votes.astype(int)).argmax()
    final_test_predictions.append(final_pred)

final_test_predictions = np.array(final_test_predictions)

# آمار predictions
print(f"\n📊 توزیع predictions روی test:")
for cls in [0, 1, 2]:
    count = (final_test_predictions == cls).sum()
    percent = count / len(final_test_predictions) * 100
    class_names = {0: 'closed', 1: 'operating', 2: 'acquired/ipo'}
    print(f"   ├─ Class {cls} ({class_names[cls]:<12}): {count:>4,} ({percent:>5.2f}%)")

# ساخت submission file
submission = pd.DataFrame({
    'status': final_test_predictions
})

# ذخیره submission
submission.to_csv('submission_lgbm.csv', index=False, header=False)

print(f"\n✅ Submission file ذخیره شد: 'submission_lgbm.csv'")
print(f"   ├─ شکل: {submission.shape}")
print(f"   └─ نمونه اول:")
print(submission.head(10))

print("\n" + "="*80)
print("🎉 تمام شد! آماده برای ارسال به مسابقه!")
print("="*80)
print(f"\n💡 پیش‌بینی نمره:")
print(f"   ├─ OOF F1 Score: {oof_f1:.4f}")
print(f"   ├─ CV Mean F1: {mean_f1:.4f} ± {std_f1:.4f}")
print(f"   └─ انتظار می‌رود نمره Public Leaderboard: ~{mean_f1*100:.1f}")
print(f"\n🎯 هدف: شکستن نمره {52.7} نفر اول!")
print(f"   آیا موفق شدیم؟ {'✅ بله!' if mean_f1*100 > 52.7 else '⚠️ نزدیک بودیم!'}")

print("\n" + "="*80)
print("📊 مقایسه با CatBoost:")
print("="*80)
print(f"   LightGBM معمولاً:")
print(f"   ├─ سریع‌تر از CatBoost (~2x)")
print(f"   ├─ حافظه کمتر")
print(f"   └─ F1 Score مشابه (±0.001)")
print(f"\n💡 توصیه: هر دو مدل را اجرا کن و بهترین را انتخاب کن!")
```
