```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# بارگذاری داده
train = pd.read_csv(r'F:\Quera\olympic 2024\internet\train.csv')

# بررسی ابعاد داده
print("Shape of data:", train.shape)
print("\n" + "="*50)

# بررسی 5 سطر اول
print("\nFirst 5 rows:")
print(train.head())
print("\n" + "="*50)

# بررسی اطلاعات کلی
print("\nData Info:")
print(train.info())
print("\n" + "="*50)

# بررسی نام ستون‌ها
print("\nColumn names (first 10):")
print(train.columns[:10].tolist())
print(f"\nTotal columns: {len(train.columns)}")
```

```python
# تبدیل timestamp به string برای پردازش
train['timestamp'] = train['timestamp'].astype(str)

# استخراج اجزای تاریخ و زمان جلالی
train['year'] = train['timestamp'].str[:4].astype(int)
train['month'] = train['timestamp'].str[5:7].astype(int)
train['day'] = train['timestamp'].str[8:10].astype(int)
train['hour'] = train['timestamp'].str[11:13].astype(int)
train['minute'] = train['timestamp'].str[14:16].astype(int)

# بررسی بازه زمانی داده
print("Date range:")
print(f"Start: {train['timestamp'].iloc[0]}")
print(f"End: {train['timestamp'].iloc[-1]}")
print(f"\nTotal records: {len(train)}")
print(f"Expected records for 15-min data: {len(train) / 96} days")
print("\n" + "="*50)

# محاسبه مجموع مصرف همه کاربران در هر 15 دقیقه
user_columns = [col for col in train.columns if col.startswith('U_')]
print(f"\nNumber of user columns: {len(user_columns)}")

# ایجاد ستون total_15min (مجموع مصرف همه کاربران در هر 15 دقیقه)
train['total_15min'] = train[user_columns].sum(axis=1)

print("\nSample of total_15min:")
print(train[['timestamp', 'total_15min']].head(10))
print("\n" + "="*50)

# بررسی آماره‌های total_15min
print("\nStatistics of total_15min:")
print(train['total_15min'].describe())
```

```python
# ایجاد یک timestamp ساعتی (حذف دقیقه)
train['timestamp_hourly'] = (train['year'].astype(str) + '-' + 
                              train['month'].astype(str).str.zfill(2) + '-' + 
                              train['day'].astype(str).str.zfill(2) + ' ' + 
                              train['hour'].astype(str).str.zfill(2) + ':00:00')

# Aggregate کردن به ساعتی با جمع زدن مصرف 15 دقیقه‌ای
hourly_data = train.groupby('timestamp_hourly').agg({
    'total_15min': 'sum',  # جمع مصرف 4 بازه 15 دقیقه‌ای
    'year': 'first',
    'month': 'first',
    'day': 'first',
    'hour': 'first'
}).reset_index()

# تغییر نام ستون
hourly_data.rename(columns={'total_15min': 'total'}, inplace=True)

print("Hourly data shape:", hourly_data.shape)
print(f"Expected hourly records: {1401 * 24} hours")
print("\n" + "="*50)

# بررسی نمونه داده ساعتی
print("\nFirst 10 hourly records:")
print(hourly_data.head(10))
print("\n" + "="*50)

# بررسی آخرین رکوردها
print("\nLast 10 hourly records:")
print(hourly_data.tail(10))
print("\n" + "="*50)

# آماره‌های داده ساعتی
print("\nHourly total statistics:")
print(hourly_data['total'].describe())
print("\n" + "="*50)

# بررسی missing values
print("\nMissing values:")
print(hourly_data.isnull().sum())
```

```python
# تبدیل timestamp به datetime برای راحتی کار
# چون تاریخ جلالی است، از index عددی استفاده می‌کنیم
hourly_data['time_index'] = range(len(hourly_data))

# ویژگی‌های زمانی
# اولین روز داده (1397-11-01) چهارشنبه است، پس با 0 شروع می‌کنیم
hourly_data['day_of_week'] = (hourly_data['time_index'] // 24) % 7  # 0=چهارشنبه, 1=پنجشنبه, ...
hourly_data['day_of_year'] = hourly_data['time_index'] // 24  # روز سال از شروع داده
hourly_data['hour_of_day'] = hourly_data['hour']  # ساعت روز (0-23)

# ویژگی‌های دوره‌ای (cyclic) برای ساعت
hourly_data['hour_sin'] = np.sin(2 * np.pi * hourly_data['hour'] / 24)
hourly_data['hour_cos'] = np.cos(2 * np.pi * hourly_data['hour'] / 24)

# ویژگی‌های دوره‌ای برای روز هفته
hourly_data['dow_sin'] = np.sin(2 * np.pi * hourly_data['day_of_week'] / 7)
hourly_data['dow_cos'] = np.cos(2 * np.pi * hourly_data['day_of_week'] / 7)

# آیا روز تعطیل است؟ پنجشنبه (1) و جمعه (2)
# چهارشنبه=0, پنجشنبه=1, جمعه=2, شنبه=3, یکشنبه=4, دوشنبه=5, سه‌شنبه=6
hourly_data['is_weekend'] = ((hourly_data['day_of_week'] == 1) | 
                              (hourly_data['day_of_week'] == 2)).astype(int)

print("Feature engineering completed!")
print("\nDataframe shape:", hourly_data.shape)
print("\n" + "="*50)

# نمایش چند رکورد با ویژگی‌های جدید
print("\nSample data with new features:")
print(hourly_data[['timestamp_hourly', 'total', 'hour', 'day_of_week', 
                    'is_weekend', 'hour_sin', 'hour_cos']].head(48))
print("\n" + "="*50)

# بررسی تعداد روزهای تعطیل
print(f"\nWeekend days: {hourly_data['is_weekend'].sum() / 24:.0f} days")
print(f"Weekday days: {(1 - hourly_data['is_weekend']).sum() / 24:.0f} days")
print(f"Total days: {len(hourly_data) / 24:.0f} days")
print("\n" + "="*50)

# بررسی همبستگی ساعت روز با مصرف
print("\nCorrelation with total:")
corr_cols = ['hour', 'day_of_week', 'is_weekend', 'hour_sin', 'hour_cos', 
             'dow_sin', 'dow_cos', 'time_index']
print(hourly_data[corr_cols + ['total']].corr()['total'].sort_values(ascending=False))
```

```python
# Lag features - مقادیر ساعات قبل
for lag in [1, 2, 3, 24, 48, 168]:  # 1h, 2h, 3h, 1day, 2day, 1week
    hourly_data[f'lag_{lag}'] = hourly_data['total'].shift(lag)

# Rolling statistics - میانگین و انحراف معیار دوره‌های گذشته
for window in [24, 168]:  # 1 روز، 1 هفته
    hourly_data[f'rolling_mean_{window}'] = hourly_data['total'].shift(1).rolling(window=window).mean()
    hourly_data[f'rolling_std_{window}'] = hourly_data['total'].shift(1).rolling(window=window).std()

# Rolling min/max برای شناسایی اوج و فرود
hourly_data['rolling_min_24'] = hourly_data['total'].shift(1).rolling(window=24).min()
hourly_data['rolling_max_24'] = hourly_data['total'].shift(1).rolling(window=24).max()

# Difference features - تغییرات نسبت به گذشته
hourly_data['diff_1'] = hourly_data['total'].diff(1)
hourly_data['diff_24'] = hourly_data['total'].diff(24)
hourly_data['diff_168'] = hourly_data['total'].diff(168)

print("Lag and rolling features created!")
print("\nDataframe shape:", hourly_data.shape)
print("\n" + "="*50)

# بررسی missing values بعد از lag features
print("\nMissing values after lag features:")
print(hourly_data.isnull().sum().sort_values(ascending=False).head(15))
print("\n" + "="*50)

# نمایش نمونه داده با lag features
print("\nSample data with lag features:")
cols_to_show = ['timestamp_hourly', 'total', 'lag_1', 'lag_24', 'rolling_mean_24', 
                'diff_1', 'diff_24']
print(hourly_data[cols_to_show].iloc[170:180])
```

```python
# بررسی تعداد missing values
print("Total records:", len(hourly_data))
print("Records with any missing value:", hourly_data.isnull().any(axis=1).sum())
print("Clean records (no missing):", (~hourly_data.isnull().any(axis=1)).sum())
print("\n" + "="*50)

# بررسی اولین و آخرین تاریخ بدون missing
clean_data = hourly_data.dropna()
print("\nClean data date range:")
print(f"Start: {clean_data['timestamp_hourly'].iloc[0]}")
print(f"End: {clean_data['timestamp_hourly'].iloc[-1]}")
print(f"Total clean records: {len(clean_data)}")
print(f"Lost records: {len(hourly_data) - len(clean_data)}")
print(f"Percentage kept: {len(clean_data) / len(hourly_data) * 100:.2f}%")
```

```python
# حذف missing values
hourly_data_clean = hourly_data.dropna().reset_index(drop=True)

print("Clean data shape:", hourly_data_clean.shape)
print(f"Date range: {hourly_data_clean['timestamp_hourly'].iloc[0]} to {hourly_data_clean['timestamp_hourly'].iloc[-1]}")
print("\n" + "="*50)

# مشخص کردن feature columns
feature_cols = [
    'time_index', 'hour', 'day_of_week', 'day_of_year', 'hour_of_day',
    'hour_sin', 'hour_cos', 'dow_sin', 'dow_cos', 'is_weekend',
    'lag_1', 'lag_2', 'lag_3', 'lag_24', 'lag_48', 'lag_168',
    'rolling_mean_24', 'rolling_std_24', 'rolling_mean_168', 'rolling_std_168',
    'rolling_min_24', 'rolling_max_24',
    'diff_1', 'diff_24', 'diff_168'
]

target_col = 'total'

# بررسی وجود همه feature ها
print("Feature columns:")
for col in feature_cols:
    if col in hourly_data_clean.columns:
        print(f"  ✓ {col}")
    else:
        print(f"  ✗ {col} - MISSING!")
print("\n" + "="*50)

# Split: 80% train, 20% validation (آخرین 20% برای validation)
split_idx = int(len(hourly_data_clean) * 0.8)

train_data = hourly_data_clean.iloc[:split_idx].copy()
val_data = hourly_data_clean.iloc[split_idx:].copy()

print(f"\nTrain data: {len(train_data)} records")
print(f"  Date range: {train_data['timestamp_hourly'].iloc[0]} to {train_data['timestamp_hourly'].iloc[-1]}")
print(f"\nValidation data: {len(val_data)} records")
print(f"  Date range: {val_data['timestamp_hourly'].iloc[0]} to {val_data['timestamp_hourly'].iloc[-1]}")
print("\n" + "="*50)

# ایجاد X و y
X_train = train_data[feature_cols]
y_train = train_data[target_col]

X_val = val_data[feature_cols]
y_val = val_data[target_col]

print(f"\nX_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_val shape: {X_val.shape}")
print(f"y_val shape: {y_val.shape}")
```

```python
import lightgbm as lgb
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

# ایجاد مدل LightGBM
lgb_model = lgb.LGBMRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=8,
    num_leaves=64,
    min_child_samples=20,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    verbose=-1
)

print("Training LightGBM model...")
print("="*50)

# آموزش مدل با early stopping
lgb_model.fit(
    X_train, y_train,
    eval_set=[(X_val, y_val)],
    eval_metric='rmse',
    callbacks=[lgb.early_stopping(stopping_rounds=50), lgb.log_evaluation(period=100)]
)

print("\n" + "="*50)
print("Training completed!")
print("\n" + "="*50)

# پیش‌بینی روی train و validation
y_train_pred = lgb_model.predict(X_train)
y_val_pred = lgb_model.predict(X_val)

# محاسبه metrics
train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
val_rmse = np.sqrt(mean_squared_error(y_val, y_val_pred))
train_mae = mean_absolute_error(y_train, y_train_pred)
val_mae = mean_absolute_error(y_val, y_val_pred)

# محاسبه امتیاز مسابقه
val_std = y_val.std()
score = 100 * np.exp(-val_rmse / val_std)

print("\n📊 Model Performance:")
print(f"  Train RMSE: {train_rmse:,.2f}")
print(f"  Train MAE:  {train_mae:,.2f}")
print(f"  Val RMSE:   {val_rmse:,.2f}")
print(f"  Val MAE:    {val_mae:,.2f}")
print(f"  Val Std:    {val_std:,.2f}")
print(f"\n🏆 Competition Score: {score:.2f}")
print("="*50)

# Feature importance
print("\n📈 Top 15 Most Important Features:")
feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': lgb_model.feature_importances_
}).sort_values('importance', ascending=False)

print(feature_importance.head(15).to_string(index=False))
```

```python
# تاریخ‌های آذر 1401 (1401-09-01 to 1401-09-30)
# آخرین تاریخ در داده: 1401-08-30 23:00:00
# اولین تاریخ پیش‌بینی: 1401-09-01 00:00:00

# ایجاد لیست تاریخ‌های آذر 1401
prediction_dates = []
for day in range(1, 31):  # 30 روز آذر
    for hour in range(24):  # 24 ساعت
        prediction_dates.append(f"1401-09-{str(day).zfill(2)} {str(hour).zfill(2)}:00:00")

print(f"Total prediction timestamps: {len(prediction_dates)}")
print(f"First: {prediction_dates[0]}")
print(f"Last: {prediction_dates[-1]}")
print("\n" + "="*50)

# آخرین داده‌های موجود را برای ساخت lag features استفاده می‌کنیم
# باید از تمام داده (train + val) استفاده کنیم
full_data = hourly_data_clean.copy()
print(f"\nFull historical data: {len(full_data)} records")
print(f"Last timestamp: {full_data['timestamp_hourly'].iloc[-1]}")
print(f"Last total value: {full_data['total'].iloc[-1]:,.2f}")
print("\n" + "="*50)

# ایجاد dataframe برای predictions
prediction_df = pd.DataFrame({'timestamp_hourly': prediction_dates})

# محاسبه time_index برای ماه آینده
last_time_index = full_data['time_index'].iloc[-1]
prediction_df['time_index'] = range(last_time_index + 1, last_time_index + 1 + len(prediction_df))

# استخراج اجزای تاریخ
prediction_df['year'] = 1401
prediction_df['month'] = 9
prediction_df['day'] = prediction_df['timestamp_hourly'].str[8:10].astype(int)
prediction_df['hour'] = prediction_df['timestamp_hourly'].str[11:13].astype(int)

# ویژگی‌های زمانی
prediction_df['day_of_year'] = prediction_df['time_index'] // 24
prediction_df['day_of_week'] = (prediction_df['time_index'] // 24) % 7
prediction_df['hour_of_day'] = prediction_df['hour']
prediction_df['hour_sin'] = np.sin(2 * np.pi * prediction_df['hour'] / 24)
prediction_df['hour_cos'] = np.cos(2 * np.pi * prediction_df['hour'] / 24)
prediction_df['dow_sin'] = np.sin(2 * np.pi * prediction_df['day_of_week'] / 7)
prediction_df['dow_cos'] = np.cos(2 * np.pi * prediction_df['day_of_week'] / 7)
prediction_df['is_weekend'] = ((prediction_df['day_of_week'] == 1) | 
                                (prediction_df['day_of_week'] == 2)).astype(int)

print("\nPrediction dataframe created!")
print(prediction_df[['timestamp_hourly', 'time_index', 'hour', 'day_of_week', 'is_weekend']].head(10))
```

```python
# ترکیب داده‌های تاریخی با prediction_df
# برای محاسبه lag features از آخرین داده‌های واقعی استفاده می‌کنیم
combined_data = pd.concat([full_data, prediction_df], ignore_index=True)

print("Combined data shape:", combined_data.shape)
print(f"Historical records: {len(full_data)}")
print(f"Future records: {len(prediction_df)}")
print("\n" + "="*50)

# Iterative prediction
predictions = []

for i in range(len(prediction_df)):
    current_idx = len(full_data) + i  # index در combined_data
    
    # محاسبه lag features از داده‌های قبلی
    # برای اولین پیش‌بینی، از داده واقعی استفاده می‌کنیم
    # برای بقیه، از پیش‌بینی‌های قبلی
    
    if i == 0:
        # اولین پیش‌بینی - از داده واقعی lag می‌گیریم
        lag_1 = full_data['total'].iloc[-1]
        lag_2 = full_data['total'].iloc[-2]
        lag_3 = full_data['total'].iloc[-3]
        lag_24 = full_data['total'].iloc[-24]
        lag_48 = full_data['total'].iloc[-48]
        lag_168 = full_data['total'].iloc[-168]
    else:
        # از ترکیبی از داده واقعی و پیش‌بینی‌های قبلی استفاده می‌کنیم
        if i >= 1:
            lag_1 = predictions[i-1]
        else:
            lag_1 = full_data['total'].iloc[-1]
            
        if i >= 2:
            lag_2 = predictions[i-2]
        else:
            lag_2 = full_data['total'].iloc[-(2-i)-1]
            
        if i >= 3:
            lag_3 = predictions[i-3]
        else:
            lag_3 = full_data['total'].iloc[-(3-i)-1]
            
        if i >= 24:
            lag_24 = predictions[i-24]
        else:
            lag_24 = full_data['total'].iloc[-(24-i)-1]
            
        if i >= 48:
            lag_48 = predictions[i-48]
        else:
            lag_48 = full_data['total'].iloc[-(48-i)-1]
            
        if i >= 168:
            lag_168 = predictions[i-168]
        else:
            lag_168 = full_data['total'].iloc[-(168-i)-1]
    
    # محاسبه rolling statistics و diff features
    # برای rolling از ترکیب داده واقعی + پیش‌بینی‌های قبلی استفاده می‌کنیم
    recent_values = list(full_data['total'].tail(168)) + predictions
    
    rolling_mean_24 = np.mean(recent_values[-24:]) if len(recent_values) >= 24 else np.mean(full_data['total'].tail(24))
    rolling_std_24 = np.std(recent_values[-24:]) if len(recent_values) >= 24 else np.std(full_data['total'].tail(24))
    rolling_min_24 = np.min(recent_values[-24:]) if len(recent_values) >= 24 else np.min(full_data['total'].tail(24))
    rolling_max_24 = np.max(recent_values[-24:]) if len(recent_values) >= 24 else np.max(full_data['total'].tail(24))
    
    rolling_mean_168 = np.mean(recent_values[-168:]) if len(recent_values) >= 168 else np.mean(full_data['total'].tail(168))
    rolling_std_168 = np.std(recent_values[-168:]) if len(recent_values) >= 168 else np.std(full_data['total'].tail(168))
    
    # محاسبه diff features
    diff_1 = lag_1 - (predictions[i-2] if i >= 2 else full_data['total'].iloc[-(2-i)-1])
    diff_24 = lag_1 - lag_24
    diff_168 = lag_1 - lag_168
    
    # ساخت feature vector
    features = {
        'time_index': prediction_df['time_index'].iloc[i],
        'hour': prediction_df['hour'].iloc[i],
        'day_of_week': prediction_df['day_of_week'].iloc[i],
        'day_of_year': prediction_df['day_of_year'].iloc[i],
        'hour_of_day': prediction_df['hour_of_day'].iloc[i],
        'hour_sin': prediction_df['hour_sin'].iloc[i],
        'hour_cos': prediction_df['hour_cos'].iloc[i],
        'dow_sin': prediction_df['dow_sin'].iloc[i],
        'dow_cos': prediction_df['dow_cos'].iloc[i],
        'is_weekend': prediction_df['is_weekend'].iloc[i],
        'lag_1': lag_1,
        'lag_2': lag_2,
        'lag_3': lag_3,
        'lag_24': lag_24,
        'lag_48': lag_48,
        'lag_168': lag_168,
        'rolling_mean_24': rolling_mean_24,
        'rolling_std_24': rolling_std_24,
        'rolling_mean_168': rolling_mean_168,
        'rolling_std_168': rolling_std_168,
        'rolling_min_24': rolling_min_24,
        'rolling_max_24': rolling_max_24,
        'diff_1': diff_1,
        'diff_24': diff_24,
        'diff_168': diff_168
    }
    
    # پیش‌بینی
    X_pred = pd.DataFrame([features])[feature_cols]
    pred = lgb_model.predict(X_pred)[0]
    predictions.append(pred)
    
    # نمایش پیشرفت
    if (i + 1) % 72 == 0:  # هر 3 روز
        print(f"Predicted {i+1}/{len(prediction_df)} hours ({(i+1)/len(prediction_df)*100:.1f}%)")

print("\n" + "="*50)
print("✅ Prediction completed!")
print(f"Total predictions: {len(predictions)}")
print(f"Mean prediction: {np.mean(predictions):,.2f}")
print(f"Std prediction: {np.std(predictions):,.2f}")
print(f"Min prediction: {np.min(predictions):,.2f}")
print(f"Max prediction: {np.max(predictions):,.2f}")
```
