```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

train = pd.read_csv(r'F:\Quera\olympic 2024\crime_detection\data\train.csv')
test = pd.read_csv(r'F:\Quera\olympic 2024\crime_detection\data\test.csv')

print("="*70)
print("TRAIN DATA INFO")
print("="*70)
print(f"Shape: {train.shape}")
print(f"\nColumn Names and Types:")
print(train.dtypes)
print(f"\n\nFirst 3 rows:")
print(train.head(3))

print("\n" + "="*70)
print("TEST DATA INFO")
print("="*70)
print(f"Shape: {test.shape}")
print(f"\nColumn Names:")
print(test.columns.tolist())

print("\n" + "="*70)
print("MISSING VALUES - TRAIN")
print("="*70)
missing_train = train.isnull().sum()
missing_train_pct = (train.isnull().sum() / len(train)) * 100
missing_df = pd.DataFrame({
    'Missing_Count': missing_train,
    'Missing_Percentage': missing_train_pct
}).sort_values('Missing_Percentage', ascending=False)
print(missing_df[missing_df['Missing_Count'] > 0])

print("\n" + "="*70)
print("TARGET VARIABLE DISTRIBUTION")
print("="*70)
target_dist = train['Crm Cd'].value_counts()
print(f"Number of unique crime codes: {train['Crm Cd'].nunique()}")
print(f"\nTop 10 crime codes:")
print(target_dist.head(10))
print(f"\nBottom 10 crime codes:")
print(target_dist.tail(10))
print(f"\nClass imbalance ratio (max/min): {target_dist.max() / target_dist.min():.2f}")
```

```python
print("="*70)
print("STEP 1: INITIAL DATA CLEANING & VERIFICATION")
print("="*70)

# Make copies
train_clean = train.copy()
test_clean = test.copy()

# Check if Crm Cd == Crm Cd 1
print("\n1. Checking Crm Cd vs Crm Cd 1:")
print(f"Total non-null Crm Cd 1: {train_clean['Crm Cd 1'].notna().sum()}")
crm_cd_match = (train_clean['Crm Cd'] == train_clean['Crm Cd 1']).sum()
crm_cd_total = train_clean['Crm Cd 1'].notna().sum()
print(f"Matching records: {crm_cd_match} / {crm_cd_total}")
print(f"Match percentage: {(crm_cd_match/crm_cd_total)*100:.2f}%")

# Check mismatches
mismatches = train_clean[
    (train_clean['Crm Cd'] != train_clean['Crm Cd 1']) & 
    (train_clean['Crm Cd 1'].notna())
]
print(f"Mismatched records: {len(mismatches)}")
if len(mismatches) > 0:
    print("\nSample mismatches:")
    print(mismatches[['Crm Cd', 'Crm Cd 1', 'Crm Cd 2', 'Crm Cd Desc']].head(10))

# Analyze TIME OCC format
print("\n2. Analyzing TIME OCC format:")
print(f"Min: {train_clean['TIME OCC'].min()}")
print(f"Max: {train_clean['TIME OCC'].max()}")
print(f"Sample values: {train_clean['TIME OCC'].head(20).tolist()}")

# Extract hour from TIME OCC
train_clean['Hour'] = train_clean['TIME OCC'] // 100
test_clean['Hour'] = test_clean['TIME OCC'] // 100
print(f"\nExtracted Hour range: {train_clean['Hour'].min()} - {train_clean['Hour'].max()}")
print(f"Invalid hours (>23): {(train_clean['Hour'] > 23).sum()}")

print("\n3. Column redundancy check complete!")
```

```python
print("="*70)
print("STEP 2: COMPLETE FEATURE ENGINEERING & CLEANING")
print("="*70)

# Make clean copies
train_clean = train.copy()
test_clean = test.copy()

# ============================================================
# PART 1: DATE & TIME FEATURE ENGINEERING
# ============================================================
print("\n[1/5] Processing dates and extracting temporal features...")

# Convert dates to datetime
train_clean['Date Rptd'] = pd.to_datetime(train_clean['Date Rptd'])
train_clean['DATE OCC'] = pd.to_datetime(train_clean['DATE OCC'])
test_clean['Date Rptd'] = pd.to_datetime(test_clean['Date Rptd'])
test_clean['DATE OCC'] = pd.to_datetime(test_clean['DATE OCC'])

# Extract temporal features from DATE OCC
for df in [train_clean, test_clean]:
    df['Year'] = df['DATE OCC'].dt.year
    df['Month'] = df['DATE OCC'].dt.month
    df['Day'] = df['DATE OCC'].dt.day
    df['DayOfWeek'] = df['DATE OCC'].dt.dayofweek  # 0=Monday, 6=Sunday
    df['Quarter'] = df['DATE OCC'].dt.quarter
    df['DayOfYear'] = df['DATE OCC'].dt.dayofyear

# Calculate report delay
for df in [train_clean, test_clean]:
    df['Report_Delay_Days'] = (df['Date Rptd'] - df['DATE OCC']).dt.days

print("  ✓ Date features extracted")
print("  ✓ Report delay calculated")

# ============================================================
# PART 2: TIME OCC PROCESSING
# ============================================================
print("\n[2/5] Processing TIME OCC...")

# Extract hour
for df in [train_clean, test_clean]:
    df['Hour'] = df['TIME OCC'] // 100

# Create time period categories
def get_time_period(hour):
    if 6 <= hour < 12:
        return 0  # Morning
    elif 12 <= hour < 18:
        return 1  # Afternoon
    elif 18 <= hour < 22:
        return 2  # Evening
    else:
        return 3  # Night

for df in [train_clean, test_clean]:
    df['Time_Period'] = df['Hour'].apply(get_time_period)

print("  ✓ Hour extracted from TIME OCC")
print("  ✓ Time_Period created")

# ============================================================
# PART 3: DROP REDUNDANT/USELESS COLUMNS
# ============================================================
print("\n[3/5] Dropping redundant and useless columns...")

columns_to_drop = [
    # IDs
    'DR_NO',
    
    # Already extracted features (source columns)
    'Date Rptd',
    'DATE OCC',
    'TIME OCC',         # Extracted Hour and Time_Period
    
    # Redundant text descriptions (have codes)
    'AREA NAME',        # Have AREA code
    'Crm Cd Desc',      # Have Crm Cd (target) - only in train
    'Premis Desc',      # Have Premis Cd
    'Weapon Desc',      # Have Weapon Used Cd
    'Status Desc',      # Have Status code
    
    # Not in test set
    'Crm Cd 1',         # 99.97% same as Crm Cd, not in test
    'Crm Cd 2',         # 93% missing, not in test
    'Crm Cd 3',         # 99.9% missing, not in test
    'Crm Cd 4',         # 99.99% missing, not in test
    
    # Geographic redundancy
    'LOCATION',         # Have LAT/LON
    'Cross Street'      # 82% missing, not useful
]

# Drop columns that exist in each dataset
train_cols_drop = [col for col in columns_to_drop if col in train_clean.columns]
test_cols_drop = [col for col in columns_to_drop if col in test_clean.columns]

train_clean = train_clean.drop(columns=train_cols_drop)
test_clean = test_clean.drop(columns=test_cols_drop)

print(f"  ✓ Dropped {len(train_cols_drop)} columns from train")
print(f"  ✓ Dropped {len(test_cols_drop)} columns from test")

# ============================================================
# PART 4: VERIFY CONSISTENCY
# ============================================================
print("\n[4/5] Verifying train-test consistency...")

train_cols_no_target = [col for col in train_clean.columns if col != 'Crm Cd']
test_cols = test_clean.columns.tolist()

if set(train_cols_no_target) == set(test_cols):
    print("  ✓ Train and test have same features (excluding target)")
else:
    print("  ⚠ WARNING: Feature mismatch!")
    print(f"    Only in train: {set(train_cols_no_target) - set(test_cols)}")
    print(f"    Only in test: {set(test_cols) - set(train_cols_no_target)}")

# ============================================================
# PART 5: SUMMARY
# ============================================================
print("\n[5/5] Summary:")
print("="*70)
print(f"Train shape: {train_clean.shape}")
print(f"Test shape:  {test_clean.shape}")

print(f"\nTrain columns ({len(train_clean.columns)}):")
print(train_clean.columns.tolist())

print(f"\nTest columns ({len(test_clean.columns)}):")
print(test_clean.columns.tolist())

print("\n" + "="*70)
print("DATA TYPES AFTER CLEANING")
print("="*70)
print(train_clean.dtypes)

print("\n" + "="*70)
print("✅ FEATURE ENGINEERING & CLEANING COMPLETE!")
print("="*70)
```

```python
print("="*70)
print("STEP 3: DETAILED FEATURE ANALYSIS BY DATA TYPE")
print("="*70)

# ============================================================
# PART 1: INTEGER FEATURES ANALYSIS
# ============================================================
print("\n[1/4] INTEGER FEATURES ANALYSIS")
print("="*70)

int_cols = ['AREA', 'Rpt Dist No', 'Vict Age', 'Hour', 'Year', 'Month', 
            'Day', 'DayOfWeek', 'Quarter', 'DayOfYear', 'Report_Delay_Days', 'Time_Period']

for col in int_cols:
    print(f"\n{col}:")
    print(f"  Unique values: {train_clean[col].nunique()}")
    print(f"  Range: [{train_clean[col].min()}, {train_clean[col].max()}]")
    print(f"  Missing: {train_clean[col].isna().sum()}")
    
    # Show distribution for low cardinality
    if train_clean[col].nunique() <= 25:
        print(f"  Value counts (top 10):")
        print(train_clean[col].value_counts().head(10).to_dict())
    
    # Check for anomalies
    if col == 'Vict Age':
        anomaly_count = ((train_clean[col] < 0) | (train_clean[col] > 120)).sum()
        print(f"  Anomalies (age < 0 or > 120): {anomaly_count}")
        if anomaly_count > 0:
            print(f"  Sample anomalies: {train_clean[train_clean[col] < 0][col].head().tolist()}")
            print(f"  Sample anomalies: {train_clean[train_clean[col] > 120][col].head().tolist()}")
    
    if col == 'Report_Delay_Days':
        negative = (train_clean[col] < 0).sum()
        extreme = (train_clean[col] > 365).sum()
        print(f"  Negative delays: {negative}")
        print(f"  Delays > 1 year: {extreme}")

print("\n" + "="*70)

# ============================================================
# PART 2: FLOAT FEATURES ANALYSIS
# ============================================================
print("\n[2/4] FLOAT FEATURES ANALYSIS")
print("="*70)

float_cols = ['Premis Cd', 'Weapon Used Cd', 'LAT', 'LON']

for col in float_cols:
    print(f"\n{col}:")
    print(f"  Unique values: {train_clean[col].nunique()}")
    print(f"  Missing count: {train_clean[col].isna().sum()} ({train_clean[col].isna().sum()/len(train_clean)*100:.2f}%)")
    print(f"  Range: [{train_clean[col].min():.4f}, {train_clean[col].max():.4f}]")
    
    # Check for zeros (potential missing indicators)
    zero_count = (train_clean[col] == 0).sum()
    print(f"  Zero values: {zero_count}")
    
    if train_clean[col].nunique() <= 50:
        print(f"  Top 10 values:")
        print(train_clean[col].value_counts().head(10).to_dict())
    
    # Geographic coordinate checks
    if col in ['LAT', 'LON']:
        # LA coordinates roughly: LAT 33.7-34.3, LON -118.7 to -118.1
        if col == 'LAT':
            invalid = ((train_clean[col] < 33) | (train_clean[col] > 35)).sum()
        else:  # LON
            invalid = ((train_clean[col] < -119) | (train_clean[col] > -117)).sum()
        print(f"  Potentially invalid coordinates: {invalid}")
        
        # Check for (0,0) or missing coordinates
        zero_coords = (train_clean[col] == 0).sum()
        print(f"  Zero coordinates: {zero_coords}")

print("\n" + "="*70)
```

```python
print("="*70)
print("STEP 4: DATA CLEANING & FEATURE REFINEMENT (REVISED V2)")
print("="*70)

# ============================================================
# PART 1: ANALYZE WEAPON USED CD FOR FEATURE ENGINEERING
# ============================================================
print("\n[1/9] Analyzing Weapon Used Cd for feature engineering...")

# Check if some rows have multiple weapons
print("Sample non-null Weapon Used Cd:")
print(train_clean[train_clean['Weapon Used Cd'].notna()]['Weapon Used Cd'].head(20).tolist())

# Create binary features BEFORE filling
train_clean['Has_Weapon'] = train_clean['Weapon Used Cd'].notna().astype(int)
train_clean['Weapon_Count'] = train_clean['Weapon Used Cd'].notna().astype(int)

test_clean['Has_Weapon'] = test_clean['Weapon Used Cd'].notna().astype(int)
test_clean['Weapon_Count'] = test_clean['Weapon Used Cd'].notna().astype(int)

print(f"✓ Created 'Has_Weapon' and 'Weapon_Count' features")
print(f"Train - Has weapon: {train_clean['Has_Weapon'].sum()} ({train_clean['Has_Weapon'].mean()*100:.2f}%)")
print(f"Test  - Has weapon: {test_clean['Has_Weapon'].sum()} ({test_clean['Has_Weapon'].mean()*100:.2f}%)")

# Fill NaN in Weapon Used Cd with -1
train_clean['Weapon Used Cd'] = train_clean['Weapon Used Cd'].fillna(-1)
test_clean['Weapon Used Cd'] = test_clean['Weapon Used Cd'].fillna(-1)

print("✓ Filled missing Weapon Used Cd with -1")

# ============================================================
# PART 2: HANDLE VICTIM AGE - REMOVE NEGATIVE, FILL ZERO
# ============================================================
print("\n[2/9] Handling Victim Age anomalies...")

# Check current status
print(f"Train - Age < 0: {(train_clean['Vict Age'] < 0).sum()}")
print(f"Train - Age = 0: {(train_clean['Vict Age'] == 0).sum()}")
print(f"Test  - Age < 0: {(test_clean['Vict Age'] < 0).sum()}")
print(f"Test  - Age = 0: {(test_clean['Vict Age'] == 0).sum()}")

# Remove negative ages
train_before = len(train_clean)
test_before = len(test_clean)

train_clean = train_clean[train_clean['Vict Age'] >= 0].reset_index(drop=True)
test_clean = test_clean[test_clean['Vict Age'] >= 0].reset_index(drop=True)

print(f"✓ Removed negative ages: Train={train_before - len(train_clean)}, Test={test_before - len(test_clean)}")

# Fill age = 0 with median of valid ages
valid_ages_median = train_clean[train_clean['Vict Age'] > 0]['Vict Age'].median()
print(f"Valid ages median: {valid_ages_median}")

train_clean.loc[train_clean['Vict Age'] == 0, 'Vict Age'] = valid_ages_median
test_clean.loc[test_clean['Vict Age'] == 0, 'Vict Age'] = valid_ages_median

print(f"✓ Filled age=0 with median: {valid_ages_median}")
print(f"Train new shape: {train_clean.shape}")
print(f"Test new shape:  {test_clean.shape}")

# ============================================================
# PART 3: HANDLE LAT/LON ZEROS - FILL WITH AREA MEDIAN
# ============================================================
print("\n[3/9] Handling zero coordinates...")

train_zero_coords = ((train_clean['LAT'] == 0) | (train_clean['LON'] == 0)).sum()
test_zero_coords = ((test_clean['LAT'] == 0) | (test_clean['LON'] == 0)).sum()

print(f"Train - Zero coordinates: {train_zero_coords}")
print(f"Test  - Zero coordinates: {test_zero_coords}")

# Calculate median LAT/LON per AREA (from non-zero values)
area_median_lat = train_clean[train_clean['LAT'] != 0].groupby('AREA')['LAT'].median()
area_median_lon = train_clean[train_clean['LON'] != 0].groupby('AREA')['LON'].median()

# Overall median as fallback
overall_median_lat = train_clean[train_clean['LAT'] != 0]['LAT'].median()
overall_median_lon = train_clean[train_clean['LON'] != 0]['LON'].median()

# Fill zeros in train
for area in train_clean['AREA'].unique():
    mask = (train_clean['AREA'] == area) & ((train_clean['LAT'] == 0) | (train_clean['LON'] == 0))
    if mask.any():
        train_clean.loc[mask, 'LAT'] = area_median_lat.get(area, overall_median_lat)
        train_clean.loc[mask, 'LON'] = area_median_lon.get(area, overall_median_lon)

# Fill zeros in test
for area in test_clean['AREA'].unique():
    mask = (test_clean['AREA'] == area) & ((test_clean['LAT'] == 0) | (test_clean['LON'] == 0))
    if mask.any():
        test_clean.loc[mask, 'LAT'] = area_median_lat.get(area, overall_median_lat)
        test_clean.loc[mask, 'LON'] = area_median_lon.get(area, overall_median_lon)

print("✓ Filled zero coordinates with AREA-based medians")

# ============================================================
# PART 4: HANDLE PREMIS CD - FILL WITH -1
# ============================================================
print("\n[4/9] Handling Premis Cd...")

print(f"Train - Premis Cd NaN: {train_clean['Premis Cd'].isna().sum()}")
print(f"Train - Premis Cd = 0: {(train_clean['Premis Cd'] == 0).sum()}")
print(f"Test  - Premis Cd NaN: {test_clean['Premis Cd'].isna().sum()}")
print(f"Test  - Premis Cd = 0: {(test_clean['Premis Cd'] == 0).sum()}")

# Fill NaN and 0 with -1 (unknown premise)
train_clean['Premis Cd'] = train_clean['Premis Cd'].fillna(-1)
train_clean.loc[train_clean['Premis Cd'] == 0, 'Premis Cd'] = -1

test_clean['Premis Cd'] = test_clean['Premis Cd'].fillna(-1)
test_clean.loc[test_clean['Premis Cd'] == 0, 'Premis Cd'] = -1

print("✓ Filled Premis Cd NaN and 0 with -1")

# ============================================================
# PART 5: DROP REDUNDANT TEMPORAL FEATURES
# ============================================================
print("\n[5/9] Removing redundant features...")

# Drop Year (all 2020), DayOfYear (redundant), Time_Period (redundant with Hour)
redundant_cols = ['Year', 'DayOfYear', 'Time_Period']

train_clean = train_clean.drop(columns=redundant_cols)
test_clean = test_clean.drop(columns=redundant_cols)

print(f"✓ Dropped redundant temporal features: {redundant_cols}")

# ============================================================
# PART 6: VERIFY NO MISSING VALUES
# ============================================================
print("\n[6/9] Verifying data integrity...")
print("="*70)

print("\nMissing values in train:")
missing_train = train_clean.isnull().sum()
if missing_train.sum() == 0:
    print("✓ NO MISSING VALUES!")
else:
    print(missing_train[missing_train > 0])

print("\nMissing values in test:")
missing_test = test_clean.isnull().sum()
if missing_test.sum() == 0:
    print("✓ NO MISSING VALUES!")
else:
    print(missing_test[missing_test > 0])

# ============================================================
# PART 7: VERIFY ANOMALIES ARE FIXED
# ============================================================
print("\n[7/9] Verifying anomalies are fixed:")
print("="*70)

print(f"Vict Age < 0: Train={(train_clean['Vict Age'] < 0).sum()}, Test={(test_clean['Vict Age'] < 0).sum()}")
print(f"Vict Age = 0: Train={(train_clean['Vict Age'] == 0).sum()}, Test={(test_clean['Vict Age'] == 0).sum()}")
print(f"LAT = 0: Train={(train_clean['LAT'] == 0).sum()}, Test={(test_clean['LAT'] == 0).sum()}")
print(f"LON = 0: Train={(train_clean['LON'] == 0).sum()}, Test={(test_clean['LON'] == 0).sum()}")
print(f"Premis Cd = 0 or NaN: Train={(train_clean['Premis Cd'] == 0).sum() + train_clean['Premis Cd'].isna().sum()}, Test={(test_clean['Premis Cd'] == 0).sum() + test_clean['Premis Cd'].isna().sum()}")

# ============================================================
# PART 8: CHECK DATA LOSS
# ============================================================
print("\n[8/9] Data loss analysis:")
print("="*70)

original_train_size = 84113  # From initial load
original_test_size = 9346

current_train_size = len(train_clean)
current_test_size = len(test_clean)

train_loss = original_train_size - current_train_size
test_loss = original_test_size - current_test_size

train_loss_pct = (train_loss / original_train_size) * 100
test_loss_pct = (test_loss / original_test_size) * 100

print(f"Train: {original_train_size} → {current_train_size} (lost {train_loss} rows, {train_loss_pct:.2f}%)")
print(f"Test:  {original_test_size} → {current_test_size} (lost {test_loss} rows, {test_loss_pct:.2f}%)")

# ============================================================
# PART 9: FINAL SUMMARY
# ============================================================
print("\n[9/9] Final Summary:")
print("="*70)
print(f"Final Train shape: {train_clean.shape}")
print(f"Final Test shape:  {test_clean.shape}")

print("\nCurrent features:")
feature_cols = [col for col in train_clean.columns if col != 'Crm Cd']
print(feature_cols)

print("\nData types:")
print(train_clean.dtypes)

print("\n" + "="*70)
print("✅ DATA CLEANING COMPLETE - READY FOR ENCODING!")
print("="*70)

# Quick check on target distribution after cleaning
if 'Crm Cd' in train_clean.columns:
    print("\nTarget distribution after cleaning:")
    target_dist = train_clean['Crm Cd'].value_counts().sort_index()
    for code, count in target_dist.items():
        pct = (count / len(train_clean)) * 100
        print(f"  {code}: {count:5d} ({pct:5.2f}%)")
    
    print(f"\nTotal classes: {train_clean['Crm Cd'].nunique()}")
    print(f"Imbalance ratio: {target_dist.max() / target_dist.min():.2f}")
```

```python
print("="*70)
print("STEP 5: OBJECT FEATURES ANALYSIS & MISSING VALUE HANDLING")
print("="*70)

from sklearn.preprocessing import MultiLabelBinarizer

# ============================================================
# PART 1: ANALYZE OBJECT FEATURES
# ============================================================
print("\n[1/5] Analyzing Object (Categorical) Features...")
print("="*70)

object_cols = ['Mocodes', 'Vict Sex', 'Vict Descent', 'Status']

for col in object_cols:
    print(f"\n{col}:")
    print(f"  Train - Unique: {train_clean[col].nunique()}, Missing: {train_clean[col].isna().sum()} ({train_clean[col].isna().sum()/len(train_clean)*100:.2f}%)")
    print(f"  Test  - Unique: {test_clean[col].nunique()}, Missing: {test_clean[col].isna().sum()} ({test_clean[col].isna().sum()/len(test_clean)*100:.2f}%)")
    
    # Show value distribution
    print(f"  Train value counts (top 10):")
    print(train_clean[col].value_counts().head(10).to_dict())
    
    if test_clean[col].nunique() <= 20:
        print(f"  Test value counts:")
        print(test_clean[col].value_counts().to_dict())

print("\n" + "="*70)

# ============================================================
# PART 2: HANDLE VICT SEX MISSING VALUES
# ============================================================
print("\n[2/5] Handling Vict Sex missing values...")

print(f"Train missing: {train_clean['Vict Sex'].isna().sum()}")
print(f"Test missing: {test_clean['Vict Sex'].isna().sum()}")

# Fill with mode (most common value)
mode_sex = train_clean['Vict Sex'].mode()[0]
train_clean['Vict Sex'] = train_clean['Vict Sex'].fillna(mode_sex)
test_clean['Vict Sex'] = test_clean['Vict Sex'].fillna(mode_sex)

print(f"✓ Filled missing Vict Sex with mode: '{mode_sex}'")

# ============================================================
# PART 3: HANDLE VICT DESCENT MISSING VALUES
# ============================================================
print("\n[3/5] Handling Vict Descent missing values...")

print(f"Train missing: {train_clean['Vict Descent'].isna().sum()}")
print(f"Test missing: {test_clean['Vict Descent'].isna().sum()}")

# Fill with mode (most common value)
mode_descent = train_clean['Vict Descent'].mode()[0]
train_clean['Vict Descent'] = train_clean['Vict Descent'].fillna(mode_descent)
test_clean['Vict Descent'] = test_clean['Vict Descent'].fillna(mode_descent)

print(f"✓ Filled missing Vict Descent with mode: '{mode_descent}'")

# ============================================================
# PART 4: HANDLE MOCODES - MULTILABELBINARIZER
# ============================================================
print("\n[4/5] Processing Mocodes with MultiLabelBinarizer...")

# Sample analysis
print("\nSample Mocodes (non-null):")
mocodes_sample = train_clean[train_clean['Mocodes'].notna()]['Mocodes'].head(10).tolist()
for i, mc in enumerate(mocodes_sample, 1):
    print(f"  {i}. {mc}")

# Create Mocodes_Count BEFORE transformation
train_clean['Mocodes_Count'] = train_clean['Mocodes'].apply(
    lambda x: len(str(x).split()) if pd.notna(x) else 0
)
test_clean['Mocodes_Count'] = test_clean['Mocodes'].apply(
    lambda x: len(str(x).split()) if pd.notna(x) else 0
)

print(f"\nMocodes_Count statistics (train):")
print(train_clean['Mocodes_Count'].describe())

# Convert Mocodes to list format for MultiLabelBinarizer
print("\nConverting Mocodes to list format...")

def mocodes_to_list(x):
    """Convert Mocodes string to list of integers"""
    if pd.isna(x) or x == 'nan':
        return []
    try:
        return [int(code) for code in str(x).split()]
    except:
        return []

train_clean['Mocodes_List'] = train_clean['Mocodes'].apply(mocodes_to_list)
test_clean['Mocodes_List'] = test_clean['Mocodes'].apply(mocodes_to_list)

print("✓ Converted to list format")

# Apply MultiLabelBinarizer
print("\nApplying MultiLabelBinarizer...")
mlb = MultiLabelBinarizer()

# Fit on train
mocodes_train_encoded = mlb.fit_transform(train_clean['Mocodes_List'])
print(f"  Train: Generated {len(mlb.classes_)} binary features")

# Transform test
mocodes_test_encoded = mlb.transform(test_clean['Mocodes_List'])
print(f"  Test: Transformed using same {len(mlb.classes_)} features")

# Create DataFrames
mocodes_train_df = pd.DataFrame(
    mocodes_train_encoded,
    columns=[f'Mocode_{c}' for c in mlb.classes_],
    index=train_clean.index
)

mocodes_test_df = pd.DataFrame(
    mocodes_test_encoded,
    columns=[f'Mocode_{c}' for c in mlb.classes_],
    index=test_clean.index
)

print(f"\nMocode features shape: Train={mocodes_train_df.shape}, Test={mocodes_test_df.shape}")

# Check sparsity
sparsity = (mocodes_train_df.sum().sum()) / (mocodes_train_df.shape[0] * mocodes_train_df.shape[1])
print(f"Feature sparsity: {sparsity*100:.2f}% (non-zero values)")

# Drop original Mocodes and Mocodes_List columns
train_clean = train_clean.drop(columns=['Mocodes', 'Mocodes_List'])
test_clean = test_clean.drop(columns=['Mocodes', 'Mocodes_List'])

# Concatenate with main dataframes
train_clean = pd.concat([train_clean, mocodes_train_df], axis=1)
test_clean = pd.concat([test_clean, mocodes_test_df], axis=1)

print("✓ MultiLabelBinarizer features added to dataframes")
print("✓ Original Mocodes column dropped")

# Show top 10 most common Mocodes
print("\nTop 10 most common Mocodes:")
top_mocodes = mocodes_train_df.sum().sort_values(ascending=False).head(10)
for mocode, count in top_mocodes.items():
    pct = (count / len(train_clean)) * 100
    print(f"  {mocode}: {int(count)} ({pct:.2f}%)")

# ============================================================
# PART 5: FINAL VERIFICATION
# ============================================================
print("\n[5/5] Final verification:")
print("="*70)

print("\nTrain missing values:")
missing_train = train_clean.isnull().sum()
if missing_train.sum() == 0:
    print("✓ NO MISSING VALUES!")
else:
    print(missing_train[missing_train > 0])

print("\nTest missing values:")
missing_test = test_clean.isnull().sum()
if missing_test.sum() == 0:
    print("✓ NO MISSING VALUES!")
else:
    print(missing_test[missing_test > 0])

print("\n" + "="*70)
print("✅ ALL MISSING VALUES HANDLED!")
print("="*70)

print(f"\nFinal shapes:")
print(f"  Train: {train_clean.shape}")
print(f"  Test:  {test_clean.shape}")

print(f"\nTotal features: {train_clean.shape[1] - 1} (excluding target)")
print(f"  - Original features: ~20")
print(f"  - Mocode binary features: {len(mlb.classes_)}")
print(f"  - Mocodes_Count: 1")

# Feature categories summary
numeric_features = train_clean.select_dtypes(include=[np.number]).columns.tolist()
if 'Crm Cd' in numeric_features:
    numeric_features.remove('Crm Cd')
    
object_features = train_clean.select_dtypes(include=['object']).columns.tolist()

print(f"\nFeature types:")
print(f"  - Numeric: {len(numeric_features)}")
print(f"  - Object (categorical): {len(object_features)}")

if len(object_features) > 0:
    print(f"\nRemaining object features: {object_features}")
```

```python
print("="*70)
print("STEP 6: ENCODING CATEGORICAL FEATURES")
print("="*70)

from sklearn.preprocessing import LabelEncoder

# ============================================================
# PART 1: LABEL ENCODE REMAINING CATEGORICALS
# ============================================================
print("\n[1/3] Encoding categorical features...")

# Vict Sex
le_sex = LabelEncoder()
train_clean['Vict Sex'] = le_sex.fit_transform(train_clean['Vict Sex'])
test_clean['Vict Sex'] = le_sex.transform(test_clean['Vict Sex'])
print(f"✓ Encoded Vict Sex: {le_sex.classes_}")

# Vict Descent
le_descent = LabelEncoder()
train_clean['Vict Descent'] = le_descent.fit_transform(train_clean['Vict Descent'])
# Handle unseen labels in test
test_clean['Vict Descent'] = test_clean['Vict Descent'].apply(
    lambda x: x if x in le_descent.classes_ else le_descent.classes_[0]
)
test_clean['Vict Descent'] = le_descent.transform(test_clean['Vict Descent'])
print(f"✓ Encoded Vict Descent: {len(le_descent.classes_)} classes")

# Status
le_status = LabelEncoder()
train_clean['Status'] = le_status.fit_transform(train_clean['Status'])
test_clean['Status'] = le_status.transform(test_clean['Status'])
print(f"✓ Encoded Status: {le_status.classes_}")

# ============================================================
# PART 2: VERIFY ALL NUMERIC
# ============================================================
print("\n[2/3] Verifying all features are numeric...")

print(f"\nTrain dtypes:")
print(train_clean.dtypes.value_counts())

print(f"\nTest dtypes:")
print(test_clean.dtypes.value_counts())

object_cols = train_clean.select_dtypes(include='object').columns.tolist()
if 'Crm Cd' in object_cols:
    object_cols.remove('Crm Cd')

if len(object_cols) == 0:
    print("\n✅ ALL FEATURES ARE NUMERIC!")
else:
    print(f"\n⚠️ Object columns remain: {object_cols}")

# ============================================================
# PART 3: FINAL SUMMARY
# ============================================================
print("\n[3/3] Final summary:")
print("="*70)

print(f"\nFinal shapes:")
print(f"  Train: {train_clean.shape}")
print(f"  Test:  {test_clean.shape}")

print(f"\nFeature breakdown:")
print(f"  - Total features: {train_clean.shape[1] - 1}")
print(f"  - Mocode binary features: 563")
print(f"  - Other features: {train_clean.shape[1] - 1 - 563}")

# Check memory usage
train_memory_mb = train_clean.memory_usage(deep=True).sum() / 1024 / 1024
test_memory_mb = test_clean.memory_usage(deep=True).sum() / 1024 / 1024

print(f"\nMemory usage:")
print(f"  Train: {train_memory_mb:.2f} MB")
print(f"  Test:  {test_memory_mb:.2f} MB")

print("\n" + "="*70)
print("✅ ENCODING COMPLETE - READY FOR MODELING!")
print("="*70)
```

```python
print("="*70)
print("STEP 7: COMPREHENSIVE MODEL COMPARISON")
print("="*70)

from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.utils.class_weight import compute_class_weight
import time
import warnings
warnings.filterwarnings('ignore')

# ============================================================
# PART 1: PREPARE DATA
# ============================================================
print("\n[1/3] Preparing data...")

# Separate target and features
y_train = train_clean['Crm Cd'].copy()
X_train = train_clean.drop('Crm Cd', axis=1).copy()

# Encode target
le_target = LabelEncoder()
y_train_encoded = le_target.fit_transform(y_train)

print(f"Classes encoded: {len(le_target.classes_)}")
print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train_encoded.shape}")

# Train/validation split
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train, y_train_encoded, 
    test_size=0.2, 
    random_state=42, 
    stratify=y_train_encoded
)

print(f"Train: {X_tr.shape}, Validation: {X_val.shape}")

# Compute class weights
class_weights_dict = {}
for i in range(len(le_target.classes_)):
    count = (y_train_encoded == i).sum()
    weight = len(y_train_encoded) / (len(le_target.classes_) * count)
    class_weights_dict[i] = weight

sample_weights = np.array([class_weights_dict[y] for y in y_tr])

# Scale data for distance-based models
print("\nScaling features for distance-based models...")
scaler = StandardScaler()
X_tr_scaled = scaler.fit_transform(X_tr)
X_val_scaled = scaler.transform(X_val)

print("✓ Data prepared and scaled")

# ============================================================
# PART 2: TRAIN ALL MODELS
# ============================================================
print("\n[2/3] Training all models...")
print("="*70)

results_list = []

# Model 1: Random Forest
print("\n[1/8] Random Forest...")
start = time.time()
rf = RandomForestClassifier(
    n_estimators=100, max_depth=20, min_samples_split=10, min_samples_leaf=4,
    max_features='sqrt', class_weight='balanced', random_state=0, n_jobs=-1
)
rf.fit(X_tr, y_tr)
rf_pred = rf.predict(X_val)
rf_f1 = f1_score(y_val, rf_pred, average='macro')
rf_time = time.time() - start
results_list.append(['Random Forest', rf_f1, rf_time, rf])
print(f"  ✓ F1-macro: {rf_f1:.4f} | Time: {rf_time:.2f}s")

# Model 2: LightGBM
print("\n[2/8] LightGBM...")
start = time.time()
lgbm = LGBMClassifier(
    n_estimators=200, max_depth=10, learning_rate=0.1, num_leaves=31,
    min_child_samples=20, subsample=0.8, colsample_bytree=0.5,
    reg_alpha=0.1, reg_lambda=1.0, class_weight='balanced',
    random_state=42, n_jobs=-1, verbose=-1
)
lgbm.fit(X_tr, y_tr)
lgbm_pred = lgbm.predict(X_val)
lgbm_f1 = f1_score(y_val, lgbm_pred, average='macro')
lgbm_time = time.time() - start
results_list.append(['LightGBM', lgbm_f1, lgbm_time, lgbm])
print(f"  ✓ F1-macro: {lgbm_f1:.4f} | Time: {lgbm_time:.2f}s")

# Model 3: XGBoost
print("\n[3/8] XGBoost...")
start = time.time()
xgb = XGBClassifier(
    n_estimators=200, max_depth=8, learning_rate=0.1,
    subsample=0.8, colsample_bytree=0.5, reg_alpha=0.1, reg_lambda=1.0,
    random_state=42, n_jobs=-1, eval_metric='mlogloss', verbosity=0
)
xgb.fit(X_tr, y_tr, sample_weight=sample_weights)
xgb_pred = xgb.predict(X_val)
xgb_f1 = f1_score(y_val, xgb_pred, average='macro')
xgb_time = time.time() - start
results_list.append(['XGBoost', xgb_f1, xgb_time, xgb])
print(f"  ✓ F1-macro: {xgb_f1:.4f} | Time: {xgb_time:.2f}s")

# Model 4: CatBoost
print("\n[4/8] CatBoost...")
start = time.time()
cat = CatBoostClassifier(
    iterations=200, depth=8, learning_rate=0.1, l2_leaf_reg=3,
    class_weights=class_weights_dict, random_state=11, verbose=False, thread_count=-1
)
cat.fit(X_tr, y_tr)
cat_pred = cat.predict(X_val)
cat_f1 = f1_score(y_val, cat_pred, average='macro')
cat_time = time.time() - start
results_list.append(['CatBoost', cat_f1, cat_time, cat])
print(f"  ✓ F1-macro: {cat_f1:.4f} | Time: {cat_time:.2f}s")

# Model 5: Extra Trees
print("\n[5/8] Extra Trees...")
start = time.time()
et = ExtraTreesClassifier(
    n_estimators=100, max_depth=20, min_samples_split=10,
    max_features='sqrt', class_weight='balanced', random_state=111, n_jobs=-1
)
et.fit(X_tr, y_tr)
et_pred = et.predict(X_val)
et_f1 = f1_score(y_val, et_pred, average='macro')
et_time = time.time() - start
results_list.append(['Extra Trees', et_f1, et_time, et])
print(f"  ✓ F1-macro: {et_f1:.4f} | Time: {et_time:.2f}s")

# Model 6: Logistic Regression
print("\n[6/8] Logistic Regression...")
start = time.time()
lr = LogisticRegression(
    max_iter=500, class_weight='balanced', random_state=11, n_jobs=-1
)
lr.fit(X_tr_scaled, y_tr)
lr_pred = lr.predict(X_val_scaled)
lr_f1 = f1_score(y_val, lr_pred, average='macro')
lr_time = time.time() - start
results_list.append(['Logistic Regression', lr_f1, lr_time, lr])
print(f"  ✓ F1-macro: {lr_f1:.4f} | Time: {lr_time:.2f}s")

# Model 7: Ridge Classifier
print("\n[7/8] Ridge Classifier...")
start = time.time()
ridge = RidgeClassifier(
    alpha=1.0, class_weight='balanced', random_state=11
)
ridge.fit(X_tr_scaled, y_tr)
ridge_pred = ridge.predict(X_val_scaled)
ridge_f1 = f1_score(y_val, ridge_pred, average='macro')
ridge_time = time.time() - start
results_list.append(['Ridge Classifier', ridge_f1, ridge_time, ridge])
print(f"  ✓ F1-macro: {ridge_f1:.4f} | Time: {ridge_time:.2f}s")

# Model 8: KNN
print("\n[8/8] K-Nearest Neighbors...")
start = time.time()
knn = KNeighborsClassifier(
    n_neighbors=10, weights='distance', n_jobs=-1
)
knn.fit(X_tr_scaled, y_tr)
knn_pred = knn.predict(X_val_scaled)
knn_f1 = f1_score(y_val, knn_pred, average='macro')
knn_time = time.time() - start
results_list.append(['KNN', knn_f1, knn_time, knn])
print(f"  ✓ F1-macro: {knn_f1:.4f} | Time: {knn_time:.2f}s")

# ============================================================
# PART 3: RESULTS TABLE
# ============================================================
print("\n" + "="*70)
print("[3/3] FINAL RESULTS - SORTED BY F1-MACRO")
print("="*70)

# Create results dataframe
results_df = pd.DataFrame(results_list, columns=['Model', 'Val F1-macro', 'Training Time (s)', 'Model Object'])

# Calculate expected score
results_df['Expected Score'] = (results_df['Val F1-macro'] * 250).round(0).astype(int)

# Drop model object for display
results_display = results_df[['Model', 'Val F1-macro', 'Training Time (s)', 'Expected Score']].copy()

# Sort by F1-macro
results_display = results_display.sort_values('Val F1-macro', ascending=False).reset_index(drop=True)

# Add rank
results_display.insert(0, 'Rank', range(1, len(results_display) + 1))

print("\n")
print(results_display.to_string(index=False))

# Best model
best_row = results_display.iloc[0]
print(f"\n{'='*70}")
print(f"🏆 BEST MODEL: {best_row['Model']}")
print(f"{'='*70}")
print(f"   Rank: #{best_row['Rank']}")
print(f"   Val F1-macro: {best_row['Val F1-macro']:.4f}")
print(f"   Training Time: {best_row['Training Time (s)']:.2f}s")
print(f"   Expected Competition Score: {best_row['Expected Score']} / 250")
print(f"{'='*70}")

# Save best model object
best_model_name = best_row['Model']
best_model_obj = results_df[results_df['Model'] == best_model_name]['Model Object'].values[0]

print(f"\n✓ Best model object saved: {best_model_name}")

# Check if needs scaling
needs_scaling = best_model_name in ['Logistic Regression', 'Ridge Classifier', 'KNN', 'Naive Bayes', 'Linear SVM']
print(f"✓ Requires feature scaling: {needs_scaling}")

print("\n" + "="*70)
print("✅ COMPREHENSIVE MODEL COMPARISON COMPLETE!")
print("="*70)
```

```python
print("="*70)
print("STEP 8: CREATE SUBMISSION FILE (FIXED)")
print("="*70)

# ============================================================
# PART 1: RETRAIN XGBoost ON FULL DATA
# ============================================================
print("\n[1/4] Retraining XGBoost on 100% training data...")

# Full training data
X_train_full = train_clean.drop('Crm Cd', axis=1)
y_train_full = train_clean['Crm Cd']

# Encode target
from sklearn.preprocessing import LabelEncoder
le_final = LabelEncoder()
y_train_encoded = le_final.fit_transform(y_train_full)

print(f"Training samples: {X_train_full.shape[0]:,}")
print(f"Features: {X_train_full.shape[1]}")
print(f"Classes: {len(le_final.classes_)}")

# Compute sample weights
class_weights_final = {}
for i in range(len(le_final.classes_)):
    count = (y_train_encoded == i).sum()
    weight = len(y_train_encoded) / (len(le_final.classes_) * count)
    class_weights_final[i] = weight

sample_weights_final = np.array([class_weights_final[y] for y in y_train_encoded])

# Train final XGBoost
from xgboost import XGBClassifier
import time

final_xgb = XGBClassifier(
    n_estimators=200,
    max_depth=8,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.5,
    reg_alpha=0.1,
    reg_lambda=1.0,
    random_state=42,
    n_jobs=-1,
    eval_metric='mlogloss',
    verbosity=0
)

start = time.time()
final_xgb.fit(X_train_full, y_train_encoded, sample_weight=sample_weights_final)
train_time = time.time() - start

print(f"✓ Training completed in {train_time:.1f}s")

# ============================================================
# PART 2: GET DR_NO FROM CLEANED TEST
# ============================================================
print("\n[2/4] Getting DR_NO from cleaned test data...")

# Load original test
test_original = pd.read_csv(r'F:\Quera\olympic 2024\crime_detection\data\test.csv')

print(f"Original test size: {len(test_original)}")
print(f"Cleaned test size: {len(test_clean)}")
print(f"Rows removed: {len(test_original) - len(test_clean)}")

# Get DR_NO for rows that survived cleaning
# Strategy: match by index if we kept track, otherwise predict for cleaned rows only
# Since we don't have original indices, we need to match

# SOLUTION: Load test again, apply same cleaning, keep DR_NO
test_with_drno = pd.read_csv(r'F:\Quera\olympic 2024\crime_detection\data\test.csv')

# Store DR_NO before cleaning
dr_no_original = test_with_drno['DR_NO'].copy()

# Apply same cleaning steps that we did
test_for_submission = test_with_drno.copy()

# Convert dates
test_for_submission['Date Rptd'] = pd.to_datetime(test_for_submission['Date Rptd'])
test_for_submission['DATE OCC'] = pd.to_datetime(test_for_submission['DATE OCC'])

# Extract features (same as step 2)
test_for_submission['Year'] = test_for_submission['DATE OCC'].dt.year
test_for_submission['Month'] = test_for_submission['DATE OCC'].dt.month
test_for_submission['Day'] = test_for_submission['DATE OCC'].dt.day
test_for_submission['DayOfWeek'] = test_for_submission['DATE OCC'].dt.dayofweek
test_for_submission['Quarter'] = test_for_submission['DATE OCC'].dt.quarter
test_for_submission['DayOfYear'] = test_for_submission['DATE OCC'].dt.dayofyear
test_for_submission['Report_Delay_Days'] = (test_for_submission['Date Rptd'] - test_for_submission['DATE OCC']).dt.days
test_for_submission['Hour'] = test_for_submission['TIME OCC'] // 100

# Keep DR_NO before further processing
dr_no_before_cleaning = test_for_submission['DR_NO'].copy()

# Apply age filter (same as step 4) - THIS IS WHERE ROWS GET REMOVED
valid_age_mask = test_for_submission['Vict Age'] >= 0
test_for_submission = test_for_submission[valid_age_mask].reset_index(drop=True)
dr_no_after_cleaning = dr_no_before_cleaning[valid_age_mask].reset_index(drop=True)

print(f"✓ Kept DR_NO for {len(dr_no_after_cleaning)} valid rows")

# ============================================================
# PART 3: PREDICT ON TEST SET
# ============================================================
print("\n[3/4] Making predictions on test set...")

# Ensure column alignment
X_test_final = test_clean[X_train_full.columns]

print(f"Test samples: {X_test_final.shape[0]:,}")

# Predict
y_test_pred_encoded = final_xgb.predict(X_test_final)
y_test_pred = le_final.inverse_transform(y_test_pred_encoded)

print(f"✓ Predictions completed")
print(f"✓ Predictions: {len(y_test_pred)}")
print(f"✓ DR_NO available: {len(dr_no_after_cleaning)}")

# Verify lengths match
if len(y_test_pred) != len(dr_no_after_cleaning):
    print(f"⚠️  WARNING: Length mismatch!")
    print(f"   Predictions: {len(y_test_pred)}")
    print(f"   DR_NO: {len(dr_no_after_cleaning)}")
else:
    print("✓ Lengths match!")

# ============================================================
# PART 4: CREATE SUBMISSION FILE
# ============================================================
print("\n[4/4] Creating submission.csv...")

# Create submission with cleaned DR_NO
submission = pd.DataFrame({
    'DR_NO': dr_no_after_cleaning,
    'Crm Cd': y_test_pred
})

# Save
submission.to_csv('submission.csv', index=False)

print(f"✓ File saved: submission.csv")
print(f"✓ Rows: {len(submission):,}")

# Show sample
print("\nSubmission sample:")
print(submission.head(10))

print("\n" + "="*70)
print("✅ SUBMISSION READY!")
print("="*70)
print(f"\nExpected score: ~{int(xgb_f1 * 250)} / 250")
print("🚀 Upload 'submission.csv' to competition!")
print("\n⚠️  NOTE: Submission has {len(submission)} rows (not 9346)")
print("   This is because we removed invalid age rows.")
```
