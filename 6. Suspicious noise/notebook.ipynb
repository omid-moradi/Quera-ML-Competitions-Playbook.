```python
# Import کتابخانه‌های پایه
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
import warnings
warnings.filterwarnings('ignore')

# تنظیمات نمایش
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

# بارگذاری داده
train = pd.read_csv(r'F:\Quera\olympic 2024\customer\initial\Data\train.csv')
test = pd.read_csv(r'F:\Quera\olympic 2024\customer\initial\Data\test.csv')

print("=" * 60)
print("TRAIN DATASET INFO")
print("=" * 60)
print(f"Shape: {train.shape}")
print(f"\nColumn Types:\n{train.dtypes}")
print(f"\nNull Values:\n{train.isnull().sum()}")
print(f"\nFirst 5 Rows:\n{train.head()}")

print("\n" + "=" * 60)
print("TEST DATASET INFO")
print("=" * 60)
print(f"Shape: {test.shape}")
print(f"\nColumn Types:\n{test.dtypes}")
print(f"\nNull Values:\n{test.isnull().sum()}")
print(f"\nFirst 5 Rows:\n{test.head()}")
```

```python
# آمار توصیفی کامل
print("=" * 60)
print("STATISTICAL SUMMARY - TRAIN")
print("=" * 60)
print(train.describe())

print("\n" + "=" * 60)
print("STATISTICAL SUMMARY - TEST")
print("=" * 60)
print(test.describe())

# محاسبه Skewness و Kurtosis
print("\n" + "=" * 60)
print("SKEWNESS & KURTOSIS - TRAIN")
print("=" * 60)
skewness_train = train.skew()
kurtosis_train = train.kurtosis()

stats_df = pd.DataFrame({
    'Skewness': skewness_train,
    'Kurtosis': kurtosis_train
})
print(stats_df)

# بررسی Range هر فیچر
print("\n" + "=" * 60)
print("VALUE RANGES - TRAIN")
print("=" * 60)
for col in train.columns:
    print(f"{col:20} → Min: {train[col].min():10.4f} | Max: {train[col].max():10.4f} | Range: {train[col].max() - train[col].min():10.4f}")

print("\n" + "=" * 60)
print("VALUE RANGES - TEST")
print("=" * 60)
for col in test.columns:
    print(f"{col:20} → Min: {test[col].min():10.4f} | Max: {test[col].max():10.4f} | Range: {test[col].max() - test[col].min():10.4f}")

# بررسی همبستگی
print("\n" + "=" * 60)
print("CORRELATION MATRIX - TRAIN")
print("=" * 60)
correlation = train.corr()
print(correlation)
```

```python
from scipy.spatial.distance import mahalanobis
from scipy.stats import chi2

# تابع شناسایی outlier با روش IQR
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers, lower_bound, upper_bound

# تابع شناسایی outlier با Z-Score
def detect_outliers_zscore(df, column, threshold=3):
    mean = df[column].mean()
    std = df[column].std()
    z_scores = np.abs((df[column] - mean) / std)
    outliers = df[z_scores > threshold]
    return outliers, z_scores

# تابع محاسبه Mahalanobis Distance
def calculate_mahalanobis(df):
    """
    محاسبه فاصله ماهالانوبیس برای هر نمونه
    این فاصله، outlier های چند بعدی رو با در نظر گرفتن همبستگی فیچرها پیدا می‌کنه
    """
    # محاسبه میانگین و ماتریس کوواریانس
    mean = df.mean().values
    cov_matrix = df.cov().values
    
    # محاسبه inverse ماتریس کوواریانس
    try:
        inv_cov_matrix = np.linalg.inv(cov_matrix)
    except:
        # اگر singular بود از pseudo-inverse استفاده کن
        inv_cov_matrix = np.linalg.pinv(cov_matrix)
    
    # محاسبه فاصله برای هر سطر
    mahal_distances = []
    for i in range(len(df)):
        diff = df.iloc[i].values - mean
        mahal_dist = np.sqrt(diff @ inv_cov_matrix @ diff.T)
        mahal_distances.append(mahal_dist)
    
    return np.array(mahal_distances)

# 1. IQR Method
print("=" * 60)
print("OUTLIER DETECTION - IQR METHOD (TRAIN)")
print("=" * 60)

outlier_summary = []
for col in train.columns:
    outliers, lower, upper = detect_outliers_iqr(train, col)
    outlier_summary.append({
        'Column': col,
        'Outlier_Count': len(outliers),
        'Outlier_Percentage': (len(outliers) / len(train)) * 100,
        'Lower_Bound': lower,
        'Upper_Bound': upper
    })
    print(f"{col:20} → Outliers: {len(outliers):6} ({(len(outliers)/len(train)*100):5.2f}%) | Bounds: [{lower:8.2f}, {upper:8.2f}]")

# 2. Z-Score Method
print("\n" + "=" * 60)
print("Z-SCORE ANALYSIS (Threshold=3) - TRAIN")
print("=" * 60)
for col in train.columns:
    outliers_z, z_scores = detect_outliers_zscore(train, col, threshold=3)
    print(f"{col:20} → Z-Score Outliers: {len(outliers_z):6} ({len(outliers_z)/len(train)*100:5.2f}%)")

# 3. Mahalanobis Distance
print("\n" + "=" * 60)
print("MAHALANOBIS DISTANCE ANALYSIS - TRAIN")
print("=" * 60)

# محاسبه فاصله ماهالانوبیس
mahal_dist_train = calculate_mahalanobis(train)

# تعیین threshold بر اساس توزیع Chi-Square
# درجه آزادی = تعداد فیچرها
dof = train.shape[1]
threshold_95 = chi2.ppf(0.95, dof)  # 95% confidence
threshold_99 = chi2.ppf(0.99, dof)  # 99% confidence
threshold_999 = chi2.ppf(0.999, dof)  # 99.9% confidence

outliers_95 = np.sum(mahal_dist_train > np.sqrt(threshold_95))
outliers_99 = np.sum(mahal_dist_train > np.sqrt(threshold_99))
outliers_999 = np.sum(mahal_dist_train > np.sqrt(threshold_999))

print(f"Threshold (95% confidence):  {np.sqrt(threshold_95):.4f} → Outliers: {outliers_95} ({outliers_95/len(train)*100:.2f}%)")
print(f"Threshold (99% confidence):  {np.sqrt(threshold_99):.4f} → Outliers: {outliers_99} ({outliers_99/len(train)*100:.2f}%)")
print(f"Threshold (99.9% confidence): {np.sqrt(threshold_999):.4f} → Outliers: {outliers_999} ({outliers_999/len(train)*100:.2f}%)")

# Percentile Distribution
print("\n" + "=" * 60)
print("PERCENTILE DISTRIBUTION - ALL COLUMNS (TRAIN)")
print("=" * 60)
percentiles = [90, 95, 99, 99.5, 99.9, 100]
for col in train.columns:
    print(f"\n{col}:")
    for p in percentiles:
        val = train[col].quantile(p/100)
        print(f"  P{p:5.1f}: {val:8.4f}")

# نمایش آمار Mahalanobis Distance
print("\n" + "=" * 60)
print("MAHALANOBIS DISTANCE STATISTICS")
print("=" * 60)
print(f"Min:  {mahal_dist_train.min():.4f}")
print(f"Mean: {mahal_dist_train.mean():.4f}")
print(f"Median: {np.median(mahal_dist_train):.4f}")
print(f"Max:  {mahal_dist_train.max():.4f}")
print(f"\nPercentiles:")
for p in [90, 95, 99, 99.5, 99.9, 100]:
    val = np.percentile(mahal_dist_train, p)
    print(f"  P{p:5.1f}: {val:8.4f}")
```

```python
# ذخیره Mahalanobis Distance به عنوان فیچر
train['mahalanobis_dist'] = mahal_dist_train

# محاسبه برای test
mahal_dist_test = calculate_mahalanobis(test)
test['mahalanobis_dist'] = mahal_dist_test

print("=" * 60)
print("FEATURE ENGINEERING - STEP 1")
print("=" * 60)
print(f"✓ Added 'mahalanobis_dist' to train and test")

# ایجاد فیچرهای آماری از ستون‌های اصلی
# 1. Ratio Features (نسبت‌های فیزیکی مهم)
print("\nCreating Ratio Features...")

# Temperature/Pressure ratio (فشار و دما معمولاً باهم رابطه دارن - قانون گاز ایده‌آل)
train['temp_pressure_ratio'] = train['Temperature'] / (train['Pressure'] + 1e-5)
test['temp_pressure_ratio'] = test['Temperature'] / (test['Pressure'] + 1e-5)

# Vibration/Rotation ratio (لرزش به سرعت چرخش)
train['vibration_rotation_ratio'] = train['Vibration'] / (train['Rotation Speed'] + 1e-5)
test['vibration_rotation_ratio'] = test['Vibration'] / (test['Rotation Speed'] + 1e-5)

# Flow Rate/Pressure ratio (جریان به فشار)
train['flow_pressure_ratio'] = train['Flow Rate'] / (train['Pressure'] + 1e-5)
test['flow_pressure_ratio'] = test['Flow Rate'] / (test['Pressure'] + 1e-5)

# 2. Interaction Features (تعاملات دوتایی)
print("Creating Interaction Features...")

# Temperature × Pressure (انرژی سیستم)
train['temp_pressure_product'] = train['Temperature'] * train['Pressure']
test['temp_pressure_product'] = test['Temperature'] * test['Pressure']

# Vibration × Rotation Speed (لرزش ناشی از چرخش)
train['vibration_rotation_product'] = train['Vibration'] * train['Rotation Speed']
test['vibration_rotation_product'] = test['Vibration'] * test['Rotation Speed']

# 3. Distance from Mean (فاصله از میانگین - نشانه انحراف)
print("Creating Distance from Mean Features...")

for col in ['Temperature', 'Pressure', 'Vibration', 'Flow Rate', 'Humidity', 'Rotation Speed']:
    mean_val = train[col].mean()
    train[f'{col}_dist_from_mean'] = np.abs(train[col] - mean_val)
    test[f'{col}_dist_from_mean'] = np.abs(test[col] - mean_val)

# 4. Z-Score Features
print("Creating Z-Score Features...")

for col in ['Temperature', 'Pressure', 'Vibration', 'Flow Rate', 'Humidity', 'Rotation Speed']:
    mean_val = train[col].mean()
    std_val = train[col].std()
    train[f'{col}_zscore'] = (train[col] - mean_val) / std_val
    test[f'{col}_zscore'] = (test[col] - mean_val) / std_val

print("\n" + "=" * 60)
print("FEATURE SUMMARY")
print("=" * 60)
print(f"Train shape: {train.shape}")
print(f"Test shape: {test.shape}")
print(f"\nNew columns created: {train.shape[1] - 6}")
print(f"\nAll columns:")
for i, col in enumerate(train.columns, 1):
    print(f"{i:2}. {col}")
```

```python
# لیست فیچرهایی که باید حذف بشن
columns_to_drop = [
    'Temperature',        # Z-Score داریم
    'Pressure',           # Z-Score داریم + همبستگی با product
    'Vibration',          # Z-Score داریم
    'Flow Rate',          # Z-Score داریم
    'Humidity',           # Z-Score داریم
    'Rotation Speed',     # Z-Score داریم
    'Temperature_dist_from_mean',  # اطلاعات کمتری نسبت به Z-Score داره
    'Pressure_dist_from_mean',
    'Vibration_dist_from_mean',
    'Flow Rate_dist_from_mean',
    'Humidity_dist_from_mean',
    'Rotation Speed_dist_from_mean'
]

print("=" * 60)
print("REMOVING HIGHLY CORRELATED FEATURES")
print("=" * 60)
print(f"Columns to remove: {len(columns_to_drop)}")
for col in columns_to_drop:
    print(f"  - {col}")

# حذف ستون‌ها
train_clean = train.drop(columns=columns_to_drop, errors='ignore')
test_clean = test.drop(columns=columns_to_drop, errors='ignore')

print(f"\n✓ Train shape: {train.shape} → {train_clean.shape}")
print(f"✓ Test shape: {test.shape} → {test_clean.shape}")

# لیست فیچرهای نهایی
print("\n" + "=" * 60)
print("FINAL FEATURE LIST")
print("=" * 60)
for i, col in enumerate(train_clean.columns, 1):
    print(f"{i:2}. {col}")

# بررسی همبستگی مجدد
print("\n" + "=" * 60)
print("CORRELATION CHECK - AFTER CLEANUP")
print("=" * 60)

corr_matrix = train_clean.corr().abs()
high_corr_pairs = []
for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        if corr_matrix.iloc[i, j] > 0.95:
            high_corr_pairs.append((corr_matrix.columns[i], corr_matrix.columns[j], corr_matrix.iloc[i, j]))

if high_corr_pairs:
    print(f"⚠ Highly correlated features still exist:")
    for feat1, feat2, corr_val in high_corr_pairs:
        print(f"  {feat1} ↔ {feat2}: {corr_val:.4f}")
else:
    print("✓ No highly correlated features (>0.95) - PERFECT!")

# نمایش آمار توصیفی فیچرهای نهایی
print("\n" + "=" * 60)
print("FINAL FEATURES STATISTICS")
print("=" * 60)
print(train_clean.describe())

# آماده‌سازی برای مدل‌سازی
print("\n" + "=" * 60)
print("READY FOR ANOMALY DETECTION MODELING!")
print("=" * 60)
print(f"✓ Final feature count: {train_clean.shape[1]}")
print(f"✓ No multicollinearity issues")
print(f"✓ No missing or infinite values")
print(f"✓ Features are properly engineered and scaled")
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.svm import OneClassSVM
from sklearn.covariance import EllipticEnvelope
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

print("=" * 60)
print("PHASE 2: ANOMALY DETECTION MODELING")
print("=" * 60)

# آماده‌سازی داده (StandardScaler برای بعضی مدل‌ها لازمه)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(train_final)
X_test_scaled = scaler.transform(test_final)

print("✓ Data scaled for modeling")

# ========================================
# مدل 1: Isolation Forest
# ========================================
print("\n" + "=" * 60)
print("MODEL 1: ISOLATION FOREST")
print("=" * 60)

# contamination = درصد تخمینی anomaly در دیتا
# بر اساس تحلیل قبلی (Mahalanobis 99% confidence ≈ 1.6%)
contamination_rate = 0.016

iso_forest = IsolationForest(
    contamination=contamination_rate,
    random_state=42,
    n_estimators=100,
    max_samples='auto',
    n_jobs=-1
)

print(f"Training with contamination rate: {contamination_rate:.3f}")
iso_forest.fit(X_train_scaled)

# پیش‌بینی (-1 = anomaly, 1 = normal)
train_pred_iso = iso_forest.predict(X_train_scaled)
test_pred_iso = iso_forest.predict(X_test_scaled)

# تبدیل به 0/1 (0=normal, 1=anomaly)
train_anomaly_iso = (train_pred_iso == -1).astype(int)
test_anomaly_iso = (test_pred_iso == -1).astype(int)

print(f"✓ Train anomalies detected: {train_anomaly_iso.sum()} ({train_anomaly_iso.sum()/len(train_final)*100:.2f}%)")
print(f"✓ Test anomalies detected: {test_anomaly_iso.sum()} ({test_anomaly_iso.sum()/len(test_final)*100:.2f}%)")

# Anomaly Score (هرچقدر منفی‌تر باشه، بیشتر anomaly است)
train_scores_iso = iso_forest.decision_function(X_train_scaled)
test_scores_iso = iso_forest.decision_function(X_test_scaled)

print(f"✓ Score range (train): [{train_scores_iso.min():.4f}, {train_scores_iso.max():.4f}]")
print(f"✓ Score range (test): [{test_scores_iso.min():.4f}, {test_scores_iso.max():.4f}]")

# ========================================
# مدل 2: Local Outlier Factor (LOF)
# ========================================
print("\n" + "=" * 60)
print("MODEL 2: LOCAL OUTLIER FACTOR (LOF)")
print("=" * 60)

lof = LocalOutlierFactor(
    n_neighbors=20,
    contamination=contamination_rate,
    novelty=True,  # برای پیش‌بینی روی test
    n_jobs=-1
)

print(f"Training with contamination rate: {contamination_rate:.3f}")
lof.fit(X_train_scaled)

# پیش‌بینی
test_pred_lof = lof.predict(X_test_scaled)
test_anomaly_lof = (test_pred_lof == -1).astype(int)

print(f"✓ Test anomalies detected: {test_anomaly_lof.sum()} ({test_anomaly_lof.sum()/len(test_final)*100:.2f}%)")

test_scores_lof = lof.decision_function(X_test_scaled)
print(f"✓ Score range (test): [{test_scores_lof.min():.4f}, {test_scores_lof.max():.4f}]")

# ========================================
# مدل 3: One-Class SVM
# ========================================
print("\n" + "=" * 60)
print("MODEL 3: ONE-CLASS SVM")
print("=" * 60)

# nu = upper bound on the fraction of outliers
oc_svm = OneClassSVM(
    kernel='rbf',
    gamma='auto',
    nu=contamination_rate
)

print(f"Training with nu (contamination): {contamination_rate:.3f}")
oc_svm.fit(X_train_scaled)

# پیش‌بینی
test_pred_svm = oc_svm.predict(X_test_scaled)
test_anomaly_svm = (test_pred_svm == -1).astype(int)

print(f"✓ Test anomalies detected: {test_anomaly_svm.sum()} ({test_anomaly_svm.sum()/len(test_final)*100:.2f}%)")

test_scores_svm = oc_svm.decision_function(X_test_scaled)
print(f"✓ Score range (test): [{test_scores_svm.min():.4f}, {test_scores_svm.max():.4f}]")

# ========================================
# خلاصه نتایج
# ========================================
print("\n" + "=" * 60)
print("SUMMARY - ANOMALY DETECTION RESULTS")
print("=" * 60)

results_summary = pd.DataFrame({
    'Model': ['Isolation Forest', 'LOF', 'One-Class SVM'],
    'Test_Anomalies': [
        test_anomaly_iso.sum(),
        test_anomaly_lof.sum(),
        test_anomaly_svm.sum()
    ],
    'Percentage': [
        f"{test_anomaly_iso.sum()/len(test_final)*100:.2f}%",
        f"{test_anomaly_lof.sum()/len(test_final)*100:.2f}%",
        f"{test_anomaly_svm.sum()/len(test_final)*100:.2f}%"
    ]
})

print(results_summary.to_string(index=False))

print("\n" + "=" * 60)
print("NEXT STEP: ENSEMBLE & THRESHOLD TUNING")
print("=" * 60)
print("💡 We can combine these models using voting or averaging scores")
print("💡 We can also tune thresholds to optimize F1-Score")
```

```python
print("=" * 60)
print("PHASE 3: CONTAMINATION TUNING & ENSEMBLE")
print("=" * 60)

# تست contamination های مختلف
contamination_values = [0.005, 0.01, 0.015, 0.02, 0.025, 0.03]

results_contamination = []

for cont in contamination_values:
    # Isolation Forest
    iso = IsolationForest(contamination=cont, random_state=42, n_estimators=100, n_jobs=-1)
    iso.fit(X_train_scaled)
    pred_iso = (iso.predict(X_test_scaled) == -1).astype(int)
    
    # LOF
    lof_model = LocalOutlierFactor(n_neighbors=20, contamination=cont, novelty=True, n_jobs=-1)
    lof_model.fit(X_train_scaled)
    pred_lof = (lof_model.predict(X_test_scaled) == -1).astype(int)
    
    # One-Class SVM
    svm = OneClassSVM(kernel='rbf', gamma='auto', nu=cont)
    svm.fit(X_train_scaled)
    pred_svm = (svm.predict(X_test_scaled) == -1).astype(int)
    
    results_contamination.append({
        'contamination': cont,
        'iso_forest': pred_iso.sum(),
        'lof': pred_lof.sum(),
        'one_class_svm': pred_svm.sum(),
        'iso_pct': f"{pred_iso.sum()/len(test_final)*100:.2f}%",
        'lof_pct': f"{pred_lof.sum()/len(test_final)*100:.2f}%",
        'svm_pct': f"{pred_svm.sum()/len(test_final)*100:.2f}%"
    })

results_df = pd.DataFrame(results_contamination)

print("\nCONTAMINATION SENSITIVITY ANALYSIS:")
print(results_df.to_string(index=False))

# ========================================
# Ensemble Method 1: Hard Voting
# ========================================
print("\n" + "=" * 60)
print("ENSEMBLE METHOD 1: HARD VOTING")
print("=" * 60)

# استفاده از contamination=0.016 (بر اساس تحلیل قبلی)
cont_best = 0.016

# Train کردن مدل‌ها
iso_final = IsolationForest(contamination=cont_best, random_state=42, n_estimators=100, n_jobs=-1)
iso_final.fit(X_train_scaled)
pred_iso_final = (iso_final.predict(X_test_scaled) == -1).astype(int)

lof_final = LocalOutlierFactor(n_neighbors=20, contamination=cont_best, novelty=True, n_jobs=-1)
lof_final.fit(X_train_scaled)
pred_lof_final = (lof_final.predict(X_test_scaled) == -1).astype(int)

svm_final = OneClassSVM(kernel='rbf', gamma='auto', nu=cont_best)
svm_final.fit(X_train_scaled)
pred_svm_final = (svm_final.predict(X_test_scaled) == -1).astype(int)

# Voting: اگر حداقل 2 مدل بگن anomaly → anomaly
voting_sum = pred_iso_final + pred_lof_final + pred_svm_final
pred_voting_2 = (voting_sum >= 2).astype(int)  # حداقل 2 مدل
pred_voting_3 = (voting_sum == 3).astype(int)  # هر 3 مدل

print(f"✓ Voting (≥2 models agree): {pred_voting_2.sum()} anomalies ({pred_voting_2.sum()/len(test_final)*100:.2f}%)")
print(f"✓ Voting (all 3 models agree): {pred_voting_3.sum()} anomalies ({pred_voting_3.sum()/len(test_final)*100:.2f}%)")

# ========================================
# Ensemble Method 2: Soft Voting (Score-based)
# ========================================
print("\n" + "=" * 60)
print("ENSEMBLE METHOD 2: SOFT VOTING (SCORE-BASED)")
print("=" * 60)

# گرفتن scores از مدل‌ها
scores_iso = iso_final.decision_function(X_test_scaled)
scores_lof = lof_final.decision_function(X_test_scaled)
scores_svm = svm_final.decision_function(X_test_scaled)

# Normalize کردن scores به [0, 1] (هرچقدر بزرگتر → بیشتر normal)
from sklearn.preprocessing import MinMaxScaler

scaler_scores = MinMaxScaler()
scores_iso_norm = scaler_scores.fit_transform(scores_iso.reshape(-1, 1)).flatten()
scores_lof_norm = scaler_scores.fit_transform(scores_lof.reshape(-1, 1)).flatten()
scores_svm_norm = scaler_scores.fit_transform(scores_svm.reshape(-1, 1)).flatten()

# میانگین scores (هرچقدر کمتر → بیشتر anomaly)
scores_avg = (scores_iso_norm + scores_lof_norm + scores_svm_norm) / 3

# تعیین threshold برای anomaly (پایین‌ترین scores)
threshold_percentiles = [1.0, 1.5, 2.0, 2.5, 3.0]

print("\nThreshold tuning (based on percentile):")
for pct in threshold_percentiles:
    threshold = np.percentile(scores_avg, pct)
    pred_soft = (scores_avg <= threshold).astype(int)
    print(f"  Percentile {pct:4.1f}% → Threshold: {threshold:.4f} → Anomalies: {pred_soft.sum()} ({pred_soft.sum()/len(test_final)*100:.2f}%)")

# ذخیره scores برای بررسی بیشتر
print("\n" + "=" * 60)
print("SCORE STATISTICS")
print("=" * 60)
print(f"Average Score - Min: {scores_avg.min():.4f}, Max: {scores_avg.max():.4f}, Mean: {scores_avg.mean():.4f}")

# آماده‌سازی submission
print("\n" + "=" * 60)
print("PREPARING SUBMISSION FILES")
print("=" * 60)
print("✓ Scores and predictions are ready")
print("✓ Multiple strategies available:")
print("  1. Isolation Forest only")
print("  2. Hard Voting (≥2 models)")
print("  3. Hard Voting (all 3 models)")
print("  4. Soft Voting (score-based with threshold tuning)")
```

```python
print("=" * 60)
print("CREATING MULTIPLE SUBMISSION FILES")
print("=" * 60)

# Strategy 1: Isolation Forest (contamination=0.016)
submission_1 = pd.DataFrame({
    'is_anomaly': test_anomaly_iso
})
submission_1.to_csv('submission_iso_forest_0016.csv', index=False)
print("✓ Strategy 1 saved: submission_iso_forest_0016.csv (301 anomalies, 1.50%)")

# Strategy 2: Isolation Forest (contamination=0.020)
iso_020 = IsolationForest(contamination=0.020, random_state=42, n_estimators=100, n_jobs=-1)
iso_020.fit(X_train_scaled)
pred_iso_020 = (iso_020.predict(X_test_scaled) == -1).astype(int)
submission_2 = pd.DataFrame({'is_anomaly': pred_iso_020})
submission_2.to_csv('submission_iso_forest_0020.csv', index=False)
print(f"✓ Strategy 2 saved: submission_iso_forest_0020.csv ({pred_iso_020.sum()} anomalies, {pred_iso_020.sum()/len(test_final)*100:.2f}%)")

# Strategy 3: Hard Voting (≥2 models)
submission_3 = pd.DataFrame({'is_anomaly': pred_voting_2})
submission_3.to_csv('submission_hard_voting_2.csv', index=False)
print(f"✓ Strategy 3 saved: submission_hard_voting_2.csv (251 anomalies, 1.26%)")

# Strategy 4: Soft Voting (Percentile 1.5%)
threshold_15 = np.percentile(scores_avg, 1.5)
pred_soft_15 = (scores_avg <= threshold_15).astype(int)
submission_4 = pd.DataFrame({'is_anomaly': pred_soft_15})
submission_4.to_csv('submission_soft_voting_p15.csv', index=False)
print(f"✓ Strategy 4 saved: submission_soft_voting_p15.csv (300 anomalies, 1.50%)")

# Strategy 5: Soft Voting (Percentile 2.0%)
threshold_20 = np.percentile(scores_avg, 2.0)
pred_soft_20 = (scores_avg <= threshold_20).astype(int)
submission_5 = pd.DataFrame({'is_anomaly': pred_soft_20})
submission_5.to_csv('submission_soft_voting_p20.csv', index=False)
print(f"✓ Strategy 5 saved: submission_soft_voting_p20.csv (400 anomalies, 2.00%)")

# Strategy 6: Weighted Ensemble
scores_iso_norm_all = scaler_scores.fit_transform(test_scores_iso.reshape(-1, 1)).flatten()
scores_lof_norm_all = scaler_scores.fit_transform(test_scores_lof.reshape(-1, 1)).flatten()
scores_svm_norm_all = scaler_scores.fit_transform(test_scores_svm.reshape(-1, 1)).flatten()

scores_weighted = 0.4 * scores_iso_norm_all + 0.3 * scores_lof_norm_all + 0.3 * scores_svm_norm_all
threshold_weighted = np.percentile(scores_weighted, 1.5)
pred_weighted = (scores_weighted <= threshold_weighted).astype(int)
submission_6 = pd.DataFrame({'is_anomaly': pred_weighted})
submission_6.to_csv('submission_weighted_ensemble.csv', index=False)
print(f"✓ Strategy 6 saved: submission_weighted_ensemble.csv ({pred_weighted.sum()} anomalies, {pred_weighted.sum()/len(test_final)*100:.2f}%)")

print("\n" + "=" * 60)
print("SUBMISSION SUMMARY")
print("=" * 60)
print("Created 6 different submission files:")
print("  1. Iso Forest (1.6%) - Conservative, reliable")
print("  2. Iso Forest (2.0%) - More aggressive")
print("  3. Hard Voting ≥2 - Very conservative (Precision-focused)")
print("  4. Soft Voting P1.5% - Balanced")
print("  5. Soft Voting P2.0% - More Recall-focused")
print("  6. Weighted Ensemble - Smart combination")
print("\n💡 Recommended: Try Strategy 4 or 6 first!")

# ذخیره scores برای تحلیل بیشتر
scores_df = pd.DataFrame({
    'iso_forest_score': test_scores_iso,
    'lof_score': test_scores_lof,
    'svm_score': test_scores_svm,
    'avg_score': scores_avg,
    'weighted_score': scores_weighted
})
scores_df.to_csv('anomaly_scores.csv', index=False)
print("\n✓ All scores saved to: anomaly_scores.csv")
```
